# 📄 Understanding CSV Files

A **CSV (Comma-Separated Values)** file stores tabular data.

Each row represents one record, and each value is separated by a comma.

Example:

```text
Name,Age,Course
Rahul,21,BCA
Aman,20,B.Tech
Nikita,22,CSE
```

CSV files are simple, lightweight, and supported by almost every spreadsheet and database application.

## 💡 Why Use CSV?

CSV files are useful because:

- They are easy to create.
- They are easy to read.
- They can be opened in spreadsheet software like Microsoft Excel and Google Sheets.
- They are commonly used for importing and exporting data.
- They are widely used in Data Analysis and Machine Learning.

# 📝 Creating and Writing a CSV File

Python provides a built-in module called **`csv`** for working with CSV (Comma-Separated Values) files.

Instead of manually adding commas between values, the `csv` module automatically formats the data correctly.

Before working with CSV files, we need to import the `csv` module.

## 📦 Importing the `csv` Module

````python
import csv
````

### Explanation

The `csv` module is part of Python's **Standard Library**, so no additional installation is required.

After importing it, we can create, write, and read CSV files easily.

## 📂 File Path

````python
FILE_PATH = "../datasets/students.csv"
````

We'll store all CSV examples in the **datasets/** folder.

The file **students.csv** does not exist yet. Python will create it for us when we open it in **Write (`w`) Mode**.

## 📝 Writing Data Using `csv.writer()`

The `csv.writer()` function creates a **writer object**.

This object provides methods to write rows into a CSV file.

The most commonly used methods are:

- `writerow()` → Writes a single row.
- `writerows()` → Writes multiple rows.

## 📝 Syntax

````python
with open(FILE_PATH, "w", newline="") as file:
    writer = csv.writer(file)
````

### Understanding the Syntax

- `open()` opens the CSV file.
- `"w"` opens the file in **Write Mode**.
- `newline=""` prevents extra blank lines in CSV files (especially on Windows).
- `csv.writer()` creates a writer object.

> **💡 Why do we use `newline=""`?**
>
> When writing CSV files, especially on Windows, omitting `newline=""` may result in an extra blank line between each row.
>
> Using `newline=""` ensures that each row is written correctly without unwanted empty lines.

## 💻 Example 1: Creating a CSV File

````python
import csv

FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "w", newline="") as file:
    writer = csv.writer(file)

print("CSV file created successfully!")
````

### Explanation

In this example:

1. The `csv` module was imported.
2. `students.csv` was opened in **Write Mode**.
3. Since the file didn't exist, Python created it automatically.
4. A CSV writer object was created.
5. No data has been written yet, so the file is currently empty.

## 💻 Example 2: Writing the Header Row

````python
import csv

FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Age", "Course"])

print("Header row written successfully!")
````

### Explanation

The `writerow()` method writes a single row into the CSV file.

The list:

```python
["Name", "Age", "Course"]
```

becomes:

```text
Name,Age,Course
```

inside the file.

This first row is called the **header row**, because it describes the columns in the CSV file.

## 💻 Example 3: Writing Student Records

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Name", "Age", "Course"])
    writer.writerow(["Rahul", 21, "BCA"])
    writer.writerow(["Aman", 20, "B.Tech"])
    writer.writerow(["Nikita", 22, "CSE"])

print("Student records written successfully!")
````

### Explanation

Each call to `writer.writerow()` writes one complete row into the CSV file.

After running the code, **students.csv** contains:

```text
Name,Age,Course
Rahul,21,BCA
Aman,20,B.Tech
Nikita,22,CSE
```

Notice that the `csv` module automatically separates the values with commas.

This structured format makes it easy to import the data into spreadsheet software or databases.

## 💡 Key Points

- Import the `csv` module before working with CSV files.
- Use `csv.writer()` to create a writer object.
- Use `writerow()` to write a single row.
- The first row is usually the **header row**.
- `newline=""` prevents extra blank lines in CSV files.

# 📖 Reading Data from a CSV File

Once data has been stored in a CSV file, we often need to retrieve it for processing, analysis, or display.

Python provides the **`csv.reader()`** function to read data from a CSV file.

Each row in the CSV file is returned as a **list**.

## 📦 `csv.reader()`

The `csv.reader()` function creates a **reader object**.

This object reads the CSV file one row at a time.

Each row is returned as a Python list.

## 💻 Example 1: Reading the Entire CSV File

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "r", newline="") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
````

### Explanation

Let's understand what happened:

- `csv.reader()` read the file row by row.
- Each row was returned as a **Python list**.
- The `for` loop printed one row at a time.

Notice that even numeric values such as `21` and `20` are returned as **strings** because CSV files store data as text.

## 💻 Example 2: Reading Only the Data Rows

Sometimes we don't want to process the header row.

We can skip it using the `next()` function.

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "r", newline="") as file:
    reader = csv.reader(file)

    next(reader)

    for row in reader:
        print(row)
````

### Explanation

The first row of the CSV file contains the column names.

```text
Name,Age,Course
```

The `next(reader)` statement reads this row once and moves the reader to the next row.

As a result, the loop processes only the student records.

## 💻 Example 3: Accessing Individual Columns

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "r", newline="") as file:
    reader = csv.reader(file)

    next(reader)

    for row in reader:
        print(f"Name   : {row[0]}")
        print(f"Age    : {row[1]}")
        print(f"Course : {row[2]}")
        print("-" * 20)
````

### Explanation

Each row is a Python list.

For example:

```python
['Rahul', '21', 'BCA']
```

We can access individual values using indexing:

- `row[0]` → Name
- `row[1]` → Age
- `row[2]` → Course

This works exactly like accessing elements in any Python list.

## 🌍 Real-World Example

Imagine a school stores student records in a CSV file.

The application reads each row, extracts the student's name, age, and course, and displays them on the screen.

This is one of the most common ways CSV files are processed in Python applications.

## 💡 Key Points

- `csv.reader()` reads CSV files row by row.
- Each row is returned as a Python list.
- Use `next(reader)` to skip the header row.
- List indexing can be used to access individual columns.
- CSV values are read as strings by default.

# 📖 Reading CSV Files Using `DictReader`

So far, we've used `csv.reader()`, where each row was returned as a Python list.

Sometimes, working with column indexes like `row[0]` and `row[1]` makes the code harder to read.

Python provides **`csv.DictReader()`**, which reads each row as a dictionary.

The column names from the header row become the dictionary keys.

## 📦 `csv.DictReader()`

`DictReader()` automatically uses the first row of the CSV file as the dictionary keys.

Each remaining row is returned as a dictionary.

## 💻 Example 1: Reading CSV as Dictionaries

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
````

### Explanation

Instead of returning a list, each row is returned as a dictionary.

For example:

```python
{
    "Name": "Rahul",
    "Age": "21",
    "Course": "BCA"
}
```

This makes the code more readable because we can access values using column names.

## 💻 Example 2: Accessing Values by Column Name

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "r", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(f"Name   : {row['Name']}")
        print(f"Age    : {row['Age']}")
        print(f"Course : {row['Course']}")
        print("-" * 20)
````

### Explanation

Instead of using:

```python
row[0]
row[1]
row[2]
```

we directly use:

```python
row["Name"]
row["Age"]
row["Course"]
```

This makes the program easier to understand and maintain.

# ✍️ Writing CSV Files Using `DictWriter`

`DictWriter()` is used when we want to write dictionaries into a CSV file.

Instead of providing a list of values, we provide a dictionary where the keys match the column names.

## 💻 Example 1: Writing Data Using `DictWriter`

````python
FILE_PATH = "../datasets/students.csv"

with open(FILE_PATH, "w", newline="") as file:
    fieldnames = ["Name", "Age", "Course"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow(
        {
            "Name": "Riya",
            "Age": 20,
            "Course": "B.Sc"
        }
    )

print("CSV written using DictWriter successfully!")
````

### Explanation

In this example:

- `fieldnames` defines the column names.
- `writeheader()` writes the header row.
- `writerow()` writes one dictionary as one row.

The CSV file now contains:

```text
Name,Age,Course
Riya,20,B.Sc
```

## 🌍 Real-World Example

Many web applications receive user information as dictionaries.

Using `DictWriter()`, this data can be directly stored in a CSV file without manually arranging the values in the correct order.

This makes the code cleaner and less error-prone.

## 💡 Key Points

- `DictReader()` returns each row as a dictionary.
- Column names become dictionary keys.
- `DictWriter()` writes dictionaries into a CSV file.
- `writeheader()` creates the header row.
- `DictReader()` is often more readable than `csv.reader()`.

# 📦 Working with JSON Files in Python

## 🤔 What is JSON?

JSON stands for **JavaScript Object Notation**.

It is a lightweight and widely used format for storing and exchanging structured data.

Although JSON originated from JavaScript, it is supported by almost every modern programming language, including Python.

JSON is commonly used in APIs, web applications, cloud services, mobile applications, and configuration files.

## 💡 Why Do We Use JSON?

JSON makes it easy to represent structured data using **key-value pairs**.

Unlike CSV, JSON can also represent nested objects and lists, making it suitable for storing complex data.

It is one of the most common formats used to exchange information between applications.

## 🌍 Real-World Applications

JSON is widely used in:

- 🌐 REST APIs
- 📱 Mobile Applications
- ☁️ Cloud Services
- 🤖 Machine Learning Projects
- ⚙️ Configuration Files
- 🛒 E-commerce Platforms
- 🏦 Banking Applications
- 📊 Data Exchange Between Systems

# 📦 Importing the `json` Module

````python
import json
````

### Explanation

The `json` module is included in Python's Standard Library.

No additional installation is required.

It provides functions to convert Python objects into JSON and JSON data back into Python objects.

## 📂 File Path

````python
FILE_PATH = "../datasets/students.json"
````

# ✍️ Writing Data into a JSON File

Python provides the **`json.dump()`** function to write Python objects into a JSON file.

The data is automatically converted into JSON format before being saved.

## 💻 Example 1: Creating a JSON File

````python
FILE_PATH = "../datasets/students.json"

student = {
    "Name": "Nikita",
    "Age": 22,
    "Course": "CSE"
}

with open(FILE_PATH, "w") as file:
    json.dump(student, file)

print("JSON file created successfully!")
````

### Explanation

In this example:

1. A Python dictionary was created.
2. `students.json` was opened in **Write Mode**.
3. `json.dump()` converted the dictionary into JSON format.
4. The JSON data was saved into the file.

The file now contains:

```json
{"Name": "Nikita", "Age": 22, "Course": "CSE"}
```

## 💡 Making JSON More Readable

By default, JSON is written on a single line.

We can use the **`indent`** parameter to format the JSON file and make it easier to read.

## 💻 Example 1: Writing Formatted JSON

````python
FILE_PATH = "../datasets/students.json"

student = {
    "Name": "Nikita",
    "Age": 22,
    "Course": "CSE"
}

with open(FILE_PATH, "w") as file:
    json.dump(student, file, indent=4)

print("Formatted JSON written successfully!")
````

### Explanation

Using `indent=4` formats the JSON file with proper indentation.

The file now looks like this:

```json
{
    "Name": "Nikita",
    "Age": 22,
    "Course": "CSE"
}
```

This format is much easier for humans to read and edit.

## 🌍 Real-World Example

Many web APIs return data in JSON format.

Python programs read this JSON data, process it, and display the required information to users.

JSON is one of the most widely used formats for communication between applications.

## 💡 Key Points

- Import the `json` module before working with JSON files.
- `json.dump()` writes Python objects into a JSON file.
- Dictionaries are automatically converted into JSON objects.
- Use `indent` to make JSON files more readable.
- JSON is commonly used for storing and exchanging structured data.

# 📖 Reading Data from a JSON File

Once data has been stored in a JSON file, we can read it back into our Python program.

Python provides the **`json.load()`** function to read JSON data from a file and convert it into the corresponding Python object.

For example:

- A JSON object becomes a Python dictionary.
- A JSON array becomes a Python list.

## 📦 `json.load()`

The `json.load()` function reads data from a JSON file and converts it into a Python object.

After loading the data, we can work with it just like any normal Python dictionary or list.

## 💻 Example 1: Reading a JSON File

````python

FILE_PATH = "../datasets/students.json"

with open(FILE_PATH, "r") as file:
    student = json.load(file)

print(student)
````

### Explanation

The JSON file was read successfully.

The JSON object was automatically converted into a Python dictionary.

Now we can use dictionary operations on the loaded data.

## 💻 Example 2: Checking the Data Type

````python
FILE_PATH = "../datasets/students.json"

with open(FILE_PATH, "r") as file:
    student = json.load(file)

print(type(student))
````

### Explanation

Although the data was stored in a JSON file, after reading it with `json.load()`, it becomes a **Python dictionary**.

This means we can use all dictionary methods and operations on it.

## 💻 Example 3: Accessing JSON Values

````python
FILE_PATH = "../datasets/students.json"

with open(FILE_PATH, "r") as file:
    student = json.load(file)

print(f"Name   : {student['Name']}")
print(f"Age    : {student['Age']}")
print(f"Course : {student['Course']}")
````

### Explanation

Since the loaded JSON data is a Python dictionary, we can access values using their keys.

For example:

- `student["Name"]`
- `student["Age"]`
- `student["Course"]`

This works exactly like accessing values in any other Python dictionary.

# 📚 Working with Multiple Records

In real-world applications, we usually store multiple records instead of just one.

To do this, we use a **list of dictionaries**.

Each dictionary represents one record.

## 💻 Example 1: Writing Multiple Student Records

````python
FILE_PATH = "../datasets/students.json"

students = [
    {"Name": "Rahul", "Age": 21, "Course": "BCA"},
    {"Name": "Aman", "Age": 20, "Course": "B.Tech"},
    {"Name": "Nikita", "Age": 22, "Course": "CSE"}
]

with open(FILE_PATH, "w") as file:
    json.dump(students, file, indent=4)

print("Multiple student records written successfully!")
````

### Explanation

Here, the data being written is a **list** containing three dictionaries.

When saved, the JSON file contains an array of student objects.

This structure is commonly used in APIs, web applications, and data exchange.

## 💻 Example 2: Reading Multiple Student Records

````python
FILE_PATH = "../datasets/students.json"

with open(FILE_PATH, "r") as file:
    students = json.load(file)

for student in students:
    print(student)
````

### Explanation

The JSON file now contains a list of dictionaries.

After reading it with `json.load()`, the result is a Python list.

Each item in the list is a dictionary representing one student record.

We use a `for` loop to process each student one by one.

## 🌍 Real-World Example

Most REST APIs return a list of JSON objects.

For example, an online shopping application may return a list of products, where each product is represented as a JSON object.

Python applications read this data using `json.load()` (or `response.json()` when working with APIs) and process each object individually.

## 💡 Key Points

- `json.load()` reads JSON data from a file.
- JSON objects become Python dictionaries.
- JSON arrays become Python lists.
- Dictionary keys are used to access values.
- Lists of dictionaries are commonly used to store multiple records.

# ⚖️ CSV vs JSON

Both CSV and JSON are popular formats for storing and exchanging data, but they are designed for different use cases.

CSV is best suited for simple tabular data, while JSON is more suitable for complex and nested data structures.

## 📊 Comparison Table

| Feature | CSV | JSON |
|---------|-----|------|
| Full Form | Comma-Separated Values | JavaScript Object Notation |
| Data Structure | Rows and Columns | Key-Value Pairs |
| Supports Nested Data | ❌ No | ✅ Yes |
| Human Readability | Easy | Easy (with indentation) |
| File Size | Usually Smaller | Usually Larger |
| Common Use Cases | Excel, Reports, Datasets | APIs, Configuration Files, Web Applications |
| Python Module | `csv` | `json` |

# 📌 Quick Reference Table

| Function / Method | Purpose |
|-------------------|---------|
| `csv.writer()` | Create a CSV writer object |
| `writer.writerow()` | Write a single row |
| `writer.writerows()` | Write multiple rows |
| `csv.reader()` | Read CSV rows as lists |
| `csv.DictReader()` | Read CSV rows as dictionaries |
| `csv.DictWriter()` | Write dictionaries to CSV |
| `writeheader()` | Write the CSV header row |
| `json.dump()` | Write Python objects to a JSON file |
| `json.load()` | Read JSON data from a file |
| `indent=4` | Format JSON for better readability |

# ❌ Common Mistakes

When working with CSV and JSON files, beginners often make the following mistakes:

### CSV

- Forgetting to import the `csv` module.
- Not using `newline=""` while writing CSV files.
- Assuming numeric values are automatically read as integers.
- Forgetting to skip the header row when required.

### JSON

- Forgetting to import the `json` module.
- Using `json.load()` instead of `json.loads()` (or vice versa).
- Using invalid JSON syntax (such as single quotes or trailing commas).
- Forgetting to use `indent` when readability is important.

# ✅ Best Practices

Follow these best practices while working with CSV and JSON files:

- Use the `with` statement to automatically close files.
- Store files inside a dedicated `datasets/` folder.
- Use descriptive file names.
- Validate the file contents after writing.
- Use `indent=4` for readable JSON files.
- Keep headers meaningful in CSV files.
- Handle exceptions when working with important files (covered in the Exception Handling module).
