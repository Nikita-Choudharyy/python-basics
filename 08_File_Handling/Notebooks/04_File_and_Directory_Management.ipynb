# 📦 The `os` Module

The **`os`** module provides functions for interacting with the operating system.

It allows us to work with files, directories, and paths in a platform-independent way.

Using the `os` module, we can:

- Navigate directories.
- Create folders.
- Rename files.
- Delete files.
- Check whether files or folders exist.

## 📥 Importing the `os` Module

````python
import os
````

### Explanation

The `os` module is part of Python's Standard Library.

No installation is required.

After importing it, we can perform many operating system-related tasks directly from Python.

# 📍 Current Working Directory (CWD)

Whenever a Python program runs, it has a **Current Working Directory (CWD)**.

This is the default folder where Python looks for files if no absolute path is provided.

Knowing the current working directory is useful when working with relative paths and project files.

## 💻 Example : Display the Current Working Directory

````python
current_directory = os.getcwd()

print("Current Working Directory:")
print(current_directory)
````

### Explanation

- `os.getcwd()` returns the current working directory as a string.
- This helps you verify where your Python program is currently running.
- Relative file paths are resolved based on this directory.

# 📂 Navigating Directories

Python allows us to move from one directory to another using the `os.chdir()` function.

Changing the current working directory is useful when files or folders are located in different locations.

After changing the directory, all relative paths are resolved from the new current working directory.

## 📦 `os.chdir()`

The `os.chdir()` function changes the current working directory to the specified path.

## 💻 Example : Changing the Current Working Directory

````python
print("Before:")
print(os.getcwd())

# Example path (replace with a valid folder on your computer)
os.chdir("..")

print("\nAfter:")
print(os.getcwd())
````

### Explanation

In this example:

- `os.getcwd()` displays the current directory.
- `os.chdir("..")` moves one level up to the parent directory.
- Calling `os.getcwd()` again confirms that the directory has changed.

> **💡 Understanding `".."`**
>
> The symbol `".."` refers to the **parent directory** (one level above the current folder).
>
> Example:
>
> ```text
> Project/
> ├── notebooks/
> └── datasets/
> ```
>
> If your current folder is `notebooks/`, then `os.chdir("..")` moves you to `Project/`.

# 📋 Listing Files and Directories

Once we know our current location, we often need to see what files and folders are available.

Python provides the `os.listdir()` function for this purpose.

## 📦 `os.listdir()`

`os.listdir()` returns a list containing the names of files and directories inside a specified folder.

If no path is provided, it lists the contents of the current working directory.

## 💻 Example : Listing the Current Directory

````python
items = os.listdir()

print(items)
````

### Explanation

The `os.listdir()` function returned a Python list containing all files and folders in the current working directory.

Each item in the list represents either:

- A file
- A directory (folder)

At this stage, `os.listdir()` does **not** tell us which items are files and which are directories. We'll learn how to check that later using `os.path.isfile()` and `os.path.isdir()`.

## 💻 Example : Listing a Specific Directory

````python
items = os.listdir("Datasets")

for item in items:
    print(item)
````

### Explanation

Instead of listing the current directory, we provided a specific path:

```python
"Datasets"
```

Python displayed all files inside the `Datasets` folder.

This approach is useful when you want to inspect a folder without changing the current working directory.

## 🌍 Real-World Example

Imagine you're building a photo management application.

Before processing images, your program first lists all files inside the `images/` folder using `os.listdir()`. It then processes each image one by one.

Similarly, data analysis projects often list all CSV files inside a dataset folder before loading them.

## 💡 Key Points

- `os.chdir()` changes the current working directory.
- `os.getcwd()` displays the current working directory.
- `os.listdir()` returns a list of files and folders.
- If no path is provided, `os.listdir()` uses the current working directory.
- `os.listdir()` returns only names, not file details.

# 📁 Creating Directories

In many applications, we need to create folders to organize files.

For example:

- Creating a folder for uploaded files.
- Creating a project directory.
- Creating folders for logs or backups.
- Organizing datasets into separate folders.

Python provides the `os.mkdir()` and `os.makedirs()` functions to create directories.

## 📦 `os.mkdir()`

The `os.mkdir()` function creates **a single directory**.

If the directory already exists, Python raises a `FileExistsError`.

## 💻 Example : Creating a Single Directory

````python
DIRECTORY_NAME = "Demo_Folder"

os.mkdir(DIRECTORY_NAME)

print("Directory created successfully!")
````

### Explanation

In this example:

- A new folder named **Demo_Folder** is created.
- The folder is created inside the current working directory.
- If the folder already exists, Python raises a `FileExistsError`.

> **⚠️ Important**
>
> Run this example **only once**.
>
> If you run it again without deleting the folder, you'll get:
>
> ```text
> FileExistsError
> ```

## 📦 `os.makedirs()`

Sometimes we need to create multiple folders at once.

Python provides `os.makedirs()` for creating **nested directories**.

## 💻 Example : Creating Nested Directories

````python
DIRECTORY_PATH = "Projects/Python/File_Handling"

os.makedirs(DIRECTORY_PATH)

print("Nested directories created successfully!")
````

### Explanation

Python creates the following folder structure automatically:

```text
Projects/
└── Python/
    └── File_Handling/
```

Even if **Projects** and **Python** don't already exist, `os.makedirs()` creates them automatically.

## ⚖️ `mkdir()` vs `makedirs()`

| Feature | `os.mkdir()` | `os.makedirs()` |
|---------|--------------|-----------------|
| Creates One Folder | ✅ Yes | ✅ Yes |
| Creates Nested Folders | ❌ No | ✅ Yes |
| Creates Missing Parent Folders | ❌ No | ✅ Yes |
| Common Use | Single folder | Folder hierarchy |

## 🌍 Real-World Example

Imagine you're creating a Data Science project.

Instead of manually creating folders like:

```text
Project/
├── data/
├── notebooks/
├── models/
└── reports/
```

you can automate the process using Python.

This saves time and ensures every project follows the same folder structure.

## 💡 Key Points

- Use `os.mkdir()` to create a single directory.
- Use `os.makedirs()` to create nested directories.
- `os.makedirs()` automatically creates missing parent folders.
- Running these examples multiple times without deleting the folders will raise `FileExistsError`.

# ✏️ Renaming Files and Directories

Sometimes we need to change the name of a file or directory.

Python provides the `os.rename()` function for this purpose.

It can rename both files and directories.

## 📦 `os.rename()`

## 💻 Example : Renaming a Directory

````python
os.rename("Demo_Folder", "Practice_Folder")

print("Directory renamed successfully!")
````

### Explanation

The directory named **Demo_Folder** is renamed to **Practice_Folder**.

Only the name changes—the contents of the directory remain unchanged.

# 🗑️ Deleting Files

Python provides the `os.remove()` function to delete files.

Once a file is deleted using `os.remove()`, it cannot be recovered from Python directly.

## 📦 `os.remove()`

## 💻 Example : Deleting a File

````python
import os

os.remove("sample.txt")

print("File deleted successfully!")
````

### Explanation

This example deletes the file named `sample.txt`.

If the file does not exist, Python raises a `FileNotFoundError`.

# 📁 Deleting Empty Directories

Python provides the `os.rmdir()` function to remove an **empty directory**.

If the directory contains files or subdirectories, Python raises an error.

## 💻 Example : Removing an Empty Directory

````python
os.rmdir("Practice_Folder")

print("Directory removed successfully!")
````

### Explanation

`Practice_Folder` is removed because it is empty.

If the directory contains any files or folders, Python raises an `OSError`.

# 📂 Removing Nested Empty Directories

When multiple nested directories are empty, Python provides the `os.removedirs()` function to remove them all.

## 💻 Example : Removing Nested Directories

````python
os.removedirs("Projects/Python/File_Handling")

print("Nested directories removed successfully!")
````

### Explanation

If the following structure exists and all folders are empty:

```text
Projects/
└── Python/
    └── File_Handling/
```

then `os.removedirs()` removes:

- `File_Handling`
- `Python`
- `Projects` (if it also becomes empty)

If any parent directory contains other files or folders, Python stops removing directories at that point.

## 🌍 Real-World Example

Automation scripts often create temporary folders during execution.

After the task is completed, these folders can be removed automatically using `os.rmdir()` or `os.removedirs()`, helping keep the system clean and organized.

## 💡 Key Points

- `os.rename()` renames files and directories.
- `os.remove()` deletes files.
- `os.rmdir()` removes only empty directories.
- `os.removedirs()` removes nested empty directories.
- Always verify the path before deleting files or folders.

# ✅ Checking Files and Directories

Before performing operations such as reading, deleting, or renaming a file, it is a good practice to check whether the file or directory exists.

Python provides several useful functions in the `os.path` module for this purpose.

# 📦 `os.path.exists()`

The `os.path.exists()` function checks whether a given file or directory exists.

It returns:

- `True` if the path exists.
- `False` if the path does not exist.

## 💻 Example : Checking Whether a File Exists

````python
FILE_PATH = "Datasets/students.csv"

print(os.path.exists(FILE_PATH))
````

### Explanation

Since `students.csv` exists inside the `Datasets` folder, Python returns `True`.

If the file didn't exist, the result would be `False`.

## 💻 Example : Checking Whether a Directory Exists

````python
print(os.path.exists("Datasets"))
````

### Explanation

The `Datasets` folder exists, so Python returns `True`.

# 📄 `os.path.isfile()`

Sometimes we need to know whether a path points specifically to a **file**.

For this purpose, Python provides the `os.path.isfile()` function.

## 💻 Example : Checking Whether a Path is a File

````python
print(os.path.isfile("Datasets/students.csv"))
````

### Explanation

The given path points to a file.

Therefore, `os.path.isfile()` returns `True`.

# 📁 `os.path.isdir()`

The `os.path.isdir()` function checks whether a given path points to a directory.

## 💻 Example : Checking Whether a Path is a Directory

````python
print(os.path.isdir("Datasets"))
````

### Explanation

Since `Datasets` is a folder, `os.path.isdir()` returns `True`.

If the path referred to a file instead, the function would return `False`.

## 🌍 Real-World Example

Many applications first check whether a file exists before attempting to open it.

For example, a data analysis script may verify that `students.csv` is available before loading it.

This helps prevent runtime errors and makes programs more reliable.

## 💡 Key Points

- `os.path.exists()` checks whether a path exists.
- `os.path.isfile()` checks whether the path is a file.
- `os.path.isdir()` checks whether the path is a directory.
- These functions help make programs safer and more reliable.

# 🌟 Introduction to the `pathlib` Module

The `os` module has been used in Python for many years and is still widely used.

However, Python introduced the **`pathlib`** module to provide a more modern and object-oriented way of working with file system paths.

`pathlib` makes path handling more readable and easier to understand.

## 🤔 Why Use `pathlib`?

Some advantages of `pathlib` include:

- Cleaner and more readable code.
- Easy path manipulation.
- Platform-independent path handling.
- Object-oriented design.
- Recommended for modern Python projects.

# 📦 Importing `pathlib`

````python
from pathlib import Path
````

### Explanation

The `Path` class is the main component of the `pathlib` module.

It allows us to represent file and directory paths as Python objects instead of plain strings.

## 💻 Example : Creating a Path Object

````python
path = Path("Datasets/students.csv")

print(path)
````

### Explanation

`Path()` creates a path object representing the file path.

Unlike the `os` module, we don't work with plain strings—we work with `Path` objects, which provide many useful methods for file and directory operations.
