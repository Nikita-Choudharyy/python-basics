## 🤔 What is File Handling?

File Handling is the process of creating, opening, reading, writing, updating, and managing files using a programming language.

Instead of storing data only in memory (RAM), Python allows us to store information permanently inside files on a storage device such as an SSD or Hard Drive.

This makes it possible to reuse the same data whenever the program is executed again.

## 💡 Why Do We Need File Handling?

Imagine you are building a Student Management System.

You store student names in a Python list.

```python
students = ["Rahul", "Aman", "Nikita"]
```

Everything works perfectly while the program is running.

However, once the program is closed, the list disappears because it exists only in memory.

To store student records permanently, we need File Handling.

Using files, the data remains available even after restarting the computer or running the program again.

## 📂 Types of Files

Files are mainly divided into two categories.

### 📄 Text Files

Text files store information in a human-readable format.

Examples:

- `.txt`
- `.csv`
- `.json`
- `.xml`
- `.html`
- `.py`



### 💾 Binary Files

Binary files store data in binary format (0s and 1s).

They cannot be read directly by humans.

Examples:

- `.jpg`
- `.png`
- `.pdf`
- `.mp3`
- `.mp4`
- `.zip`
- `.exe`

## 📊 Text File vs Binary File

| Feature | Text File | Binary File |
|----------|-----------|-------------|
| Data Format | Human-readable | Binary (0s and 1s) |
| Can Humans Read It? | ✅ Yes | ❌ No |
| Common Extensions | `.txt`, `.csv`, `.json` | `.jpg`, `.png`, `.pdf` |
| Storage | Characters | Bytes |
| Examples | Notes, CSV, JSON | Images, Videos, PDFs |

# 📂 Opening a File in Python

Before performing any operation on a file, Python must first open it.

Python provides the built-in **`open()`** function to open a file.

When a file is opened, Python creates a **File Object** that allows us to interact with that file.

### General Syntax

```python
open(file_path, mode)
```

Where:

- **file_path** → Location of the file.
- **mode** → Specifies how the file will be used (read, write, append, etc.).

## 💻 Example 1: Opening a File

````python
FILE_PATH = "../datasets/students.txt"

file = open(FILE_PATH, "w")

print(file)

file.close()
````

### Explanation

- `open()` opens the file located at `FILE_PATH`.
- The `"w"` mode tells Python to open the file in **Write Mode**.
- The returned value is stored inside the variable **file**.
- This returned object is called a **File Object**.
- Finally, we close the file using `close()`.

## 📦 What is a File Object?

When Python opens a file, it creates a special object called a **File Object**.

This object acts as a connection between your Python program and the actual file stored on your computer.

All file operations such as reading, writing, and closing are performed using this File Object.

## 💻 Example 2: Checking the Type of File Object

````python
FILE_PATH = "../datasets/students.txt"

file = open(FILE_PATH, "w")

print(type(file))

file.close()
````

### Explanation

The output shows that `file` is not a string or a list.

It is an object of type **TextIOWrapper**, which is Python's built-in class for working with text files.

# 🔒 Closing a File

After finishing work with a file, it should always be closed.

Python provides the **`close()`** method to close a file.

Closing a file releases the system resources associated with it.

## 💻 Example 3: Closing a File

````python
FILE_PATH = "../datasets/students.txt"

file = open(FILE_PATH, "w")

file.close()

print("File closed successfully!")
````

### Explanation

The `close()` method ends the connection between the program and the file.

After calling `close()`, no further operations should be performed using that File Object unless the file is opened again.

## ❓ Why is Closing a File Important?

Closing a file is considered a good programming practice.

Some benefits of closing a file are:

- ✅ Frees system resources.
- ✅ Prevents data corruption.
- ✅ Ensures changes are properly saved.
- ✅ Avoids unexpected errors.

# 🤝 The `with` Statement

Python provides the **`with` statement** as a safer way to work with files.

When using `with`, Python automatically closes the file after the block of code finishes executing.

This reduces the chance of forgetting to call `close()`.

## 💻 Example 4: Opening a File Using `with`

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "w") as file:
    print(type(file))

print("File closed automatically.")
````

### Explanation

- `with` opens the file.
- `as file` stores the File Object in the variable `file`.
- When the `with` block ends, Python automatically closes the file.
- This is the recommended way to work with files in Python.

# 📚 Overview of File Modes

A **file mode** tells Python how you want to use a file.

Some commonly used file modes are:

| Mode | Description |
|------|-------------|
| `r` | Read an existing file |
| `w` | Write to a file (creates a new file if it doesn't exist) |
| `a` | Append data to the end of a file |
| `x` | Create a new file only if it doesn't already exist |
| `r+` | Read and write |
| `w+` | Write and read (overwrites existing content) |
| `a+` | Append and read |

## ⚠️ Common Mistakes

Beginners often make these mistakes while learning File Handling:

- ❌ Forgetting to close a file after opening it.
- ❌ Using the wrong file mode (`r`, `w`, `a`, etc.).
- ❌ Assuming `open()` reads or writes data automatically.
- ❌ Confusing a **File Object** with the actual file.
- ❌ Forgetting to use the correct file path.
- ❌ Using `"w"` mode without realizing it can overwrite an existing file.
