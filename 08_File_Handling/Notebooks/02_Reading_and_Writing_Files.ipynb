# ✍️ Writing Data into a File

Since our `students.txt` file is empty, the first step is to store some data inside it.

Python provides the **`write()`** method to write text into a file.

The `write()` method writes the given string exactly as it is into the file.

## 💻 Example 1: Writing Data into an Empty File

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "w") as file:
    file.write("Nikita Choudhary")

print("Data written successfully!")
````

### Explanation

Let's understand what happened in this example:

1. We stored the location of the file in the `FILE_PATH` variable.
2. The file was opened in **Write (`w`) Mode**.
3. Since `students.txt` was empty, Python wrote the text **"Nikita Choudhary"** into it.
4. After leaving the `with` block, Python automatically closed the file.

Now, if you open **students.txt**, you'll see:

```text
Nikita Choudhary
```

## 🌍 Real-World Example

Imagine you're developing a **Student Management System**.

When a new student registers, their name can be saved to a text file.

Later, the application can read this file to display all registered students.

This simple concept forms the foundation of many real-world software systems.

## 💡 Key Points

- `write()` writes a string into a file.
- `"w"` creates the file if it doesn't exist.
- If the file already contains data, `"w"` replaces the existing content.
- The `with` statement automatically closes the file after use.

## ✍️ Writing Multiple Lines into a File

In the previous example, we wrote only a single line into the file.

In real-world applications, files usually contain multiple lines of data, such as student names, employee records, or log entries.

To store data on separate lines, we use the newline character (`\n`).

## 💡 What is `\n`?

`\n` is called the **newline character**.

It tells Python to move the cursor to the beginning of the next line.

Without `\n`, everything would be written on the same line.

## 💻 Example 1: Writing Multiple Student Names

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "w") as file:
    file.write("Rahul\n")
    file.write("Aman\n")
    file.write("Nikita\n")
    file.write("Priya")

print("Student records saved successfully!")
````

### Explanation

In this example:

1. The file was opened in **Write (`w`) Mode**.
2. Four student names were written into the file.
3. `\n` was used after the first three names to move each new name to the next line.
4. The last name does not require `\n` because no additional line follows it.

The file now contains:

```text
Rahul
Aman
Nikita
Priya
```

## ⚠️ Important: What Happens if We Use `"w"` Again?

The **Write (`w`) Mode** always starts writing from the beginning of the file.

If the file already contains data, the old content is removed before the new content is written.

This process is called **overwriting**.

## 💻 Example 2: Overwriting Existing Data

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "w") as file:
    file.write("Riya")

print("Previous data has been replaced.")
````

### Explanation

Before running this code, the file contained:

```text
Rahul
Aman
Nikita
Priya
```

After running the code, the file contains only:

```text
Riya
```

This happened because **`"w"` mode overwrites the existing content** before writing new data.

## 🌍 Real-World Example

Imagine a school updates its student list every year.

If the old list is no longer needed, the application can open the file in **Write (`w`) Mode** and save the new list.

The previous records are removed, and the latest records become the new contents of the file.

## 💡 Key Points

- `write()` writes text into a file.
- `\n` moves the cursor to the next line.
- `"w"` mode starts writing from the beginning of the file.
- Existing content is removed when `"w"` mode is used.
- Use `"w"` only when you want to create a new file or replace its contents.

# 📖 Reading Data from a File

Before reading data from a file, the file should contain some information.

In the previous section, we learned how to write data into a file.

Now, we'll first prepare our file with sample student records and then learn different ways to read that data.

## 📝 Preparing the File

The following code writes sample student records into **students.txt**.

We'll use this same file throughout the reading examples.

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "w") as file:
    file.write("Rahul\n")
    file.write("Aman\n")
    file.write("Nikita\n")
    file.write("Priya")

print("Sample data prepared successfully!")
````

### Explanation

This code prepares the file for our reading examples.

The file now contains:

```text
Rahul
Aman
Nikita
Priya
```

All the following examples will use this data.

## 📖 Reading the Entire File

Python provides the **`read()`** method to read the contents of a file.

When called without any arguments, `read()` returns the **entire content** of the file as a single string.

## 💻 Example 1: Reading the Complete File

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    content = file.read()

print(content)
````

### Explanation

Let's understand what happened:

1. The file was opened in **Read (`r`) Mode**.
2. `read()` read the complete contents of the file.
3. The returned data was stored in the variable `content`.
4. Finally, `print(content)` displayed all the student names.

Notice that the line breaks are preserved because the file contains newline characters (`\n`).

## 🌍 Real-World Example

Imagine you're building a **Student Management System**.

When the application starts, it reads the student file and displays all registered students on the screen.

This is one of the most common uses of the `read()` method.

## 📖 Reading a Specific Number of Characters

Sometimes we don't want to read the entire file.

The `read()` method accepts an integer argument that specifies how many characters should be read.

### Syntax

```python
file.read(number_of_characters)
```

For example:

- `file.read(5)` → Reads the first **5 characters**.
- `file.read(10)` → Reads the first **10 characters**.

## 💻 Example 1: Reading the First 5 Characters

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    content = file.read(5)

print(content)
````

### Explanation

The first five characters in the file are:

```text
Rahul
```

Since `"Rahul"` contains exactly **5 characters**, `read(5)` returns only that portion of the file.

If you specify a larger number, Python will continue reading additional characters until the requested limit is reached or the file ends.

## 💡 Key Points

- `read()` reads the complete file.
- `read(number)` reads only the specified number of characters.
- The returned value is always a string.
- Reading does not modify the contents of the file.

# 📄 Reading One Line at a Time

Sometimes we don't need to read the entire file at once.

Instead, we may want to process one line at a time.

Python provides the **`readline()`** method for this purpose.

## 💻 Example 1: Reading the First Line

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    first_line = file.readline()

print(first_line)
````

### Explanation

The file contains:

```text
Rahul
Aman
Nikita
Priya
```

`readline()` reads only the **first line**, so the output is:

```text
Rahul
```

The file pointer automatically moves to the beginning of the second line.

## 💻 Example 2: Reading Multiple Lines Using `readline()`

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    print(file.readline(), end="")
    print(file.readline(), end="")
    print(file.readline(), end="")
````

### Explanation

Each call to `readline()` reads the **next available line**.

- First call → Rahul
- Second call → Aman
- Third call → Nikita

Notice the use of `end=""`.

This prevents `print()` from adding an extra blank line because `readline()` already includes the newline character (`\n`).

# 📚 Reading All Lines as a List

Sometimes we want to read all lines at once while keeping them separated.

Python provides the **`readlines()`** method for this.

Instead of returning a single string, `readlines()` returns a **list of strings**.

## 💻 Example 1: Reading All Lines

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    lines = file.readlines()

print(lines)
````

### Explanation

The output is a **list**.

Each element in the list represents one line from the file.

Notice that:

- `"Rahul\n"` contains a newline character.
- `"Aman\n"` contains a newline character.
- `"Nikita\n"` contains a newline character.
- `"Priya"` is the last line, so it does not end with `\n`.

## 💻 Example 2: Printing Each Line from `readlines()`

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    lines = file.readlines()

for line in lines:
    print(line, end="")
````

### Explanation

Instead of printing the entire list, we loop through it using a `for` loop.

This prints each student name on a separate line, making the output much cleaner and easier to read.

## 📊 Difference Between `read()`, `readline()`, and `readlines()`

| Method | Returns | Best Used For |
|---------|---------|---------------|
| `read()` | Entire file as a single string | Reading the complete file |
| `readline()` | One line as a string | Reading one line at a time |
| `readlines()` | List of all lines | Processing each line individually |

## 🌍 Real-World Example

Imagine you're processing a log file that contains thousands of entries.

Using `readline()`, you can process one log entry at a time without loading the entire file into memory.

If you need all the lines together for further processing, `readlines()` provides them as a list.

## 💡 Key Points

- `readline()` reads one line at a time.
- Every call to `readline()` moves to the next line.
- `readlines()` returns a list of all lines.
- Use `readline()` for sequential processing.
- Use `readlines()` when you need all lines as a list.

# ➕ Appending Data to a File

Until now, we used **Write Mode (`w`)** to store data in a file.

However, Write Mode removes the existing content before writing new data.

Sometimes we want to keep the existing data and simply add new information at the end of the file.

For this purpose, Python provides **Append Mode (`a`)**.

## 💻 Example 1: Appending a New Student

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "a") as file:
    file.write("\nRiya")

print("New student added successfully!")
````

### Explanation

Before running this code, the file contained:

```text
Rahul
Aman
Nikita
Priya
```

After running the code, the file becomes:

```text
Rahul
Aman
Nikita
Priya
Riya
```

Notice that no existing data was removed.

The new student was added at the end of the file.

## 🌍 Real-World Example

Imagine a school where new students join throughout the year.

Instead of replacing the entire student list, the application simply appends each new student's record to the end of the file.

This is a common use case for **Append Mode**.

# 📍 Understanding the File Pointer

Whenever a file is opened, Python keeps track of the current reading or writing position.

This position is called the **File Pointer**.

As data is read or written, the file pointer automatically moves forward.

## 📍 The `tell()` Method

The **`tell()`** method returns the current position of the file pointer.

The position is measured in bytes from the beginning of the file.

## 💻 Example 1: Checking the Initial File Pointer Position

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    print(file.tell())
````

### Explanation

The output is `0` because the file pointer starts at the beginning of the file.

No data has been read yet.

## 💻 Example 2: Checking the Pointer After Reading

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    print(file.read(5))
    print(file.tell())
````

### Explanation

- `read(5)` reads the first **5 characters** (`Rahul`).
- After reading these characters, the file pointer moves forward.
- `tell()` now returns **5**, indicating the current position.

# 🔄 Moving the File Pointer with `seek()`

The **`seek()`** method moves the file pointer to a specified position.

This allows you to reread data or jump to another location within the file.

## 💻 Example 1: Moving Back to the Beginning

````python
FILE_PATH = "../datasets/students.txt"

with open(FILE_PATH, "r") as file:
    print(file.read(5))

    file.seek(0)

    print(file.read(5))
````

### Explanation

Here's what happened:

1. `read(5)` read the first **5 characters** (`Rahul`).
2. The file pointer moved to position **5**.
3. `seek(0)` moved the file pointer back to the beginning of the file.
4. Calling `read(5)` again read **Rahul** once more.

Without `seek(0)`, Python would continue reading from the current position instead of starting over.

## 📊 Difference Between `tell()` and `seek()`

| Method | Purpose |
|---------|---------|
| `tell()` | Returns the current position of the file pointer |
| `seek(position)` | Moves the file pointer to the specified position |

## 🌍 Real-World Example

Suppose you're reading a large log file.

You may need to:

- Check how much of the file has been processed using `tell()`.
- Return to the beginning of the file using `seek(0)` if you want to process it again.

These methods are especially useful when working with large files or custom file-processing programs.

## 💡 Key Points

- `"a"` mode adds data to the end of a file.
- Existing content is not removed in Append Mode.
- `tell()` returns the current file pointer position.
- `seek()` moves the file pointer to a new position.
- The file pointer changes automatically as data is read or written.

## ⚠️ Common Mistakes

Beginners often make the following mistakes while reading and writing files:

- ❌ Opening a file in the wrong mode (`"r"` instead of `"w"` or `"a"`).
- ❌ Forgetting that `"w"` mode overwrites the existing content.
- ❌ Trying to read from an empty file and expecting data.
- ❌ Forgetting to use the correct file path.
- ❌ Using `read()` and expecting the file pointer to return to the beginning automatically.
- ❌ Forgetting to use `seek()` before reading the same content again.

## 💡 Best Practices

Follow these best practices while working with files:

- ✅ Prefer the `with` statement instead of manually calling `close()`.
- ✅ Store file paths in variables such as `FILE_PATH`.
- ✅ Use `"w"` only when replacing the existing content is intended.
- ✅ Use `"a"` when you want to preserve existing data.
- ✅ Read files only after ensuring that they contain data.
- ✅ Organize all practice files inside the `datasets/` folder.
