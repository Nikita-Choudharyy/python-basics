# 🏷️ `__name__` in Python

# 📖 Introduction

As Python programs grow larger, code is often divided into multiple modules and packages to improve organization and reusability.

Sometimes, a Python file is executed directly, while other times it is imported into another program.

Python needs a way to identify **how a file is being used**.

This is where the special built-in variable **`__name__`** becomes important.

The value of `__name__` changes depending on whether the file is:

- ▶️ Executed directly by the user.
- 📦 Imported into another Python program.

Understanding `__name__` is essential for writing reusable modules, testing code, avoiding unwanted execution during imports, and building professional Python applications.

One of the most common Python statements,

```python
if __name__ == "__main__":
```

is based entirely on this concept.

By the end of this notebook, you will understand what `__name__` is, how Python assigns its value, and why this concept is used in almost every professional Python project.

# 🤔 Why Do We Need `__name__`?

Imagine you create a Python file that contains:

- Functions
- Classes
- Test code
- Example programs

Now suppose another file imports your module.

If Python executes **every line** inside the imported file automatically, it could produce unwanted output or run test code that wasn't intended to execute.

To solve this problem, Python provides the special variable **`__name__`**.

Using `__name__`, a program can determine:

- ▶️ Whether the file is being executed directly.
- 📦 Whether the file has been imported into another program.

This allows developers to keep reusable code and testing code inside the same file while ensuring that only the desired code runs.

# 🌍 Real-World Analogy

Imagine a teacher writes a mathematics workbook.

The workbook contains:

- 📖 Lessons
- 📝 Practice Questions
- ✅ Answer Key

Now consider two situations.

### Situation 1

A student opens the workbook to study.

The student reads the lessons, solves the exercises, and checks the answers.

Everything in the workbook is used.

---

### Situation 2

Another teacher only wants to copy a few practice questions into a new worksheet.

The teacher does **not** want the answer key or the entire workbook.

Only the required content is taken.

---

Python works in a similar way.

When a file is **executed directly**, the entire program runs.

When the same file is **imported**, Python should make its functions and classes available without automatically running demonstration or testing code.

The special variable **`__name__`** helps Python distinguish between these two situations, making code more reusable, organized, and professional.

# 📌 What is `__name__`?

`__name__` is a **special built-in variable** provided automatically by Python.

Every Python file (module) has its own `__name__` variable.

Python assigns a value to this variable when the file is executed.

The value of `__name__` depends on **how the file is being used**.

- ▶️ If the file is executed directly, `__name__` becomes `"__main__"`.
- 📦 If the file is imported into another file, `__name__` becomes the **module name**.

This allows Python programs to identify whether a file is running as the main program or being used as a reusable module.



## 📌 Definition

> **`__name__` is a special built-in variable that stores the name of the current module being executed.**

# 🧠 Understanding `__main__`

`__main__` is a special string used by Python.

Whenever a Python file is executed directly,

Python automatically assigns

```python
__name__ = "__main__"
```

This tells Python that the current file is the **main program**.

However,

if the same file is imported into another Python file,

Python assigns the module's filename instead of `"__main__"`.

For example,

Suppose we have a file named

```text
calculator.py
```

If we execute it directly,

```text
python calculator.py
```

then

```python
__name__
```

becomes

```python
"__main__"
```

But if another file imports it,

```python
import calculator
```

then

```python
__name__
```

becomes

```python
"calculator"
```

```text

 │         │
 └────┬────┘
      ▼

```

This decision happens automatically before the program begins execution.

# 🎯 Why Does Python Do This?

Python needs to know whether a file is:

- ▶️ The starting point of the program.
- 📦 A reusable module imported by another program.

This information allows developers to control which parts of the code should execute automatically.

Without this distinction,

every imported file would execute all of its demonstration or testing code whenever it was imported.

This would make large Python applications difficult to manage.

The special variable `__name__` solves this problem.

# 💡 Did You Know?

When you import a module,

Python executes the module **only once** during the current program execution.

If the same module is imported again,

Python uses the cached version instead of executing the file a second time.

This behavior is called **Module Caching**, and it improves the performance of Python programs.

# 🎯 `if __name__ == "__main__":`

One of the most commonly used statements in Python is:

```python
if __name__ == "__main__":
```

You will find this statement in many Python programs, libraries, and open-source projects.

It is used to determine whether a Python file is being:

- ▶️ Executed directly
- 📦 Imported as a module

The code inside this block runs **only when the file is executed directly**.

If the file is imported into another program, the code inside this block is skipped.

This helps developers write reusable modules while keeping testing or demonstration code separate.

# 📝 Syntax

```python
if __name__ == "__main__":
    # Code to execute when the file is run directly
```

The condition checks whether the special variable `__name__` is equal to `"__main__"`.

If the condition is **True**, the code inside the block executes.

Otherwise, Python skips the block.

````python
# calculator.py

def add(a, b):
    return a + b


if __name__ == "__main__":
    print("calculator.py is running directly.")
    print(add(10, 20))
````

### ▶️ Run the following file

```text
python calculator.py
```

# 📝 Explanation

Here,

we executed

```text
calculator.py
```

directly.

Therefore,

Python assigned

```python
__name__ = "__main__"
```

The condition

```python
if __name__ == "__main__":
```

became **True**.

As a result,

Python executed the code inside the `if` block.

````python
# main.py

import calculator

print("main.py is running.")
````

### ▶️ Run the following file

```text
python main.py
```

# 📝 Explanation

In this example,

`calculator.py` was imported into `main.py`.

Since the file was **imported**,

Python assigned

```python
__name__ = "calculator"
```

The condition

```python
if __name__ == "__main__":
```

became **False**.

Therefore,

the code inside the `if` block did **not** execute.

Only the import was completed, and the remaining code in `main.py` continued to run.

# ⚠️ Common Mistakes

While learning `__name__`, beginners often make the following mistakes.

## ❌ Mistake 1: Forgetting the Double Underscores

Incorrect

```python
if name == "__main__":
```

or

```python
if _name_ == "__main__":
```

Correct

```python
if __name__ == "__main__":
```

Both `__name__` and `__main__` contain **double underscores (`__`)**.

---

## ❌ Mistake 2: Incorrect Comparison

Incorrect

```python
if __name__ == "__main":
```

Correct

```python
if __name__ == "__main__":
```

Always compare with the exact string `"__main__"`.

---

## ❌ Mistake 3: Expecting the Block to Execute After Importing

Suppose

```python
import calculator
```

If `calculator.py` contains

```python
if __name__ == "__main__":
    print("Hello")
```

The message **will not be printed** because the file is being imported.

---

## ❌ Mistake 4: Writing Important Functions Inside the Block

Incorrect

```python
if __name__ == "__main__":

    def add(a, b):
        return a + b
```

Functions should normally be defined **outside** the `if` block.

The `if` block should contain only execution code such as testing or demonstrations.

---

## ❌ Mistake 5: Misunderstanding the Purpose

The statement

```python
if __name__ == "__main__":
```

does **not** create a function.

It simply checks whether the current file is running as the main program.

# ✅ Best Practices

Follow these best practices while using `__name__`.

### ♻️ Keep Reusable Code Outside the Block

### 🧪 Use the Block for Testing

---

to separate reusable code from executable code.
