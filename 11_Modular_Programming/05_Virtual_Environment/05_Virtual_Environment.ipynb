# 🐍 Virtual Environment in Python

# 📖 Introduction

As Python projects become larger, they often require multiple external libraries and packages.

Different projects may depend on different versions of the same package. For example, one project might require **NumPy 1.26**, while another requires **NumPy 2.x**.

If all packages are installed globally on the system, version conflicts can occur, making projects difficult to manage.

To solve this problem, Python provides **Virtual Environments**.

A Virtual Environment is an isolated Python environment that contains its own Python interpreter, installed packages, and dependencies. Each project can have its own independent environment without affecting other projects or the system-wide Python installation.

Using Virtual Environments is considered a standard practice in professional Python development. They help developers keep projects organized, reproducible, and free from dependency conflicts.

By the end of this notebook, you will learn how to create, activate, use, and manage Virtual Environments for your Python projects.

# 🤔 Why Do We Need Virtual Environments?

Imagine you are working on multiple Python projects.

### Project A

Requires:

- NumPy 1.26
- Pandas 2.1

---

### Project B

Requires:

- NumPy 2.x
- TensorFlow
- Scikit-learn

If all these packages are installed globally, updating one package for a project may accidentally break another project that depends on an older version.

This problem is known as a **dependency conflict**.

Virtual Environments solve this issue by giving each project its own isolated Python environment.

This means:

- 📦 Each project has its own installed packages.
- 🔄 Different projects can use different package versions.
- 🚫 Changes in one project do not affect another.
- 🧹 The system Python installation remains clean.
- 🚀 Projects become easier to share and reproduce.

For these reasons, Virtual Environments are considered an essential part of professional Python development.

# 📌 What is a Virtual Environment?

A **Virtual Environment** is an isolated Python environment created for a specific project.

It contains its own:

- 🐍 Python Interpreter
- 📦 Installed Packages
- 📚 Libraries
- ⚙️ Project Dependencies

This means that every project can have its own independent Python environment without affecting other projects on the same computer.

Each Virtual Environment works separately, allowing different projects to use different package versions safely.



## 📌 Definition

> **A Virtual Environment is an isolated Python environment that contains its own interpreter and installed packages for a specific project.**

# 🧠 How Virtual Environments Work

Normally, when we install a Python package,

```bash
pip install numpy
```

the package is installed into the **global Python environment**.

Every Python project on the computer shares these globally installed packages.

This can lead to version conflicts when different projects require different versions of the same package.

A Virtual Environment solves this problem.

Instead of installing packages globally,

Python creates a separate environment for each project.

Packages installed inside one Virtual Environment remain available **only within that environment**.

Other projects remain completely unaffected.

# 📊 Global Environment vs Virtual Environment

```text
                    Global Python

            ┌───────────────────────┐
            │ Python Installation    │
            │ NumPy                  │
            │ Pandas                 │
            │ Matplotlib             │
            └───────────────────────┘
                     ▲
          Every project shares these packages
```

---

```text
          Virtual Environments

Project A                 Project B

┌───────────────┐      ┌───────────────┐
│ Python        │      │ Python        │
│ NumPy 1.26    │      │ NumPy 2.x     │
│ Pandas        │      │ TensorFlow    │
└───────────────┘      └───────────────┘

Independent from each other
```

Each project has its own packages and dependencies.

# 📂 Structure of a Virtual Environment

After creating a Virtual Environment,

Python automatically generates several folders and files.

Example:

```text
my_project/

│
├── venv/
│   ├── Include/
│   ├── Lib/
│   ├── Scripts/          (Windows)
│   ├── bin/              (Linux/macOS)
│   └── pyvenv.cfg
│
└── main.py
```

The exact structure may vary slightly depending on the operating system.

# 📁 Understanding the Generated Files

| File / Folder | Purpose |
|---------------|---------|
| `Include/` | Stores C header files used by Python extensions. |
| `Lib/` | Contains all installed Python packages and libraries. |
| `Scripts/` | Contains the Python executable and activation scripts on Windows. |
| `bin/` | Contains the Python executable and activation scripts on Linux/macOS. |
| `pyvenv.cfg` | Stores configuration information about the Virtual Environment. |

# 💻 Creating Your First Virtual Environment

Python provides the built-in **venv** module for creating Virtual Environments.

### Windows

```bash
python -m venv venv
```

---

### macOS / Linux

```bash
python3 -m venv venv
```

---

### Command Breakdown

| Part | Meaning |
|------|---------|
| `python` | Starts the Python interpreter. |
| `-m` | Runs a module as a script. |
| `venv` | Python's built-in module for creating Virtual Environments. |
| `venv` | Name of the Virtual Environment folder. |

# 📝 Explanation

Suppose your project folder is:

```text
Student_Management/

│
└── app.py
```

Run the following command inside the project folder:

```bash
python -m venv venv
```

Python will create a new folder named:

```text
venv/
```

The folder contains everything needed for an isolated Python environment.

After creation, your project will look like:

```text
Student_Management/

│
├── app.py
└── venv/
```

From now on, packages installed after activating this environment will remain inside this project only.

# 💡 Did You Know?

The name **`venv`** is only a convention.

You can choose any valid folder name, such as:

```text
env/

myenv/

project_env/

python_env/
```

However, most developers simply use:

```text
venv/
```

because it is widely recognized and follows common Python project conventions.

# ▶️ Activating a Virtual Environment

Creating a Virtual Environment is only the first step.

Before installing packages or running your project, you must **activate** the Virtual Environment.

When a Virtual Environment is activated:

- 🐍 Python uses the interpreter inside the Virtual Environment.
- 📦 Packages are installed only inside that environment.
- 🚫 The global Python installation remains unaffected.

Once activated, all Python and pip commands will use the Virtual Environment until it is deactivated.

# 💻 Activation Commands

The activation command depends on your operating system.

| Operating System | Command |
|------------------|---------|
| **Windows (Command Prompt)** | `venv\Scripts\activate` |
| **Windows (PowerShell)** | `venv\Scripts\Activate.ps1` |
| **Linux / macOS** | `source venv/bin/activate` |

Run the appropriate command from your project's root directory.

# 🪟 Windows (Command Prompt)

If your project structure is:

```text
Student_Management/

│
├── app.py
└── venv/
```

Open **Command Prompt** inside the project folder and run:

```bash
venv\Scripts\activate
```

If activation is successful, you will see something similar to:

```text
(venv) C:\Users\Nikita\Student_Management>
```

The `(venv)` prefix indicates that the Virtual Environment is now active.

# 🪟 Windows (PowerShell)

If you are using **PowerShell**, run:

```powershell
venv\Scripts\Activate.ps1
```

If PowerShell blocks the script due to security settings, you may see an execution policy error.

This is normal and can be resolved by changing the PowerShell execution policy (covered later in this notebook).

# 🐧 Linux / 🍎 macOS

On Linux or macOS, activate the Virtual Environment using:

```bash
source venv/bin/activate
```

After activation, the terminal will display something similar to:

```text
(venv) user@computer:~/Student_Management$
```

Again, the `(venv)` prefix confirms that the Virtual Environment is active.

# 🔍 How to Verify That the Virtual Environment is Active

There are several ways to confirm that the Virtual Environment is active.

### ✅ Method 1

Look at the terminal prompt.

Example:

```text
(venv) C:\Projects\Student_Management>
```

If `(venv)` appears, the environment is active.

---

### ✅ Method 2

Check the Python interpreter.

Run:

```bash
python --version
```

or

```bash
where python
```

**Windows**

```bash
where python
```

**Linux / macOS**

```bash
which python
```

The displayed path should point to the Python executable inside the `venv` folder.

# 📦 Installing Packages Inside a Virtual Environment

After activation, packages are installed only inside the Virtual Environment.

Example:

```bash
pip install numpy
```

Install multiple packages:

```bash
pip install pandas matplotlib seaborn
```

These packages will **not** affect the global Python installation.

# 📋 Viewing Installed Packages

To see all installed packages inside the active Virtual Environment, use:

```bash
pip list
```

Example Output:

```text
Package      Version
------------ -------
numpy        2.3.1
pandas       2.3.0
pip          25.1
setuptools   80.9.0
```

The exact versions may differ depending on your installation.

# ⏹️ Deactivating the Virtual Environment

Once you finish working on your project, you can deactivate the Virtual Environment.

Run:

```bash
deactivate
```

After deactivation:

- The `(venv)` prefix disappears.
- Python switches back to the global installation.
- New packages will no longer be installed inside the Virtual Environment.

# 💡 Did You Know?

You only need to create a Virtual Environment **once** for a project.

However, **every time you reopen the project**, you must activate the Virtual Environment again before running your code or installing packages.

# 📄 What is `requirements.txt`?

As Python projects grow, they often use multiple third-party packages such as:

- NumPy
- Pandas
- Matplotlib
- Scikit-learn
- Flask
- Django

If someone else wants to run your project, they also need the same packages and compatible versions.

Instead of manually listing every package, Python developers use a special file called **`requirements.txt`**.

This file contains all the project dependencies along with their versions.

It allows anyone to install the required packages with a single command.

Using a `requirements.txt` file is considered a standard practice in professional Python development.

# 📌 Why Do We Need `requirements.txt`?

Imagine you upload a Machine Learning project to GitHub.

Your project uses:

- NumPy
- Pandas
- Scikit-learn
- Matplotlib

If another developer downloads your project, they may not have these packages installed.

Without a `requirements.txt` file, they would have to guess which packages and versions are required.

This can lead to:

- ❌ Missing packages
- ❌ Version conflicts
- ❌ Import errors
- ❌ Project not working correctly

A `requirements.txt` file solves this problem by providing a complete list of the required dependencies.

# 📤 Creating a `requirements.txt` File

After activating your Virtual Environment and installing all required packages, generate the dependency file using:

```bash
pip freeze > requirements.txt
```

### Command Breakdown

| Part | Meaning |
|------|---------|
| `pip freeze` | Displays all installed packages and their versions. |
| `>` | Redirects the output to a file. |
| `requirements.txt` | Stores the list of project dependencies. |

# 📝 Example `requirements.txt`

After running:

```bash
pip freeze > requirements.txt
```

The file may look like this:

```text
numpy==2.3.1
pandas==2.3.0
matplotlib==3.10.5
scikit-learn==1.7.1
seaborn==0.13.2
```

The exact package versions may differ depending on your environment.

# 📥 Installing Packages from `requirements.txt`

Suppose another developer clones your GitHub repository.

Instead of installing each package manually, they can run:

```bash
pip install -r requirements.txt
```

### Command Breakdown

| Part | Meaning |
|------|---------|
| `pip install` | Installs Python packages. |
| `-r` | Reads package names from a file. |
| `requirements.txt` | File containing project dependencies. |

Python automatically installs all the required packages listed in the file.

# 🔄 Recreating a Virtual Environment

Suppose you share your project with another developer.

They can recreate the same environment using the following steps.

### Step 1

Create a Virtual Environment.

```bash
python -m venv venv
```

---

### Step 2

Activate the Virtual Environment.

**Windows**

```bash
venv\Scripts\activate
```

**Linux / macOS**

```bash
source venv/bin/activate
```

---

### Step 3

Install all dependencies.

```bash
pip install -r requirements.txt
```

Now the project is ready to run with the same dependencies as the original environment.

A typical professional workflow looks like this:

```text
Create Project
       │
       ▼
Create Virtual Environment
       │
       ▼
Activate Virtual Environment
       │
       ▼
Install Required Packages
       │
       ▼
Develop the Project
       │
       ▼
Generate requirements.txt
       │
       ▼
Push the Project to GitHub
       │
       ▼
Other Developers Clone the Project
       │
       ▼
Create Their Own Virtual Environment
       │
       ▼
Install Dependencies
       │
       ▼
Run the Project Successfully
```

# 🚫 Should You Upload the `venv` Folder to GitHub?

**No.**

The `venv` folder should **not** be uploaded to GitHub.

Reasons:

- 📁 It contains hundreds or thousands of generated files.
- 💾 It makes the repository unnecessarily large.
- 🖥️ Virtual Environments are platform-specific.
- 🔄 Other developers should create their own Virtual Environment.

Instead, upload only:

- ✅ Source code
- ✅ `requirements.txt`
- ✅ `README.md`

Then each developer can recreate the environment locally.

# 💡 Did You Know?

Professional Python projects usually include a `.gitignore` file.

This file tells Git which files and folders should **not** be uploaded to GitHub.

A common entry is:

```text
venv/
```

This prevents the Virtual Environment folder from being committed to the repository while allowing the source code and `requirements.txt` to be shared.

# ⚠️ Common Errors

While working with Virtual Environments, beginners often encounter the following errors.

## ❌ Error 1: `'python' is not recognized`

```bash
python -m venv venv
```

Output

```text
'python' is not recognized as an internal or external command
```

**Reason:**

Python is either not installed or its path is not added to the system's environment variables.

**Solution:**

- Install Python.
- Ensure **"Add Python to PATH"** is selected during installation.
- Restart the terminal and try again.

---

## ❌ Error 2: Activation Command Not Working

Windows

```bash
venv\Scripts\activate
```

Output

```text
The system cannot find the path specified.
```

**Reason:**

The Virtual Environment has not been created or you are running the command from the wrong directory.

**Solution:**

- Verify that the `venv` folder exists.
- Open the terminal in the project folder.
- Run the activation command again.

---

## ❌ Error 3: PowerShell Execution Policy Error

```powershell
venv\Scripts\Activate.ps1
```

Output

```text
running scripts is disabled on this system
```

**Reason:**

PowerShell blocks script execution due to its execution policy.

**Solution:**

Run PowerShell as Administrator and execute:

```powershell
Set-ExecutionPolicy RemoteSigned -Scope CurrentUser
```

Then confirm the change and try activating the Virtual Environment again.

---

## ❌ Error 4: `ModuleNotFoundError`

```python
import numpy
```

Output

```text
ModuleNotFoundError: No module named 'numpy'
```

**Reason:**

The package is not installed in the active Virtual Environment.

**Solution:**

```bash
pip install numpy
```

---

## ❌ Error 5: Forgetting to Activate the Virtual Environment

Sometimes developers install packages before activating the Virtual Environment.

As a result, packages are installed in the global Python environment instead of the project-specific environment.

Always activate the Virtual Environment before installing packages or running the project.

# ✅ Best Practices

Follow these best practices while working with Virtual Environments.

### 📦 Create One Virtual Environment Per Project

Each project should have its own isolated environment.

---

### ▶️ Always Activate Before Working

Activate the Virtual Environment before:

- Running Python programs
- Installing packages
- Updating dependencies

---

### 📄 Use `requirements.txt`

Maintain a `requirements.txt` file so others can recreate your project's dependencies easily.

---

### 🚫 Do Not Upload `venv`

Add the following entry to your `.gitignore` file:

```text
venv/
```

Only upload your source code and `requirements.txt`.

---

### 📚 Install Only Required Packages

Avoid installing unnecessary packages inside the Virtual Environment.

---

### 🔄 Recreate Instead of Sharing

Do not copy the `venv` folder between computers.

Instead:

- Create a new Virtual Environment.
- Install dependencies using `requirements.txt`.
