# 📌 What is a Module?

A **Module** is a Python file (`.py`) that contains **functions, variables, classes, or executable code** related to a specific purpose.

Instead of writing an entire program in a single file, Python allows us to organize code into multiple modules. Each module focuses on a particular task and can be imported whenever needed.

Every Python file is considered a module. This means that even the programs you have already written, such as `calculator.py` or `student.py`, are Python modules.

Modules help developers write programs that are organized, reusable, and easier to maintain.



## 📌 Definition

> **A Module is a Python file (`.py`) that contains reusable code such as functions, classes, variables, or executable statements, which can be imported and used in other Python programs.**

# 🐍 Types of Modules

Python provides two main types of modules.

- 🎲 `random`

# 📥 Importing Modules

After creating a module, the next step is to use it in another Python program.

Python provides the **`import`** statement for this purpose.

The `import` statement allows us to access functions, variables, classes, and other objects defined inside a module.

### Basic Syntax

```python
import module_name
```

Here,

- **import** → Python keyword used to import a module.
- **module_name** → Name of the module without the `.py` extension.

For example, if we have a file named:

```text
calculator.py
```

we import it as:

```python
import calculator
```

Notice that we **do not write `.py`** while importing.

Once imported, we can access everything inside the module using the **dot (`.`) operator**.

````python
# main.py

import calculator

result = calculator.add(10, 5)

print("Addition :", result)
````

# 📝 Explanation

Let's understand what happened step by step.

### Step 1

We created a module named:

```text
calculator.py
```

which contains two functions.

---

### Step 2

Inside another Python file, we wrote:

```python
import calculator
```

Python loads the entire module into memory.

---

### Step 3

To call a function inside the module, we write:

```python
calculator.add(10, 5)
```

Here,

- **calculator** → Module Name
- **add()** → Function inside the module

The dot (`.`) operator tells Python:

> "Go inside the `calculator` module and execute the `add()` function."

The output becomes:

```text
Addition : 15
```

This is the most common and recommended way of importing modules because it clearly shows where each function comes from.

# ⭐ Importing Specific Functions

Sometimes we don't need the entire module.

We only need one or two functions.

In such cases, Python allows us to import specific functions directly.

### Syntax

```python

```

Example:

```python

```

Now we can use the function directly without writing the module name.

````python
# main.py

from calculator import add

result = add(25, 15)

print("Addition :", result)
````

# 📝 Explanation

Instead of importing the entire module,

```python
import calculator
```

```python
from calculator import add
```

Now Python directly places the `add()` function into our program.

Therefore,

instead of writing:

```python

```

we simply write:

```python
add(25, 15)
```

This makes the code shorter.

However, importing too many individual functions can sometimes make it difficult to understand where a function originally came from.

Therefore, use this approach only when importing a small number of functions.

# 🏷️ Module Aliasing

Sometimes module names are long or we need to use them frequently.

Typing the complete module name every time can make the code lengthy.

Python allows us to give a module a **temporary alternative name**, called an **alias**.

This is done using the **`as`** keyword.

## Syntax

```python
import module_name as alias_name
```

Example:

```python
import calculator as calc
```

Now we can use:

```python
calc.add()
```

instead of

```python
calculator.add()
```

Aliasing makes code shorter and easier to read, especially when working with long module names.

````python
# main.py

import calculator as calc

print(calc.add(15, 10))

print(calc.subtract(15, 10))
````

# 📝 Explanation

Here,

```python
import calculator as calc
```

creates an alias named **`calc`**.

Now,

```python
calc.add()
```

is exactly the same as

```python
calculator.add()
```

The alias exists only inside the current program.

It does **not** rename the original module.

Aliasing is commonly used in real-world projects.

For example,

```python
import numpy as np

import pandas as pd

import matplotlib.pyplot as plt
```

You will see these aliases almost everywhere in Data Science and Machine Learning.

# 🌟 Wildcard Import (`*`)

Python also allows importing **everything** from a module.

## Syntax

```python
from module_name import *
```

Example:

```python
from calculator import *
```

Now every public function, variable, and class inside the module becomes directly available.

````python
# main.py

from calculator import *

print(add(20, 30))

print(subtract(50, 10))
````

# 📝 Explanation

Using

```python
from calculator import *
```

imports all **public** members from the module.

After importing,

we can directly call:

```python
add()
```

```python
subtract()
```

without writing the module name.

Although this looks convenient, it has some disadvantages.

If two different modules contain functions with the same name,

Python may become confused about which function should be used.

This can make programs difficult to understand and debug.

# ⚠️ Why Wildcard Import is Not Recommended

Although wildcard imports reduce typing,

professional Python developers generally avoid them.

### Problems with Wildcard Import

❌ Namespace Pollution

Many functions and variables are imported into the current program.

---

❌ Name Conflicts

Two different modules may contain functions with the same name.

---

❌ Poor Readability

It becomes difficult to identify where a function originally came from.

---

❌ Difficult Debugging

Finding errors becomes harder because imported objects are not clearly identified.

### ✅ Recommended Practice

Instead of writing:

```python
from calculator import *
```

prefer:

```python
import calculator
```

or

```python
from calculator import add
```

These approaches make your code cleaner, safer, and easier to maintain.

# 🔍 How Python Finds Modules

When Python encounters an **import** statement, it needs to locate the requested module.

For example,

```python
import calculator
```

Python starts searching for a file named:

```text
calculator.py
```

However, Python does **not** search randomly.

It follows a predefined order called the **Module Search Path**.

If the module is found, Python imports it successfully.

Otherwise, Python raises an error.

```text
ModuleNotFoundError
```

Understanding the Module Search Path helps us solve many import-related errors in Python.

# 📂 Module Search Path (`sys.path`)

Python searches for modules in the following order.

## 1️⃣ Current Working Directory

Python first checks the folder containing the currently running program.

If the module exists here, Python imports it immediately.

---

## 2️⃣ PYTHONPATH Environment Variable

If the module is not found in the current directory,

Python checks the directories listed in the **PYTHONPATH** environment variable.

---

## 3️⃣ Standard Library

Next,

Python searches its built-in Standard Library.

Examples include:

- `math`
- `random`
- `datetime`
- `os`
- `sys`

---

## 4️⃣ Installed Third-Party Packages

Finally,

Python checks installed packages such as:

- NumPy
- Pandas
- Matplotlib
- Scikit-learn

If the module still cannot be found,

Python raises:

```text
ModuleNotFoundError
```

````python
import sys

print(sys.path)
````

# 📝 Explanation

The statement

```python
print(sys.path)
```

displays Python's **Module Search Path**.

Each item in the list represents a directory where Python searches for modules.

Whenever an `import` statement is executed,

Python checks these directories one by one until it finds the requested module.

If no matching module exists in any of these locations,

Python raises:

```text
ModuleNotFoundError
```

Understanding `sys.path` is extremely useful when debugging import-related problems.

# 📍 Finding the Location of a Module

Sometimes we want to know where a module is stored on our computer.

Python provides the special attribute:

```python

```

It returns the file path of the imported module.

````python
import math

print(math.__file__)
````

> Note: Some built-in modules (or modules implemented in C) may not expose __file__. In such cases, attempting to access it can raise an AttributeError.

# 📝 Explanation

The special attribute

```python
__file__
```

returns the location of the module file (when available).

This is helpful when:

- locating installed packages,
- understanding which version of a module is being used,
- or debugging import issues.

Not every module has a `__file__` attribute, so always be aware that this attribute may not be available for certain built-in modules.

# ⚠️ Common Mistakes

While learning modules, beginners often make a few common mistakes.

## ❌ Mistake 1: Including `.py` While Importing

Incorrect

```python
import calculator.py
```

Correct

```python
import calculator
```

Python automatically understands that it is importing a Python file.

---

## ❌ Mistake 2: Importing a Module That Does Not Exist

```python
import mathematics
```

If the module is not available, Python raises:

```text
ModuleNotFoundError
```

Always make sure the module exists before importing it.

---

## ❌ Mistake 3: Using Wildcard Imports

```python
from calculator import *
```

Although this works, it is not recommended because it imports everything into the current namespace, making the code harder to understand and maintain.

---

## ❌ Mistake 4: Circular Imports

Avoid situations like:

```text
student.py

↓

imports teacher.py

↓

teacher.py

↓

imports student.py
```

This creates a **Circular Import**, which can lead to import errors and unexpected behavior.

---

## ❌ Mistake 5: Naming Your File Like a Built-in Module

Avoid names such as:

```text
math.py

random.py

os.py

json.py
```

If you create files with these names, Python may import your file instead of the actual built-in module.

This can cause confusing errors.
