# 📌 What is an Import Statement?

An **Import Statement** is a Python statement used to bring code from another module or package into the current program.

Instead of rewriting existing functions, classes, or variables, we can simply import them and use them whenever needed.

Import statements are one of the core features of Python that promote **code reusability**, **modularity**, and **maintainability**.


## 📌 Definition

> **An Import Statement is used to load a module, package, or specific objects from them so they can be used in the current Python program.**

# 🧠 How Import Works in Python

Whenever Python encounters an import statement, it follows a series of steps before making the imported module available.

The process is as follows:

1. 🔍 Python searches for the requested module.
2. 📂 If the module is found, Python loads it into memory.
3. ⚙️ The module is executed (only the first time it is imported).
4. 💾 Python stores the loaded module in memory (cache).
5. ♻️ Future imports reuse the cached module instead of loading it again.

This process improves performance because Python avoids executing the same module multiple times during the same program execution.

# 📊 How Python Imports a Module

```text
Program Starts
      │
      ▼
import math
      │
      ▼
Search for the Module
      │
      ▼
Module Found
      │
      ▼
Execute the Module
      │
      ▼
Store in Memory (Cache)
      │
      ▼
Module Ready to Use
```

This process happens automatically whenever a module is imported.

# 💾 Module Caching

Python imports a module **only once** during a program's execution.

After the first import, the module is stored in memory.

If the same module is imported again,

Python does **not** execute it again.

Instead,

it reuses the already loaded module from memory.

This mechanism is called **Module Caching**.

It makes Python programs faster and more efficient.

````python
# Example

import math

import math

import math

print("Module Imported Successfully!")
````

# 📝 Explanation

Although the `math` module is imported three times,

Python loads and executes it only once.

The remaining import statements simply use the already loaded module from memory.

This behavior improves performance and avoids unnecessary work.

# ✨ Different Types of Import Statements

Python provides several ways to import modules and their contents.

Each method has its own purpose.

| Import Statement | Purpose |
|-----------------|---------|
| `import module` | Import the entire module. |
| `from module import function` | Import specific functions, classes, or variables. |
| `import module as alias` | Import a module with a shorter name (alias). |
| `from module import *` | Import all public objects from a module. |
| `import module1, module2` | Import multiple modules together. |

We will learn each of these import styles with practical examples in the following sections.

# 📥 Importing an Entire Module

The simplest way to use a module is to import the entire module.

### Syntax

```python
import module_name
```

After importing the module,

its functions, classes, and variables are accessed using the **dot (`.`) operator**.

This is the most commonly used import style because it clearly indicates where each function belongs.

````python
import math

print(math.sqrt(25))

print(math.factorial(5))

print(math.pi)
````

# 📝 Explanation

In this example,

we imported the complete **math** module.

```python
import math
```

To access its members,

we use the module name followed by the **dot (`.`) operator**.

Examples:

```python
math.sqrt(25)

math.factorial(5)

math.pi
```

This approach makes the code more readable because it is immediately clear that these functions belong to the **math** module.

For most real-world projects, importing the entire module is considered a clean and recommended practice.

# 💡 Did You Know?

The `math` module is part of Python's **Standard Library**.

This means you can use it directly without installing any additional packages.

Some other commonly used built-in modules include:

- 🎲 `random`
- 📅 `datetime`
- 📂 `os`
- 📁 `pathlib`
- ⏰ `time`
- 📊 `statistics`

Python provides hundreds of built-in modules that can be imported whenever needed.

# 📥 Importing Specific Functions

Sometimes, we don't need an entire module.

Instead, we only need one or two functions from a module.

In such cases, Python allows us to import specific objects directly.

### Syntax

```python
from module_name import function_name
```

Instead of importing the complete module, we import only the required function.

This makes the code shorter and easier to write.

````python
from math import sqrt

print(sqrt(49))
````

# 📝 Explanation

In this example,

we imported only the **sqrt()** function from the **math** module.

```python
from math import sqrt
```

Since only the function is imported,

we can directly write

```python
sqrt(49)
```

instead of

```python
math.sqrt(49)
```

This approach is useful when only a few functions are needed from a module.

# 📥 Importing Multiple Objects

Python also allows importing multiple functions from the same module.

### Syntax

```python
from module_name import function1, function2, function3
```

````python
from math import sqrt, factorial, pi

print(sqrt(64))

print(factorial(6))

print(pi)
````

# 📝 Explanation

Here, we imported three objects from the **math** module.

- `sqrt()`
- `factorial()`
- `pi`

Since they are imported directly,

we don't need to use the module name before each one.

# 🏷️ Importing Modules with Aliases

Sometimes module names are long.

Python allows us to give a module a shorter name using the **`as`** keyword.

This shorter name is called an **Alias**.

### Syntax

```python
import module_name as alias_name
```

````python
import math as m

print(m.sqrt(81))

print(m.factorial(4))

print(m.pi)
````

# 📝 Explanation

Here,

we imported the **math** module using the alias **m**.

```python
import math as m
```

Now,

instead of writing

```python
math.sqrt(81)
```

we write

```python
m.sqrt(81)
```

Aliases make code shorter and improve readability when module names are long.

# ⭐ Wildcard Import

Python also allows importing **all public objects** from a module.

### Syntax

```python
from module_name import *
```

Example:

```python
from math import *
```

After this,

all public functions, constants, and variables of the module become directly available.

````python
from math import *

print(sqrt(100))

print(factorial(5))

print(pi)
````

# 📝 Explanation

The statement

```python
from math import *
```

imports all public objects from the **math** module.

This allows us to use functions like

```python
sqrt()

factorial()

pi
```

without writing the module name.



### ⚠️ Why Wildcard Imports Are Discouraged

Although wildcard imports are supported,

they are **generally not recommended** because:

- ❌ They make it difficult to identify where a function came from.
- ❌ They may overwrite existing names.
- ❌ They reduce code readability.
- ❌ They make debugging more difficult.

In professional Python projects, wildcard imports are usually avoided.

# 💡 Did You Know?

The official Python style guide (**PEP 8**) recommends avoiding wildcard imports (`from module import *`) in most situations because they reduce code clarity and can introduce unexpected name conflicts.

Instead, prefer:

```python
import math
```

or

```python
from math import sqrt
```

These approaches make your code easier to read and maintain.

# 📦 Importing Multiple Modules

Python allows us to import multiple modules in a single program.

This is useful when a program requires functionalities from different modules.

There are two common ways to import multiple modules.

### Method 1: Separate Import Statements

```python
import math
import random
import datetime
```



### Method 2: Single Import Statement

```python
import math, random, datetime
```

Although both methods work, **separate import statements** are generally preferred because they improve readability.

````python
import math
import random
import datetime

print(math.sqrt(81))

print(random.randint(1, 10))

print(datetime.date.today())
````

# 📝 Explanation

In this example, we imported three different modules.

- 📐 `math` → Mathematical operations
- 🎲 `random` → Random number generation
- 📅 `datetime` → Working with dates and times

Each module provides a different functionality.

Importing multiple modules allows us to combine their features within a single program.

# 📚 Importing Built-in vs User-defined Modules

Python supports importing both **Built-in Modules** and **User-defined Modules**.

| Built-in Module | User-defined Module |
|-----------------|---------------------|
| Comes with Python | Created by the programmer |
| No installation required | Written by the user |
| Examples: `math`, `random`, `os` | Examples: `calculator.py`, `student.py` |



### Built-in Module

```python

```



### User-defined Module

Both are imported using almost the same syntax.

# 📊 Comparison of Import Styles

| Import Style | Example | When to Use |
|--------------|---------|-------------|
| `import module` | `import math` | When you need many functions from a module. |
| `from module import function` | `from math import sqrt` | When only one or two functions are required. |

Choosing the appropriate import style improves code readability and maintainability.

# ⚠️ Common Import Errors

While working with import statements, beginners often encounter a few common errors.

## ❌ Error 1: Module Not Found

```python
import maths
```

Output

```text
ModuleNotFoundError: No module named 'maths'
```

**Reason:**

The module name is incorrect.

**Correct**

```python
import math
```

---

## ❌ Error 2: Incorrect Import Path

```python
from math import square
```

Output

```text
ImportError: cannot import name 'square'
```

**Reason:**

The `math` module does not contain a function named `square`.

**Correct**

```python
from math import sqrt
```

---

## ❌ Error 3: Forgetting the Module Name

```python
import math

print(sqrt(25))
```

Output

```text
NameError: name 'sqrt' is not defined
```

**Reason:**

The entire module was imported.

**Correct**

```python
import math

print(math.sqrt(25))
```

---

## ❌ Error 4: Using Relative Imports Incorrectly

```python
from .calculator import add
```

Output

```text
ImportError: attempted relative import with no known parent package
```

**Reason:**

Relative imports only work inside a package.

Running the above statement directly in a Jupyter Notebook or a standalone script will produce this error.

---

## ❌ Error 5: Wildcard Import Causing Name Conflicts

```python
from math import *

from statistics import *
```

Both modules may contain objects with the same name.

This can make it difficult to identify which object is being used.

For this reason, wildcard imports are generally discouraged.
