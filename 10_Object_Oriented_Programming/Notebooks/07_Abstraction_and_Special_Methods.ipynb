# 🎭 What is Abstraction?

Abstraction is one of the four fundamental principles of Object-Oriented Programming (OOP).

It focuses on showing only the **essential features** of an object while hiding unnecessary implementation details.

Instead of worrying about **how** something works internally,

the user only needs to know **what** the object can do.



## 📌 Definition

> **Abstraction is the process of hiding implementation details and exposing only the essential features or functionalities of an object.**

Abstraction allows developers to create cleaner, simpler, and more maintainable programs.



# ✨ Advantages of Abstraction

Abstraction provides several important benefits.

### 🧹 Simplicity

Users interact only with the necessary features.



### 🔒 Better Security

Internal implementation details remain hidden.



### ♻️ Easier Maintenance

Implementation can change without affecting the user interface.



### 📈 Improved Code Reusability

Common interfaces can be reused across multiple classes.



### 🚀 Better Scalability

Applications become easier to extend and modify.



### 🌍 Better Real-World Modeling

Abstraction closely matches how real-world systems work by exposing only useful functionality.



# 🏗️ What is an Abstract Class?

An **Abstract Class** is a special type of class that **cannot be instantiated directly**.

In other words,

we cannot create an object of an Abstract Class.

Instead,

it acts as a **blueprint** for other classes.

The Child Classes inherit from the Abstract Class and provide implementations for its abstract methods.



## 📌 Definition

> **An Abstract Class is a class that cannot be instantiated and is designed to be inherited by other classes.**

It defines the common structure that all child classes must follow.


# 🤔 Why Do We Need Abstract Classes?

Imagine a company.

Every employee must perform work.

However,

different employees perform different types of work.

```
Employee

↓

work()
```

```
Developer

↓

Write Code
```

```
Designer

↓

Create UI Design
```

```
Tester

↓

Test Software
```

Instead of writing a complete implementation inside the Parent Class,

we simply declare that every employee **must implement** the `work()` method.

This is exactly what an Abstract Class does.

It defines **what should be done**, but not **how it should be done**.



# 📦 The `abc` Module

Python provides a built-in module named:

```python
abc
```

which stands for

**Abstract Base Classes**.

This module allows us to create Abstract Classes and Abstract Methods.

To use it,

we import:

```python
from abc import ABC, abstractmethod
```

Here,

- `ABC` → Used to create an Abstract Class.
- `abstractmethod` → Used to create an Abstract Method.

These are the foundation of Abstraction in Python.



````python
# Importing the abc Module

from abc import ABC, abstractmethod
````

# 📝 Explanation

This statement imports two important components.

```python
ABC
```

It allows us to create an Abstract Class.



```python
abstractmethod
```

It is a decorator used to declare Abstract Methods.

Without importing these,

Python cannot create Abstract Classes.

Almost every Abstraction program begins with this import statement.



# 🔑 What is an Abstract Method?

An **Abstract Method** is a method that is **declared but not implemented** inside the Abstract Class.

It acts like a rule.

Every Child Class **must provide its own implementation** of that method.

### Syntax

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
```

Here,

```python
sound()
```

is an Abstract Method.

It contains no implementation.

The Child Classes must override it.



````python
# Example 1: Creating an Abstract Class

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
````

# 📝 Explanation

Let's understand this step by step.

First,

```python
class Animal(ABC):
```

creates an **Abstract Class**.

Next,

```python
@abstractmethod
```

marks the following method as an Abstract Method.

```python
def sound(self):
    pass
```

Notice that the method contains only:

```python
pass
```

This means:

The Parent Class only declares the method.

It does not provide any implementation.

Every Child Class must implement this method.



````python
# Example 2: Trying to Create an Object of an Abstract Class

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


animal = Animal()
````

# 📝 Explanation

This program will generate an error.

Output

```text
TypeError:
Can't instantiate abstract class Animal
with abstract method sound
```

Why?

Because Abstract Classes are **blueprints**.

They are meant to be inherited,

not instantiated directly.

Just like you cannot build a house from only a blueprint,

you cannot create an object directly from an Abstract Class.



# 💡 Key Takeaways

✔ Abstract Classes cannot be instantiated.

✔ Abstract Classes are used as blueprints.

✔ Python provides the `abc` module for creating Abstract Classes.

✔ `ABC` is used to create an Abstract Class.

✔ `@abstractmethod` is used to declare Abstract Methods.

✔ Child Classes must implement all Abstract Methods.



# 👨 Child Class Implementing an Abstract Method

An Abstract Class only declares what should be done.

It does **not** provide the implementation.

The responsibility of implementing the Abstract Method belongs to the Child Class.

If a Child Class inherits from an Abstract Class,

it **must** implement all Abstract Methods.

Otherwise,

that Child Class also becomes an Abstract Class and its object cannot be created.



````python
# Example 1: Implementing an Abstract Method

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        print("Dog says: Bark 🐶")


dog = Dog()

dog.sound()
````

# 📝 Explanation

Let's understand the execution.

The Parent Class

```python
Animal
```

contains an Abstract Method.

```python
sound()
```

The Child Class

```python
Dog
```

implements the same method.

```python
def sound(self):
    print("Dog says: Bark 🐶")
```

Now,

the Child Class is no longer abstract.

Therefore,

Python allows us to create its object.

Output

```text
Dog says: Bark 🐶
```



````python
# Example 2: Multiple Child Classes

class Shape(ABC):

    @abstractmethod
    def draw(self):
        pass


class Circle(Shape):

    def draw(self):
        print("Drawing Circle ⭕")


class Rectangle(Shape):

    def draw(self):
        print("Drawing Rectangle ▭")


circle = Circle()
rectangle = Rectangle()

circle.draw()
rectangle.draw()
````

# 📝 Explanation

The Parent Class

```python
Shape
```

only declares the method

```python
draw()
```

Each Child Class provides its own implementation.

```
Circle

↓

Drawing Circle
```

```
Rectangle

↓

Drawing Rectangle
```

Output

```text
Drawing Circle ⭕

Drawing Rectangle ▭
```

This is one of the biggest advantages of Abstraction.

The Parent defines the interface,

while the Child defines the implementation.



````python
# Example 3: Banking System

from abc import ABC, abstractmethod


class BankAccount(ABC):

    @abstractmethod
    def account_type(self):
        pass


class SavingsAccount(BankAccount):

    def account_type(self):
        print("Savings Account")


class CurrentAccount(BankAccount):

    def account_type(self):
        print("Current Account")


saving = SavingsAccount()
current = CurrentAccount()

saving.account_type()
current.account_type()
````

# 📝 Explanation

The Parent Class only specifies that every account

must implement

```python
account_type()
```

Each account decides how it should display its own type.

Output

```text
Savings Account

Current Account
```

This creates a common interface while allowing different implementations.



````python
# Example 4: Payment System

from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass


class CreditCard(Payment):

    def pay(self):
        print("Payment made using Credit Card 💳")


class UPI(Payment):

    def pay(self):
        print("Payment made using UPI 📱")


credit = CreditCard()
upi = UPI()

credit.pay()
upi.pay()
````

# 📝 Explanation

Every payment method must provide

```python
pay()
```

However,

each payment method implements it differently.

Output

```text
Payment made using Credit Card 💳

Payment made using UPI 📱
```

The Parent Class defines the rule.

The Child Classes define the behavior.



````python
# Example 5: What Happens If We Don't Implement an Abstract Method?

from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):
    pass


car = Car()
````

# 📝 Explanation

The `Vehicle` class contains the Abstract Method

```python
start()
```

However,

the `Car` class does **not** implement it.

When we try to create

```python
car = Car()
```

Python raises an error.

Output

```text
TypeError:
Can't instantiate abstract class Car
without an implementation for abstract method 'start'
```

This happens because every Child Class **must implement all Abstract Methods** inherited from the Abstract Class.

If even one Abstract Method is left unimplemented,

Python treats the Child Class as an Abstract Class.

Therefore,

its object cannot be created.



# 💡 Key Takeaways

✔ An Abstract Class defines the interface.

✔ Child Classes provide the actual implementation.

✔ Every Child Class must implement all Abstract Methods.

✔ If a Child Class does not implement an Abstract Method, it cannot be instantiated.

✔ Abstraction enforces consistency by ensuring that all Child Classes follow the same interface.

✔ This concept is widely used in frameworks, APIs, and large-scale software applications.



# 🔄 Abstraction vs Encapsulation

Abstraction and Encapsulation are two important principles of Object-Oriented Programming.

Many beginners get confused between them because both deal with **hiding** something.

However,

their purpose is completely different.



## 🎭 Abstraction

Abstraction focuses on:

> **Hiding the implementation details and showing only the essential features.**

The user only knows **what** an object does.

The user does not need to know **how** it works internally.



## 🔒 Encapsulation

Encapsulation focuses on:

> **Hiding data and protecting it from unauthorized access.**

It combines data and methods into a single unit and controls how the data is accessed.



# 🌍 Real-World Example

Let's compare them using a Car.

## 🚗 Abstraction

When you drive a car,

you only use:

- Steering Wheel
- Brake
- Accelerator
- Gear

You do not know how the engine works internally.

The complex implementation is hidden.

This is **Abstraction**.



## 🔒 Encapsulation

Now imagine the car's engine.

Many internal parts are protected inside the engine compartment.

You cannot directly access or modify them while driving.

Only authorized mechanics can safely access them.

This is similar to **Encapsulation**,

where data is protected and controlled.



# 📊 Abstraction vs Encapsulation

| Feature | Abstraction | Encapsulation |
|----------|-------------|---------------|
| Main Goal | Hide implementation details | Protect data |
| Focus | What an object does | How data is protected |
| Hides | Complex implementation | Internal data |
| Achieved Using | Abstract Classes, Interfaces | Private Members, Getter & Setter |
| Improves | Simplicity | Security |
| Example | Car Driving | Bank Account PIN |


## 📝 Easy Trick to Remember

### 🎭 Abstraction

> **"Hide the implementation."**



### 🔒 Encapsulation

> **"Hide the data."**

This simple trick helps remember the difference during interviews.



# 💡 Key Takeaways

✔ **Abstraction** hides implementation details.

✔ **Encapsulation** hides and protects data.

✔ Abstraction focuses on **what** an object does.

✔ Encapsulation focuses on **how** data is accessed and protected.

✔ Abstraction is commonly implemented using **Abstract Classes** and **Abstract Methods**.

✔ Encapsulation is commonly implemented using **Private Members**, **Getter Methods**, and **Setter Methods**.

✔ Both concepts improve software design but solve different problems.



# ⚠️ Common Mistakes

While learning Abstraction, beginners often make the following mistakes.



## ❌ Mistake 1: Creating Objects of an Abstract Class

Incorrect

```python
from abc import ABC

class Animal(ABC):
    pass

animal = Animal()
```

Abstract Classes are meant to be inherited.

They should not be instantiated directly.



## ❌ Mistake 2: Forgetting to Implement Abstract Methods

Incorrect

```python
class Dog(Animal):
    pass
```

If the Parent Class contains Abstract Methods,

every Child Class must implement them.

Otherwise,

Python will not allow object creation.



## ❌ Mistake 3: Forgetting to Import the `abc` Module

Incorrect

```python
class Animal(ABC):
```

without importing

```python
from abc import ABC, abstractmethod
```

Always import the required classes before creating an Abstract Class.



## ❌ Mistake 4: Confusing Abstraction with Encapsulation

Many beginners think both concepts are the same.

Remember:

- Abstraction → Hide implementation.
- Encapsulation → Hide and protect data.



## ❌ Mistake 5: Using Abstract Classes Unnecessarily

Not every class needs to be an Abstract Class.

Use Abstract Classes only when multiple child classes must follow a common interface.



# 💡 Best Practices

Follow these best practices while implementing Abstraction.



## ✅ Create Clear Abstract Interfaces

Keep only the essential methods inside the Abstract Class.

Avoid unnecessary implementation.



## ✅ Keep Child Classes Focused

Each Child Class should provide its own implementation for every Abstract Method.


## ✅ Use Meaningful Method Names

Examples

```text
draw()

pay()

start()

calculate_salary()

send_notification()
```

Meaningful names improve readability.



## ✅ Reuse Common Structure

If multiple classes perform similar tasks,

define a common interface using an Abstract Class.



## ✅ Avoid Duplicate Code

Keep common logic inside the Parent Class whenever possible.

Only override methods when behavior needs to change.
