# 🔒 What is Encapsulation?

Encapsulation is one of the four fundamental principles of Object-Oriented Programming (OOP).

It is the process of **bundling data (attributes) and methods (functions) into a single unit (class)** while controlling access to that data.

Instead of allowing direct access to important information, Encapsulation encourages us to use methods to read or update data safely.



## 📌 Definition

> **Encapsulation is the process of combining data and methods into a single class while protecting the object's data from unauthorized or accidental access.**

Encapsulation helps us write programs that are:

- 🔒 More Secure
- 🧹 Better Organized
- 🔄 Easier to Maintain
- ♻️ More Reusable


# ✨ Features of Encapsulation

Encapsulation provides several important benefits in Object-Oriented Programming.

### 🔐 Data Protection

Sensitive data can be protected from direct modification.



### 🎯 Controlled Access

Data can only be accessed or modified through specific methods.


### ♻️ Better Code Organization

Data and methods related to the same object remain together inside one class.



### 🧹 Easier Maintenance

Changes can be made inside the class without affecting the rest of the program.



### 📈 Improved Code Reusability

Encapsulated classes can be reused in multiple projects.



### 🛡️ Better Data Validation

We can validate data before allowing any changes.

For example:

- Prevent negative bank balances.
- Prevent invalid student ages.
- Prevent empty usernames.



# 🌟 Public Members

By default, every attribute and method in Python is **Public**.

This means they can be accessed from **inside the class** as well as **outside the class**.

No special symbol is required while creating public members.

### Syntax

```python
class Student:

    def __init__(self):
        self.name = "Rahul"
```

Here,

```python
self.name
```

is a **Public Attribute**.

Similarly,

```python
def display():
```

is a **Public Method**.

Public members are the simplest and most commonly used members in Python.



````python
# Example 1: Public Attribute

class Student:

    def __init__(self):
        self.name = "Rahul"

student = Student()

print(student.name)
````

# 📝 Explanation

In this example,

```python
self.name = "Rahul"
```

creates a **public attribute** named `name`.

Since it is public, we can access it directly outside the class.

```python
print(student.name)
```

Output

```text
Rahul
```

There are no restrictions on reading or modifying public attributes.



````python
# Example 2: Modifying a Public Attribute

class Student:

    def __init__(self):
        self.name = "Rahul"

student = Student()

print("Before:", student.name)

student.name = "Priya"

print("After :", student.name)
````

# 📝 Explanation

Initially,

```python
student.name
```

stores

```text
Rahul
```

Later,

```python
student.name = "Priya"
```

updates the value.

Output

```text
Before: Rahul

After : Priya
```

Since the attribute is public, anyone can modify it directly.

This flexibility is useful, but it may also become risky when dealing with sensitive information.



````python
# Example 3: Public Method

class Student:

    def __init__(self):
        self.name = "Rahul"

    def display(self):
        print("Student Name:", self.name)

student = Student()

student.display()
````

# 📝 Explanation

Here,

```python
display()
```

is a **public method**.

It can be called directly from outside the class.

```python
student.display()
```

Output

```text
Student Name: Rahul
```

Public methods are commonly used to display or process object data.



````python
# Example 4: Real-World Example

class BankAccount:

    def __init__(self):
        self.balance = 10000

account = BankAccount()

print("Current Balance:", account.balance)

account.balance = 50000

print("Updated Balance:", account.balance)
````

In this example,

balance

# 🛡️ Protected Members

Protected Members are attributes and methods that are **intended for internal use** within the class and its child classes.

In Python, a protected member is created by adding a **single underscore (`_`)** before its name.

### Syntax

```python
class Student:

    def __init__(self):
        self._name = "Rahul"
```

Here,

```python
self._name
```

is a **Protected Attribute**.

Similarly,

```python
def _display():
```

is a **Protected Method**.



## 📌 Important Note

Unlike some programming languages such as Java or C++, Python **does not strictly prevent access** to protected members.

The single underscore (`_`) is only a **convention** that tells other developers:

> "This member is intended for internal use. Please don't access it directly unless necessary."



# 🤔 Why Do We Need Protected Members?

Suppose you are developing an Employee Management System.

Each employee has:

- 👤 Name
- 💰 Salary
- 🏢 Department

The salary should not be modified directly by every part of the program.

Instead, only trusted methods or related classes should work with it.

Protected members help communicate this idea.

They indicate that a variable is meant for internal use and should be handled carefully.

Although Python allows access to protected members, following this convention improves code quality and teamwork.



````python
# Example 1: Protected Attribute

class Student:

    def __init__(self):
        self._name = "Rahul"

student = Student()

print(student._name)
````

# 📝 Explanation

In this example,

```python
self._name
```

creates a protected attribute.

Notice that we are still able to access it.

```python
print(student._name)
```

Output

```text
Rahul
```

This surprises many beginners.

### Why does it work?

Because Python **does not enforce** protection for a single underscore (`_`).

Instead, it follows the philosophy:

> **"We are all responsible developers."**

The underscore simply acts as a warning that this member is intended for internal use.



````python
# Example 2: Modifying a Protected Attribute

class Student:

    def __init__(self):
        self._name = "Rahul"

student = Student()

print("Before:", student._name)

student._name = "Priya"

print("After :", student._name)
````

# 📝 Explanation

Even though `_name` is protected, Python still allows us to modify it.

Output

```text
Before: Rahul

After : Priya
```

This shows that:

- Public → Fully accessible.
- Protected → Accessible, but should be treated as internal.

Protected members rely on developer discipline rather than language restrictions.



````python
# Example 3: Protected Method

class Student:

    def __init__(self):
        self._name = "Rahul"

    def _display(self):
        print("Student Name:", self._name)

student = Student()

student._display()
````

# 📝 Explanation

Methods can also be protected.

Here,

```python
def _display(self):
```

creates a protected method.

Although we can call it directly,

```python
student._display()
```

the underscore indicates that it is intended for internal use.

Output

```text
Student Name: Rahul
```

In larger projects, protected methods are usually called from other methods inside the class or from child classes.



````python
# Example 4: Real-World Example

class Employee:

    def __init__(self):
        self._salary = 50000

employee = Employee()

print("Salary:", employee._salary)

employee._salary = 60000

print("Updated Salary:", employee._salary)
````

# 📝 Explanation

Imagine `_salary` represents an employee's salary.

Technically, Python allows us to write:

```python
employee._salary = 60000
```

However, in professional projects, directly modifying protected data is discouraged.

Instead, developers usually create methods to safely update important information.

This makes the program easier to maintain and reduces accidental mistakes.



# 🌍 Public vs Protected Members

| Feature | Public Member | Protected Member |
|---------|---------------|------------------|
| Naming | `name` | `_name` |
| Accessible Outside Class | ✅ Yes | ✅ Yes (but discouraged) |
| Intended for Internal Use | ❌ No | ✅ Yes |
| Python Restricts Access | ❌ No | ❌ No |
| Purpose | General access | Internal use by class and subclasses |



### 📝 Remember

- **Public (`name`)** → Free to access and modify.
- **Protected (`_name`)** → Can be accessed, but should be treated as an internal member.

The underscore (`_`) is a convention that improves code readability and communicates developer intent.



# 🔒 Private Members

Private Members are used to provide a higher level of data protection.

In Python, a private member is created by adding **double underscores (`__`)** before its name.

### Syntax

```python
class Student:

    def __init__(self):
        self.__name = "Rahul"
```

Here,

```python
self.__name
```

is a **Private Attribute**.

Similarly,

```python
def __display():
```

creates a **Private Method**.



## 📌 Important Note

Private members are **not directly accessible** outside the class.

Python internally changes their names using a mechanism called **Name Mangling**.

This makes accidental access much more difficult.



# 🤔 Why Do We Need Private Members?

Imagine you are developing an Online Banking System.

Each customer has:

- 👤 Name
- 💳 Account Number
- 💰 Balance
- 🔑 PIN

Should everyone be able to directly access or modify the PIN?

❌ Absolutely not.

Sensitive information like:

- ATM PIN
- Password
- Salary
- Account Balance
- API Keys

should be protected.

Private members help us hide such important data from direct external access.

Instead of allowing direct modification, we usually provide controlled methods (Getter and Setter) to work with private data safely.



````python
# Example 1: Private Attribute

class Student:

    def __init__(self):
        self.__name = "Rahul"

student = Student()

print(student.__name)
````

# 📝 Explanation

This example will **NOT** work.

Python raises an error similar to:

```text
AttributeError:
'Student' object has no attribute '__name'
```

Many beginners think the variable doesn't exist.

Actually,

```
self.__name
```

was successfully created.

The problem is that Python automatically changes its name internally.

This process is called **Name Mangling**.

We will learn it in detail shortly.



````python
# Example 2: Accessing through a class method

class Student:

    def __init__(self):
        self.__name = "Rahul"

    def display(self):
        print("Student Name:", self.__name)

student = Student()

student.display()
````

# 📝 Explanation

Although private members cannot be accessed directly from outside the class,

they **can** be accessed inside the class.

Here,

```python
self.__name
```

is used inside

```python
display()
```

Since both belong to the same class,

Python allows access.

Output

```text
Student Name: Rahul
```

This is the main idea of Encapsulation.

External code cannot directly access private data,

but the class itself can work with it.



````python
# Example 3: Multiple Private Attributes

class Employee:

    def __init__(self):
        self.__name = "Rahul"
        self.__salary = 50000

    def display(self):
        print("Employee :", self.__name)
        print("Salary   :", self.__salary)

employee = Employee()

employee.display()
````

# 📝 Explanation

The Employee object contains two private attributes.

```
Employee Object

│

├── __name

└── __salary
```

Both attributes are protected from direct external access.

Only the class method

```python
display()
```

can access them.

Output

```text
Employee : Rahul

Salary   : 50000
```

This helps keep sensitive information secure.



````python
# Example 4: Attempting to Modify a Private Attribute

class BankAccount:

    def __init__(self):
        self.__balance = 10000

account = BankAccount()

account.__balance = 50000

print(account.__balance)
````

# 📝 Explanation

Many beginners expect this code to update the private balance.

But that is **not** what happens.

Instead,

```python
account.__balance = 50000
```

creates a **new public attribute** named `__balance` outside the class.

The original private attribute inside the object remains unchanged.

This happens because Python protects the original private member using **Name Mangling**.

This is one of the most common beginner mistakes.



# 💡 Key Takeaways

✔ Private members begin with **double underscores (`__`)**.

✔ They cannot be accessed directly from outside the class.

✔ They can be accessed freely inside the same class.

✔ Private members are mainly used for sensitive data.

# 🧩 What is Name Mangling?

Many beginners believe that Private Members cannot be accessed because Python completely hides them.

This is **not entirely true**.

Python actually uses a technique called **Name Mangling**.

When a variable starts with **double underscores (`__`)**, Python automatically changes its internal name.

For example,

```python
self.__name
```

internally becomes something similar to:

```python
self._Student__name
```

Here,

- `Student` → Class Name
- `__name` → Original Private Variable

Python does this automatically to reduce accidental access and name conflicts.



## 📌 Definition

> **Name Mangling is the process by which Python internally changes the name of private members by adding the class name before them.**



````python
# Example 1: Understanding Name Mangling

class Student:

    def __init__(self):
        self.__name = "Rahul"

student = Student()

print(student.__dict__)
````

# 📝 Explanation

The special attribute

```python
__dict__
```

returns all attributes stored inside the object.

Output

```python
{'_Student__name': 'Rahul'}
```

Notice something interesting.

We created:

```python
self.__name
```

But Python stored it as:

```python
_Student__name
```

This is **Name Mangling**.

Python automatically changes the variable name internally.



````python
# Example 2: Accessing a Private Attribute using Name Mangling

class Student:

    def __init__(self):
        self.__name = "Rahul"

student = Student()

print(student._Student__name)
````

# 📝 Explanation

Although direct access is not allowed,

```python
student.__name
```

raises an error.

We can still access the variable using its mangled name.

```python
student._Student__name
```

Output

```text
Rahul
```

### ⚠️ Important

This technique is mainly useful for learning and debugging.

In professional code, you should **avoid accessing private members this way**.

Instead, use Getter and Setter methods, which we will learn next.



````python
# Example 3: Modifying a Private Attribute using Name Mangling

class Student:

    def __init__(self):
        self.__name = "Rahul"

student = Student()

student._Student__name = "Priya"

print(student._Student__name)
````

# 📝 Explanation

Using the mangled name,

```python
student._Student__name
```

Python allows us to modify the private attribute.

Output

```text
Priya
```

However,

this breaks the purpose of Encapsulation.

Private members are meant to be accessed through controlled methods, not by bypassing the protection.



````python
# Example 4: Name Mangling in Another Class

class BankAccount:

    def __init__(self):
        self.__balance = 10000

account = BankAccount()

print(account.__dict__)
````

# 📝 Explanation

Output

```python
{'_BankAccount__balance': 10000}
```

Notice the pattern.

Python creates the new name as:

```text
_ClassName__VariableName
```

Examples:

```text
Student

↓

_Student__name
```

```text
BankAccount

↓

_BankAccount__balance
```

Every class gets its own unique mangled variable names.

This helps avoid accidental name conflicts and provides an extra layer of protection.

# 🔑 Getter and Setter Methods

Until now, we learned that private members cannot be accessed directly from outside the class.

So the question is...

🤔 **How can we safely read or update private data?**

The answer is:

- 🔑 Getter Method
- ✏️ Setter Method

These methods provide **controlled access** to private attributes.

Instead of allowing direct access,

```python
student.__name
```

we create methods that decide:

- Who can read the data.
- Who can modify the data.
- Which values are valid.
- Which values should be rejected.

This is one of the biggest advantages of Encapsulation.



# 📖 What is a Getter Method?

A **Getter Method** is a public method used to safely read the value of a private attribute.

Instead of accessing the private variable directly,

```python
student.__name
```

we use a getter method.

### Syntax

```python
def get_name(self):
    return self.__name
```

The getter method returns the value of the private attribute.

This allows the class to decide how the data should be exposed.



````python
# Example 1: Simple Getter Method

class Student:

    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

student = Student("Rahul")

print(student.get_name())
````

# 📝 Explanation

Let's understand the flow.

### Step 1

The constructor stores:

```python
self.__name = name
```

Now,

```
Student Object

↓

__name = Rahul
```



### Step 2

Instead of writing

```python
student.__name
```

(which gives an error),

we call

```python
student.get_name()
```

The getter method safely returns the private value.

Output

```text
Rahul
```

The private attribute remains protected while still allowing controlled access.



# ✏️ What is a Setter Method?

A **Setter Method** is a public method used to safely update the value of a private attribute.

Instead of writing:

```python
student.__name = "Priya"
```

(which is not allowed),

we use a setter method.

### Syntax

```python
def set_name(self, name):
    self.__name = name
```

The setter method allows us to update the private attribute in a controlled manner.

It can also validate data before making any changes.


````python
# Example 2: Simple Setter Method

class Student:

    def __init__(self, name):
        self.__name = name

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

student = Student("Rahul")

print("Before:", student.get_name())

student.set_name("Priya")

print("After :", student.get_name())
````

# 📝 Explanation

Initially,

```
__name = Rahul
```

The setter method executes:

```python
student.set_name("Priya")
```

Internally,

```python
self.__name = "Priya"
```

updates the private variable.

Output

```text
Before: Rahul

After : Priya
```

Notice that we never directly accessed the private attribute.

Everything happened through controlled methods.


````python
# Example 3: Setter with Validation

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):

        if balance >= 0:
            self.__balance = balance
        else:
            print("❌ Balance cannot be negative.")

account = BankAccount(10000)

print("Current Balance:", account.get_balance())

account.set_balance(25000)

print("Updated Balance:", account.get_balance())

account.set_balance(-5000)

print("Final Balance:", account.get_balance())
````

# 📝 Explanation

This is a real-world example.

Initially,

```
Balance = ₹10000
```

When we write:

```python
account.set_balance(25000)
```

The condition

```python
balance >= 0
```

is True.

So the balance becomes:

```
₹25000
```

Later,

```python
account.set_balance(-5000)
```

The condition becomes False.

Instead of updating the balance,

the program prints:

```text
❌ Balance cannot be negative.
```

The original balance remains unchanged.

This is the biggest advantage of using Setter Methods.

We can validate data before allowing changes.


````python
# Example 4: Student Age Validation

class Student:

    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):

        if age >= 5:
            self.__age = age
        else:
            print("❌ Invalid age.")

student = Student(18)

print(student.get_age())

student.set_age(20)

print(student.get_age())

student.set_age(2)

print(student.get_age())
````

# 📝 Explanation

This example validates a student's age.

Valid updates:

```python
20
```

Invalid updates:

```python
2
```

Output

```text
18

20

❌ Invalid age.

20
```

The invalid value is rejected.

This keeps the object's data accurate and reliable.

This is exactly how Encapsulation helps build secure applications.

# ⚠️ Common Mistakes

While learning Encapsulation, beginners often make the following mistakes.



## ❌ Mistake 1: Using a Single Underscore Instead of Double Underscores

Incorrect

```python
self._password = "1234"
```

If your goal is to create a **private** attribute, this is incorrect.

Correct

```python
self.__password = "1234"
```

Remember:

- `_variable` → Protected
- `__variable` → Private



## ❌ Mistake 2: Accessing Private Members Directly

Incorrect

```python
student.__name
```

This raises an `AttributeError`.

Correct

```python
student.get_name()
```

Always use Getter methods to access private data.



## ❌ Mistake 3: Modifying Private Data Directly

Incorrect

```python
student.__age = 20
```

This creates a **new attribute** instead of updating the original private attribute.

Correct

```python
student.set_age(20)
```



## ❌ Mistake 4: Ignoring Validation in Setter Methods

Incorrect

```python
def set_balance(self, balance):
    self.__balance = balance
```

Correct

```python
def set_balance(self, balance):

    if balance >= 0:
        self.__balance = balance
```

Always validate important data before updating it.


## ❌ Mistake 5: Misunderstanding Name Mangling

Many beginners think Python completely hides private variables.

Actually,

Python internally renames them using **Name Mangling**.

It does **not** make them impossible to access.

| Data Validation | Checks values before updating | `if age >= 5:` |



## 📝 Remember

✔ Public → `name`

✔ Protected → `_name`

✔ Private → `__name`

✔ Getter → Read data

✔ Setter → Update data

✔ Name Mangling → `_ClassName__Variable`

✔ Use Setters whenever validation is required.
