- Procedural Programming
- Functional Programming

- ✔ Easy to understand
- ✔ Easy to maintain
- ✔ Easy to reuse
- ✔ Easy to extend
- ✔ Suitable for large-scale applications

# 💻 Procedural Programming vs Object-Oriented Programming

Before learning OOP, let's understand why it was introduced.

### 🔹 Procedural Programming

In Procedural Programming:

- The program is divided into functions.
- Data and functions are usually kept separate.
- Suitable for small projects.
- Becomes difficult to manage in large applications.



### 🔹 Object-Oriented Programming

In Object-Oriented Programming:

- The program is divided into objects.
- Each object contains both data and methods.
- Code becomes modular and reusable.
- Easier to maintain and extend.
- Suitable for real-world and enterprise-level applications.

As software grows, OOP provides a much better way to organize and manage code.

# 🧠 What is Object-Oriented Programming (OOP)?

Object-Oriented Programming (OOP) is a programming paradigm that organizes a program into **objects** instead of just functions and procedures.

An **object** is a real-world entity that contains:

- 📦 **Data (Attributes)** → Characteristics or properties of the object.
- ⚙️ **Behavior (Methods)** → Actions the object can perform.

In OOP, we first create a **Class**, which acts as a blueprint or template. Then, we create **Objects** from that class.

For example:

- 👨‍🎓 Student
- 🚗 Car
- 📱 Mobile
- 🏦 Bank Account
- 🛒 Shopping Cart

Each object has its own data and behavior.

This approach makes programs more organized, reusable, and easier to maintain.



# 🏗️ Basic Structure of OOP

Object-Oriented Programming mainly revolves around two important concepts:

## 📘 Class

A **Class** is a blueprint or template used to create objects.

It defines:

- Attributes (Data)
- Methods (Functions)

Think of a class as a design or plan.



## 📦 Object

An **Object** is an actual instance of a class.

Every object has its own values while sharing the same structure defined by the class.



### 📊 Simple Representation

Class (Blueprint)

⬇️

Object 1

Object 2

Object 3



Example:

Class → Student

Objects:

- Rahul
- Priya
- Aman
- Neha

All students belong to the same class (Student), but each has different information such as name, age, and roll number.



# ⭐ Features of Object-Oriented Programming

Object-Oriented Programming provides several powerful features that make software development easier and more efficient.

Some of the major features are:

### ✅ Class

Defines the blueprint for creating objects.



### ✅ Object

Represents a real-world entity created from a class.



### ✅ Encapsulation

Bundles data and methods together while protecting data from unauthorized access.



### ✅ Inheritance

Allows one class to inherit properties and methods from another class.



### ✅ Polymorphism

Allows the same method or function to behave differently in different situations.



### ✅ Abstraction

Hides unnecessary implementation details and shows only essential information.



Together, these features make code:

- Reusable
- Modular
- Secure
- Easy to Maintain
- Easy to Extend



# 🏛️ Four Pillars of Object-Oriented Programming

The foundation of Object-Oriented Programming is built on four important principles.

## 1️⃣ Encapsulation 🔒

Combines data and methods into a single unit and protects the data from direct access.



## 2️⃣ Inheritance 👨‍👩‍👧

Allows a class to inherit features from another class.

This helps in code reusability.



## 3️⃣ Polymorphism 🎭

Allows the same interface or method to perform different tasks depending on the object.



## 4️⃣ Abstraction 🎨

Shows only important information while hiding internal implementation details.



These four pillars make Object-Oriented Programming powerful and suitable for building large-scale software applications.

# 🌟 Advantages of Object-Oriented Programming

Object-Oriented Programming offers many benefits that make software development easier and more efficient.

## ✅ Code Reusability

Classes can be reused in multiple projects, reducing development time.



## ✅ Better Code Organization

Related data and methods are grouped together, making code easier to understand.



## ✅ Easy Maintenance

Large applications become easier to update and maintain because the code is modular.



## ✅ Improved Security

Encapsulation helps protect sensitive data from unauthorized access.



## ✅ Scalability

New features can be added without affecting the existing code significantly.



## ✅ Real-World Modeling

Objects represent real-world entities, making software easier to design and understand.



# ⚠️ Limitations of Object-Oriented Programming

Although OOP is very powerful, it also has some limitations.

### ❌ Steeper Learning Curve

Beginners may initially find concepts like inheritance and polymorphism difficult to understand.



### ❌ More Planning Required

Designing classes and relationships requires careful planning before coding.



### ❌ Increased Memory Usage

Objects consume additional memory compared to simple procedural programs.



### ❌ Not Always Necessary

For very small programs or scripts, using OOP may add unnecessary complexity.


Despite these limitations, OOP remains one of the most widely used programming paradigms because its advantages outweigh its disadvantages in most real-world applications.
