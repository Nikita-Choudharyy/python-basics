# 🔄 Method Overriding as Polymorphism

One of the most common ways to implement Polymorphism in Python is through **Method Overriding**.

When a Child Class defines a method with the **same name** as a method in its Parent Class, the Child Class provides its own implementation.

When the method is called using the Child object,

Python automatically executes the Child Class's version.

This behavior is known as **Runtime Polymorphism**.


## 📌 Definition

> **Runtime Polymorphism occurs when Python decides which overridden method to execute while the program is running, based on the object's actual type.**



# 🤔 Why Do We Need Method Overriding?

Imagine a company.

Every employee works.

```
Employee

↓

work()
```

But,

different employees perform different types of work.

👨‍💻 Developer

→ Writes Code



🎨 Designer

→ Creates UI Designs



🧪 Tester

→ Tests Software

The action is the same:

```text
work()
```

But every role performs that action differently.

Instead of creating different method names,

we use the same method name and allow each class to provide its own implementation.

This is Method Overriding and a practical example of Polymorphism.


````python
# Example 1: Notification System

class Notification:

    def send(self):
        print("Sending a notification.")


class EmailNotification(Notification):

    def send(self):
        print("Sending notification via Email 📧")


class SMSNotification(Notification):

    def send(self):
        print("Sending notification via SMS 📱")


email = EmailNotification()
sms = SMSNotification()

email.send()
sms.send()
````

# 📝 Explanation

The Parent Class provides a general notification method.

```python

```

Different Child Classes override the same method.

```

↓

send()
```

↓

```

```

↓

```
SMS → SMS Notification
```

Output

```text
Sending notification via Email 📧
Sending notification via SMS 📱
```

Although the method name is the same,

each object performs the action differently.

This is Runtime Polymorphism.

# 🦆 Duck Typing

Duck Typing is one of Python's most powerful features.

Unlike many other programming languages,

Python does **not** check an object's type before calling its methods.

Instead,

Python checks whether the object has the required method or not.

If the object has the required method,

Python executes it successfully.

Otherwise,

it raises an error.



## 📌 Definition

> **Duck Typing is a programming concept in which Python focuses on an object's behavior (methods and attributes) instead of its actual type.**

A popular saying explains this idea:

> **"If it walks like a duck and quacks like a duck, then it is treated as a duck."**



# 🤔 Why Do We Need Duck Typing?

Imagine a TV Remote.

The remote has one button.

```text
Power
```

Different devices respond differently.

📺 Television

→ Turns ON



🖥️ Computer

→ Boots the System



🎮 Gaming Console

→ Starts the Console

The remote never checks

```
Is this a TV?

Is this a Computer?

Is this a Gaming Console?
```

It simply sends the **Power** command.

If the device supports it,

the action is performed.

Python behaves in the same way.

It checks whether the object supports the required method,

not what type the object belongs to.



````python
# Example 1: Basic Duck Typing

class Dog:

    def speak(self):
        print("Dog says: Bark 🐶")


class Cat:

    def speak(self):
        print("Cat says: Meow 🐱")


def animal_sound(animal):
    animal.speak()


dog = Dog()
cat = Cat()

animal_sound(dog)
animal_sound(cat)
````

# 📝 Explanation

Notice the function:

```python
animal_sound(animal)
```

Inside the function,

Python simply executes:

```python
animal.speak()
```

It never checks:

```python
isinstance(animal, Dog)
```

or

```python
isinstance(animal, Cat)
```

As long as the object has the method:

```python
speak()
```

the function works correctly.

Output

```text
Dog says: Bark 🐶

Cat says: Meow 🐱
```

This is the core idea of Duck Typing.



````python
# Example 2: Notification System

class Email:

    def send(self):
        print("Sending Email 📧")


class SMS:

    def send(self):
        print("Sending SMS 📱")


class WhatsApp:

    def send(self):
        print("Sending WhatsApp Message 💬")


def notify(notification):
    notification.send()


notify(Email())
notify(SMS())
notify(WhatsApp())
````

# 📝 Explanation

# ➕ Operator Overloading

Operators such as:

```python
+
-
*
/
==
<
>
```

are commonly used in Python.

Interestingly,

the **same operator** can perform different operations depending on the data type.

For example,

```python
10 + 20
```

performs addition.

Whereas,

```python
"Hello " + "World"
```

performs string concatenation.

Although the operator is the same,

its behavior changes depending on the operands.

This is called **Operator Overloading**.



## 📌 Definition

> **Operator Overloading is a feature of Polymorphism where the same operator performs different operations depending on the objects or data types involved.**



# 🤔 Why Do We Need Operator Overloading?

Imagine the **"+"** symbol.

For different data types, it performs different tasks.

```
Numbers

↓

Addition
```

```
Strings

↓

Concatenation
```

```
Lists

↓

Combine Lists
```

The operator remains the same.

Only its behavior changes depending on the operands.

Python achieves this using **Special (Magic/Dunder) Methods**.

For example,

the `+` operator internally calls:

```python
__add__()
```


````python
# Example 1: Addition of Numbers

a = 10
b = 20

print(a + b)
````

# 📝 Explanation

Here,

both operands are integers.

```python
10 + 20
```

Output

```text
30
```

Python internally calls

```python
a.__add__(b)
```

The `+` operator actually invokes the `__add__()` special method behind the scenes.



````python
# Example 2: String Concatenation

first_name = "Nikita "
last_name = "Choudhary"

print(first_name + last_name)
````

# 📝 Explanation

Here,

the operands are strings.

Instead of mathematical addition,

Python joins both strings together.

Output

```text
Nikita Choudhary
```

Internally,

Python again uses

```python
__add__()
```

but the string version of that method.

This is Operator Overloading.



````python
# Example 3: Combining Lists

list1 = [1, 2, 3]
list2 = [4, 5, 6]

print(list1 + list2)
````

# 📝 Explanation

The `+` operator behaves differently for lists.

Instead of adding elements,

it combines both lists.

Output

```text
[1, 2, 3, 4, 5, 6]
```

The same operator

```python
+
```

now performs list concatenation.

Again,

this is possible because of Operator Overloading.



````python
# Example 4: Custom Operator Overloading

class Book:

    def __init__(self, pages):
        self.pages = pages

    def __add__(self, other):
        return self.pages + other.pages


book1 = Book(250)
book2 = Book(150)

print(book1 + book2)
````

# 📝 Explanation

This is our **first custom Operator Overloading example**.

Normally,

Python does not know how to add two `Book` objects.

So,

we define:

```python
__add__()
```

Inside this method,

we return

```python
self.pages + other.pages
```

Output

```text
400
```

Now,

the expression

```python
book1 + book2
```

works because we overloaded the `+` operator for the `Book` class.



````python
# Example 5: Student Marks

class Student:

    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks


student1 = Student(85)
student2 = Student(90)

print("Total Marks:", student1 + student2)
````

# 📝 Explanation

Here,

each object stores a student's marks.

Normally,

the following statement would produce an error.

```python
student1 + student2
```

However,

after defining

```python
__add__()
```

Python understands how to add two Student objects.

Output

```text
Total Marks: 175
```

This is another practical example of **Operator Overloading**.



# 💡 Key Takeaways

✔ Operator Overloading allows the same operator to perform different operations.

✔ Python implements Operator Overloading using **Special (Magic/Dunder) Methods**.

✔ The `+` operator internally calls:

```python
__add__()
```

✔ Built-in types like integers, strings, and lists already support Operator Overloading.

✔ We can overload operators for our own custom classes.

✔ Operator Overloading makes objects behave naturally and improves code readability.



# ✨ Special (Magic / Dunder) Methods

Python provides many built-in methods that begin and end with **double underscores (`__`)**.

These are called:

- **Special Methods**
- **Magic Methods**
- **Dunder Methods** (Double Under)

Examples:

```python
__init__()

__str__()

__len__()

__add__()

__eq__()

__repr__()
```

These methods allow Python objects to work naturally with built-in functions and operators.



## 📌 Definition

> **Special (Magic/Dunder) Methods are predefined methods that allow Python classes to define how objects behave with built-in functions, operators, and language features.**



# 🤔 Why Do We Need Magic Methods?

Imagine creating your own class.

```python
class Book:
    pass
```

Now create an object.

```python
book = Book()
```

If we print it,

```python
print(book)
```

Python displays something like

```text
<__main__.Book object at 0x...>
```

This output is not user-friendly.

Instead,

we may want to display

```text
Book Name: Python Programming
```

This is possible using

```python
__str__()
```

Similarly,

operators like

```python
+
```

and built-in functions like

```python
len()
```

also work because of Magic Methods.



````python
# Example 1: __str__() Method

class Book:

    def __init__(self, title):
        self.title = title

    def __str__(self):
        return f"Book Title: {self.title}"


book = Book("Python Programming")

print(book)
````

# 📝 Explanation

Normally,

printing an object displays its memory address.

Example

```text
<__main__.Book object at ...>
```

After defining

```python
__str__()
```

Python automatically calls it when we use

```python
print(book)
```

Output

```text
Book Title: Python Programming
```

The `__str__()` method provides a readable and user-friendly string representation of an object.


````python
# Example 2: __len__() Method

class Playlist:

    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)


playlist = Playlist(["Song 1", "Song 2", "Song 3", "Song 4"])

print(len(playlist))
````

# 📝 Explanation

The built-in function

```python
len()
```

internally calls

```python
__len__()
```

Since our class defines

```python
__len__()
```

Python knows how to calculate the playlist length.

Output

```text
4
```

This allows custom objects to work naturally with the `len()` function.



````python
# Example 3: __add__() Method

class Money:

    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return self.amount + other.amount


wallet1 = Money(2500)
wallet2 = Money(3500)

print(wallet1 + wallet2)
````

# 📝 Explanation

The

```python
+
```

operator internally calls

```python
__add__()
```

Instead of producing an error,

our class now knows how to add two Money objects.

Output

```text
6000
```

This is another example of Operator Overloading using a Magic Method.


````python
# Example 4: __eq__() Method

class Student:

    def __init__(self, roll_no):
        self.roll_no = roll_no

    def __eq__(self, other):
        return self.roll_no == other.roll_no


student1 = Student(101)
student2 = Student(101)
student3 = Student(102)

print(student1 == student2)
print(student1 == student3)
````

# 📝 Explanation

The comparison operator

```python
==
```

internally calls

```python
__eq__()
```

We compare students using their roll numbers.

Output

```text
True

False
```

Without

```python
__eq__()
```

Python would compare the memory addresses of the objects instead of their actual data.


````python
# Example 5: __repr__() Method

class Laptop:

    def __init__(self, brand):
        self.brand = brand

    def __repr__(self):
        return f"Laptop('{self.brand}')"


laptop = Laptop("Dell")

print(repr(laptop))
````

# 📝 Explanation

The function

```python
repr()
```

internally calls

```python
__repr__()
```

The `__repr__()` method provides an official and developer-friendly representation of an object.

Output

```text
Laptop('Dell')
```

This representation is useful while debugging or inspecting objects.

Unlike `__str__()`, which is intended for end users, `__repr__()` is mainly intended for developers.


# 📊 Common Magic Methods

| Magic Method | Purpose | Example |
|--------------|---------|---------|
| `__init__()` | Initialize an object | Constructor |
| `__str__()` | User-friendly string representation | `print(obj)` |
| `__repr__()` | Developer-friendly representation | `repr(obj)` |
| `__len__()` | Return object length | `len(obj)` |
| `__add__()` | Overload the `+` operator | `obj1 + obj2` |
| `__eq__()` | Compare two objects | `obj1 == obj2` |



## 📝 Remember

✔ Magic Methods always begin and end with **double underscores**.

✔ Python automatically calls them when needed.

✔ They allow custom objects to behave like built-in Python objects.

✔ They are the foundation of Operator Overloading and many built-in Python features.



# 💡 Key Takeaways

✔ Magic Methods are predefined methods provided by Python.

✔ They define how custom objects behave with operators and built-in functions.

✔ `__str__()` controls what `print()` displays.

✔ `__len__()` works with `len()`.

✔ `__add__()` enables Operator Overloading.

✔ `__eq__()` customizes object comparison.

✔ `__repr__()` provides a developer-friendly representation.

✔ Magic Methods make custom classes more intuitive, readable, and Pythonic.



# ⚠️ Common Mistakes

While learning Polymorphism, beginners often make the following mistakes.



## ❌ Mistake 1: Confusing Polymorphism with Inheritance

Many beginners think Polymorphism and Inheritance are the same.

They are different concepts.

- **Inheritance** → Reuses code from another class.
- **Polymorphism** → Allows the same interface to behave differently.

Although they often work together, they solve different problems.



## ❌ Mistake 2: Giving Different Method Names

Incorrect

```python
class Dog:

    def bark(self):
        print("Bark")


class Cat:

    def meow(self):
        print("Meow")
```

Correct

```python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")
```

Using the same method name enables Polymorphism.



## ❌ Mistake 3: Forgetting the Required Method in Duck Typing

Incorrect

```python
class Car:
    pass

car = Car()

drive(car)
```

If `drive()` expects a method like

```python
start()
```

Python raises an

```text
AttributeError
```

Always make sure the object provides the required method.



## ❌ Mistake 4: Forgetting to Implement Magic Methods

Without methods like

```python
__str__()

__len__()

__add__()
```

custom objects cannot fully support Python's built-in functions and operators.



## ❌ Mistake 5: Overusing Operator Overloading

Operator Overloading should make code easier to understand.

Avoid changing the meaning of operators in confusing ways.

For example,

using `+` to perform subtraction would make the program difficult to understand and maintain.


# 💡 Best Practices

Follow these best practices while using Polymorphism.



## ✅ Use Meaningful Method Names

Use common method names across related classes.

Examples

```text
draw()

play()

send()

pay()

start()
```

This makes Polymorphism possible.



## ✅ Prefer Duck Typing in Python

Instead of checking object types,

focus on whether the object provides the required behavior.

Example

```python
obj.draw()
```

instead of

```python
isinstance(obj, Circle)
```



## ✅ Override Only When Necessary

Do not override Parent methods unless the Child Class really needs different behavior.



## ✅ Keep Operator Overloading Logical

The behavior of overloaded operators should match user expectations.

Example

```python
+
```

should represent combining or adding something.



## ✅ Implement Useful Magic Methods

For custom classes,

consider implementing methods like

```python
__str__()

__repr__()

__len__()

__eq__()
```

This makes objects more user-friendly and Pythonic.



## ✅ Write Reusable Code

Design methods so they work with multiple object types.

This improves flexibility and reduces duplication.
