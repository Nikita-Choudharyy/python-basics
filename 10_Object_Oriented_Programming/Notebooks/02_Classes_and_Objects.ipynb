# 🤔 What is a Class?

A **Class** is a blueprint, template, or design used to create objects.

It defines:

- 📋 Attributes (Data)
- ⚙️ Methods (Functions)

Think of a class as a plan that describes how an object should look and behave.

A class itself does not store actual data.

Instead, it defines the structure that every object created from it will follow.



### 📌 Definition

> **A Class is a blueprint for creating objects.**



# 🌍 Real-World Analogy of a Class

Imagine a car manufacturing company.

Before producing cars, engineers first create a blueprint.

The blueprint defines:

- 🚗 Shape
- 🎨 Color Options
- ⚙️ Engine Type
- 🛞 Number of Wheels

However, the blueprint itself is **not a real car**.

Using that blueprint, the company manufactures many cars.

Similarly,

- 🏗️ Class → Blueprint
- 📦 Object → Real Car

One class can be used to create multiple objects.



# 📦 What is an Object?

An **Object** is an actual instance of a class.

Objects contain real data and can perform actions defined inside the class.

Every object has its own identity and its own values.

For example,

Class:

👨‍🎓 Student

Objects:

- Rahul
- Priya
- Aman
- Neha

Each student has different information such as:

- Name
- Age
- Roll Number

Although all students belong to the same class, every object stores its own unique data.



### 📌 Definition

> **An Object is an instance of a class.**



# 📊 Relationship Between Class and Object

The relationship between a class and an object can be understood as follows:

```text
              Class                 
        (Blueprint / Template)      
                │                   
      ┌─────────┴─────────┐         
      │         │         │         
   Object1   Object2   Object3      
```

### Example

Class:

🏦 Bank Account

Objects:

- Nikita's Account
- Rahul's Account
- Priya's Account

Every account is created from the same class, but each account stores different information such as:

- Account Holder Name
- Account Number
- Balance

This is how classes and objects work together in Object-Oriented Programming.



# 🐍 Creating Your First Class

In Python, we use the **`class`** keyword to create a class.

### Basic Syntax

```python
class ClassName:
    # Class Body
```

### Syntax Breakdown

- **class** → Keyword used to define a class.
- **ClassName** → Name of the class.
- **:** → Indicates the beginning of the class body.
- **Indented Block** → Contains attributes and methods of the class.



### 📌 Naming Convention

According to Python's **PEP 8 Style Guide**, class names should follow **PascalCase**.

✅ Correct

- Student
- BankAccount
- EmployeeDetails

❌ Incorrect

- student
- bank_account
- employee_details

Using meaningful class names makes your code easier to read and maintain.



````python
# Creating a simple class

class Student:
    pass
````

# 📝 Explanation

Let's understand the above code step by step.

```python
class Student:
```

This line creates a class named **Student**.

The class currently has no attributes or methods.



```python
pass
```

The **`pass`** keyword is a placeholder.

Python does not allow an empty class body, so `pass` is used to indicate that the implementation will be added later.

At this stage, we have successfully created our first class.



# 📦 Creating an Object

Creating a class only defines the blueprint.

To use that blueprint, we need to create an **Object**.

The process of creating an object is called **Instantiation**.

### Syntax

```python
object_name = ClassName()
```

Here,

- **object_name** → Variable that stores the object.
- **ClassName()** → Calls the class and creates a new object.

Every time we call the class using parentheses `()`, Python creates a new object.



````python
# Creating a class

class Student:
    pass

# Creating an object

student1 = Student()

print(student1)
````

# 📝 Output Explanation

The output will look similar to:

```python
<__main__.Student object at 0x000001A3B5D6E8F0>
```

Don't worry if this output looks confusing.

It simply means:

- ✅ Python has successfully created a **Student** object.
- ✅ The object exists in memory.
- ✅ The hexadecimal value (`0x...`) represents the memory address of that object.

The memory address will be different every time you run the program.

This is completely normal.



# 👥 Creating Multiple Objects

One of the biggest advantages of a class is that we can create multiple objects from the same blueprint.

For example,

Class:

👨‍🎓 Student

Objects:

- Rahul
- Priya
- Aman
- Neha

Although all these students belong to the same class, each student represents a different object.

Each object stores its own information independently.

This makes Object-Oriented Programming highly reusable and scalable.



````python
# Creating a class

class Student:
    pass

# Creating multiple objects

student1 = Student()
student2 = Student()
student3 = Student()

print(student1)
print(student2)
print(student3)
````

# 💡 Observation

Notice that all three objects have different memory addresses.

Example:

```python
<__main__.Student object at 0x...>

<__main__.Student object at 0x...>

<__main__.Student object at 0x...>
```

Although all three objects belong to the same **Student** class, they are completely different objects.

Each object occupies its own memory location.

This is why every object can store different data independently.



# 🔍 Checking the Type of an Object

Sometimes we want to know which class an object belongs to.

Python provides the built-in **`type()`** function for this purpose.

### Syntax

```python
type(object_name)
```

The `type()` function returns the class (type) of the object.

It is very useful while debugging programs and understanding the type of variables and objects.



````python
# Example 1: Checking the type of an object

class Student:
    pass

student1 = Student()

print(type(student1))
````

````python
# Example 2: Checking the type of different objects

class Car:
    pass

class Laptop:
    pass

car = Car()
laptop = Laptop()

print(type(car))
print(type(laptop))
````

# 📝 Explanation

The `type()` function tells us the class of an object.

For Example:

```python
student1 = Student()
```

Here,

```python
type(student1)
```

returns

```python
<class '__main__.Student'>
```

Similarly,

```python
type(car)
```

returns

```python
<class '__main__.Car'>
```

This confirms that every object remembers the class from which it was created.



# 🆔 Checking the Identity of an Object

Every object created in Python has its own unique identity.

Python provides the built-in **`id()`** function to check this identity.

### Syntax

```python
id(object_name)
```

The `id()` function returns a unique integer that represents the object's identity in memory.

Even if two objects belong to the same class, their IDs are different because they are stored separately in memory.



````python
# Example 1: Checking object identity

class Student:
    pass

student1 = Student()

print(id(student1))
````

````python
# Example 2: Comparing object identities

class Student:
    pass

student1 = Student()
student2 = Student()

print(id(student1))
print(id(student2))
````

````python
# Example 3: Multiple objects have unique identities

class Mobile:
    pass

mobile1 = Mobile()
mobile2 = Mobile()
mobile3 = Mobile()

print(id(mobile1))
print(id(mobile2))
print(id(mobile3))
````

# 💡 Observation

Notice that every object has a different identity.

Example:

```text
Student Object 1 → 2872623232208

Student Object 2 → 2872623231536

Student Object 3 → 2872621601872
```

The actual numbers may vary every time you run the program.

This happens because each object occupies a different location in memory.

Even though all objects belong to the same class, they are completely independent of each other.

# ⚖️ Class vs Object

Many beginners get confused between **Class** and **Object**.

Remember this simple rule:

> **A Class is a blueprint, while an Object is the real thing created from that blueprint.**

Let's understand this with an easy example.

Imagine a company manufactures laptops.

Before making laptops, engineers first design a blueprint.

That blueprint describes:

- 💻 Brand
- 💾 RAM
- 💽 Storage
- 🖥️ Screen Size

This blueprint is called a **Class**.

Using that blueprint, many laptops are manufactured.

Each laptop is an **Object**.

Although all laptops are created from the same blueprint, every laptop has its own identity and data.



# 📊 Difference Between Class and Object

| Class 🏗️ | Object 📦 |
|-----------|------------|
| A blueprint or template. | A real instance of a class. |
| Defines structure. | Stores actual data. |
| Created only once. | Can be created many times. |
| Does not occupy memory like objects. | Occupies memory when created. |
| Used to create objects. | Used to access data and methods. |



````python
# Example 1: Class and Object

class Car:
    pass

car1 = Car()

print(type(car1))
````

# 📝 Explanation

In the above example,

```python
class Car:
```

creates a blueprint named **Car**.

Later,

```python
car1 = Car()
```

creates a real object from that blueprint.

So,

- **Car** → Class
- **car1** → Object

The object is created using the class.


````python
# Example 2: One class, multiple objects

class Car:
    pass

car1 = Car()
car2 = Car()
car3 = Car()

print(type(car1))
print(type(car2))
print(type(car3))
````

````python
# Example 3: Student class

class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
````

````python
# Example 4: Bank Account class

class BankAccount:
    pass

account1 = BankAccount()
account2 = BankAccount()

print(type(account1))
print(type(account2))
````

# ⚠️ Common Mistakes

While learning Classes and Objects, beginners often make the following mistakes.

## ❌ Mistake 1: Forgetting Parentheses While Creating an Object

Incorrect

```python
class Student:
    pass

student = Student
```

Correct

```python
class Student:
    pass

student = Student()
```

Without `()`, Python stores the class itself instead of creating an object.



## ❌ Mistake 2: Using Incorrect Indentation

Incorrect

```python
class Student:
pass
```

Correct

```python
class Student:
    pass
```

Python uses indentation to define code blocks.



## ❌ Mistake 3: Confusing Class and Object

```python
class Student:
    pass

student1 = Student()
```

Here,

- `Student` → Class
- `student1` → Object

Many beginners think both are the same, but they have different purposes.



## ❌ Mistake 4: Expecting Different Classes for Every Object

Incorrect Thinking

```
Rahul → Student Class
Priya → Another Student Class
```

Correct Thinking

```
Student Class

│
├── Rahul
├── Priya
├── Aman
└── Neha
```

One class can create any number of objects.



## ❌ Mistake 5: Forgetting Naming Conventions

Avoid using names like:

```python
class student:
```

Prefer:

```python
class Student:
```

According to Python's PEP 8 Style Guide, class names should follow **PascalCase**.

# 💡 Best Practices

Follow these best practices while working with Classes and Objects.

## ✅ Use Meaningful Class Names

Good Examples

- Student
- Employee
- BankAccount
- Car
- Laptop

Avoid names like:

- A
- Test
- Demo1



## ✅ Follow PascalCase

Correct

```python
class StudentDetails:
```

Incorrect

```python

```



## ✅ Keep One Class for One Responsibility

A class should represent only one real-world entity.

Example:

- Student
- Employee
- Product

Instead of creating one huge class for everything.



## ✅ Create Multiple Objects Instead of Multiple Classes

Incorrect

```
Student1 Class
Student2 Class
Student3 Class
```

Correct

```
Student Class

↓

student1
student2
student3
```



## ✅ Write Clean and Readable Code

- Use meaningful variable names.
- Follow proper indentation.
- Add comments whenever necessary.
- Keep your code simple and organized.

Writing clean code makes your programs easier to understand, debug, and maintain.
