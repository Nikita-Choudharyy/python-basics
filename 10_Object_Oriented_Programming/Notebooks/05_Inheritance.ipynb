# 🌳 What is Inheritance?

Inheritance is one of the four fundamental principles of Object-Oriented Programming (OOP).

It allows one class to acquire the attributes and methods of another class.

The existing class is called the **Parent Class** (or **Base Class**).

The new class is called the **Child Class** (or **Derived Class**).

Instead of creating everything from scratch, the child class automatically gets access to the parent class's features.



## 📌 Definition

> **Inheritance is the process by which one class acquires the properties and methods of another class, enabling code reusability and hierarchical relationships.**



# ✨ Advantages of Inheritance

Inheritance provides several important benefits.

### ♻️ Code Reusability

Write common code once and reuse it in multiple classes.



### 🧹 Reduced Code Duplication

Avoid writing the same attributes and methods repeatedly.



### 📈 Better Maintainability

Changes made in the parent class automatically become available to child classes.



### 🏗️ Better Code Organization

Related classes can be organized into a clear hierarchy.



### 🌍 Real-World Modeling

Inheritance represents natural relationships such as:

- Person → Student
- Vehicle → Car
- Animal → Dog
- Employee → Manager



### 🚀 Faster Development

Developers can build new classes quickly by extending existing ones instead of rewriting code.



# 👨 Parent Class and 👦 Child Class

Inheritance always involves at least two classes.

## 👨 Parent Class (Base Class)

A **Parent Class** is the class whose attributes and methods are inherited by another class.

It contains common properties and behaviors that can be reused.



## 👦 Child Class (Derived Class)

A **Child Class** is the class that inherits the properties and methods of the Parent Class.

It can:

- Use all inherited features.
- Add its own new features.
- Modify inherited behavior if needed.



### 📌 Relationship

```text
          Parent Class
               │
        Inherits
               │
               ▼
          Child Class
```

The child class automatically gets access to the parent class's public methods and attributes.



# 📝 Syntax of Inheritance

The syntax of Inheritance in Python is very simple.

```python
class Parent:
    pass

class Child(Parent):
    pass
```

### Syntax Breakdown

- **Parent** → Base Class
- **Child** → Derived Class
- **(Parent)** → Indicates that `Child` inherits from `Parent`

Whenever the child class is created, it automatically gets access to the parent class's public attributes and methods.


````python
# Example 1: Basic Inheritance

class Animal:

    def eat(self):
        print("Animals can eat.")

class Dog(Animal):
    pass

dog = Dog()

dog.eat()
````

# 📝 Explanation

Let's understand what happens.

### Step 1

```python
class Animal:
```

This creates the **Parent Class**.

It contains one method:

```python
eat()
```



### Step 2

```python
class Dog(Animal):
```

This creates the **Child Class**.

Notice that we didn't define the `eat()` method again.



### Step 3

```python
dog = Dog()
```

creates a Dog object.



### Step 4

```python
dog.eat()
```

Although `Dog` does not contain an `eat()` method,

Python searches in the Parent Class (`Animal`) and finds it there.

Output

```text
Animals can eat.
```

This is the power of Inheritance.



````python
# Example 2: Parent Class with Multiple Methods

class Vehicle:

    def start(self):
        print("Vehicle Started.")

    def stop(self):
        print("Vehicle Stopped.")

class Car(Vehicle):
    pass

car = Car()

car.start()
car.stop()
````

# 📝 Explanation

The Parent Class (`Vehicle`) contains two methods.

```
start()

stop()
```

The Child Class (`Car`) does not define any methods.

Yet,

```python
car.start()

car.stop()
```

work perfectly.

Output

```text
Vehicle Started.

Vehicle Stopped.
```

The child class automatically inherits all public methods from the parent class.

# 🏗️ Adding New Methods in a Child Class

A Child Class is **not limited** to only using the Parent Class's methods.

It can also define its **own attributes and methods**.

This means a Child Class can:

- ✅ Use all inherited features from the Parent Class.
- ✅ Add completely new features.
- ✅ Extend the functionality of the Parent Class.

This makes Inheritance flexible and powerful.



### 📌 Remember

A Child Class gets **everything from the Parent Class**, but it is also free to create **its own features**.


````python
# Example 1: Child Class Adding a New Method

class Animal:

    def eat(self):
        print("Animals can eat.")

class Dog(Animal):

    def bark(self):
        print("Dog is barking.")

dog = Dog()

dog.eat()
dog.bark()
````

Output

```text

Dog is barking.
```

This is one of the biggest advantages of Inheritance.

The child class can **reuse existing code** and **add new functionality**.

# 🔄 Method Overriding

Sometimes, a Child Class wants to change the behavior of a method inherited from the Parent Class.

Instead of using the Parent Class's version of the method, the Child Class creates its own version with the **same method name**.

This concept is called **Method Overriding**.



## 📌 Definition

> **Method Overriding is the process in which a Child Class provides its own implementation of a method that already exists in the Parent Class.**

When both classes have a method with the same name,

Python always gives priority to the **Child Class's method**.


# 🤔 Why Do We Need Method Overriding?

Imagine every vehicle can start.

```
Vehicle

↓

start()
```

But different vehicles start differently.

🚗 Car

```
Press the Start Button
```

🏍️ Bike

```
Turn the Key
```

✈️ Airplane

```
Start the Engines
```

Although every vehicle performs the same action (**start**),

the implementation is different.

Instead of creating completely different class structures,

Inheritance allows us to **override** the parent's method.



````python
# Example 1: Basic Method Overriding

class Animal:

    def sound(self):
        print("Animals make different sounds.")

class Dog(Animal):

    def sound(self):
        print("Dog barks.")

dog = Dog()

dog.sound()
````

# 📝 Explanation

The Parent Class contains:

```python

```

The Child Class also contains:

```python

```

Both methods have the **same name**.

When we execute:

```python
dog.sound()
```

Python first looks inside the **Dog** class.

Since it finds the method there,

it does **not** execute the Parent Class's version.

Output

```text
Dog barks.
```

This is called **Method Overriding**.



````python
# Example 2: Vehicle Example

class Vehicle:

    def start(self):
        print("Vehicle Started.")

class Car(Vehicle):

    def start(self):
        print("Car Started using Push Button.")

car = Car()

car.start()
````

# 📝 Explanation

```text

```

Python always executes the Child Class's method when it exists.

# ⭐ The `super()` Function

The `super()` function is used to access the **Parent Class** from the **Child Class**.

Using `super()`, we can:

- ✅ Call the Parent Class constructor.
- ✅ Call Parent Class methods.
- ✅ Reuse Parent Class functionality without rewriting code.



## 📌 Definition

> **`super()` is a built-in Python function that allows a Child Class to access the methods and constructor of its Parent Class.**

It is one of the most commonly used functions while working with Inheritance.



# 🤔 Why Do We Need `super()`?

Suppose we have a **Person** class.

Every person has:

- 👤 Name
- 🎂 Age

Now we create a **Student** class.

A student also needs:

- Name
- Age
- 📚 Course

Without `super()`,

we would have to write the same initialization code again.

```python
self.name = name
self.age = age
```

This creates duplicate code.

Using `super()`,

the Child Class can directly reuse the Parent Class constructor.

This keeps the program cleaner and easier to maintain.



````python
# Example 1: Calling the Parent Constructor

class Person:

    def __init__(self):
        print("Person Constructor Called")

class Student(Person):

    def __init__(self):
        super().__init__()
        print("Student Constructor Called")

student = Student()
````

# 📝 Explanation

Let's understand the execution flow.

When we create the object:

```python
student = Student()
```

Python first enters:

```python
Student.__init__()
```

Inside the constructor,

```python
super().__init__()
```

calls the Parent Class constructor.

Execution order:

```text
Person Constructor Called

↓

Student Constructor Called
```

Output

```text
Person Constructor Called
Student Constructor Called
```

Without `super()`, the Parent Class constructor would not execute automatically when the Child Class defines its own constructor.



````python
# Example 2: Reusing Parent Attributes

class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):

    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.course = course

student = Student("Rahul", 20, "Python")

print(student.name)
print(student.age)
print(student.course)
````

# 📝 Explanation

The Parent Class is responsible for storing:

- Name
- Age

The Child Class is responsible for storing:

- Course

Instead of writing:

```python
self.name = name
self.age = age
```

again,

we simply write:

```python
super().__init__(name, age)
```

Now the object contains:

```
Student Object

│

├── Name = Rahul

├── Age = 20

└── Course = Python
```

Output

```text
Rahul
20
Python
```

This reduces duplicate code and makes the program easier to maintain.



````python
# Example 3: Calling a Parent Method

class Animal:

    def eat(self):
        print("Animals can eat.")

class Dog(Animal):

    def eat(self):
        super().eat()
        print("Dog eats dog food.")

dog = Dog()

dog.eat()
````

# 📝 Explanation

Here,

Both Parent and Child classes contain the same method.

```
eat()
```

Inside the Child Class,

```python
super().eat()
```

executes the Parent Class version first.

After that,

the Child Class continues executing its own code.

Output

```text
Animals can eat.
Dog eats dog food.
```

This allows us to extend the Parent Class behavior instead of completely replacing it.



````python
# Example 4: Employee Example

class Employee:

    def login(self):
        print("Employee Logged In")

class Manager(Employee):

    def login(self):
        super().login()
        print("Manager Dashboard Opened")

manager = Manager()

manager.login()
````

# 📝 Explanation

The Manager class overrides the `login()` method.

However,

instead of completely replacing the Parent Class behavior,

it first executes:

```python
super().login()
```

and then adds its own functionality.

Output

```text
Employee Logged In
Manager Dashboard Opened
```

This is very common in real-world applications where we want to preserve the original behavior and then extend it.


````python
# Example 5: Banking Example

class BankAccount:

    def __init__(self, account_holder):
        self.account_holder = account_holder

class SavingsAccount(BankAccount):

    def __init__(self, account_holder, balance):
        super().__init__(account_holder)
        self.balance = balance

account = SavingsAccount("Rahul", 25000)

print("Account Holder:", account.account_holder)
print("Balance       :", account.balance)
````

# 📝 Explanation

The Parent Class stores the common information.

```
Account Holder
```

The Child Class stores the specialized information.

```
Balance
```

Using

```python
super().__init__(account_holder)
```

the Child Class reuses the Parent Class constructor instead of rewriting the same code.

Output

```text
Account Holder: Rahul
Balance       : 25000
```

This demonstrates how `super()` helps us write clean, reusable, and maintainable code.


# 💡 Key Takeaways

✔ `super()` is used inside a Child Class to access the Parent Class.

✔ It can call the Parent Class constructor using:

```python
super().__init__()
```

✔ It can call Parent Class methods using:

```python
super().method_name()
```

✔ `super()` reduces duplicate code and improves maintainability.

✔ It is widely used in real-world Python projects and frameworks.

✔ `super()` is especially useful when the Child Class wants to extend the Parent Class instead of replacing its functionality.


# 🌳 Types of Inheritance

Inheritance can be implemented in different ways depending on the relationship between classes.

Python supports the following types of Inheritance.

## Types of Inheritance

1. 🌱 Single Inheritance
2. 🌿 Multiple Inheritance
3. 🌳 Multilevel Inheritance
4. 🌲 Hierarchical Inheritance
5. 💎 Hybrid Inheritance (Combination of Multiple Types)

Each type solves a different real-world problem.

Let's understand them one by one with practical examples.


# 🌱 1. Single Inheritance

Single Inheritance is the simplest form of Inheritance.

In this type,

one Child Class inherits from one Parent Class.

### Structure

```text
Parent Class
      │
      ▼
Child Class
```

This is the most commonly used type of Inheritance.


````python
# Example: Single Inheritance

class Animal:

    def eat(self):
        print("Animals can eat.")

class Dog(Animal):

    def bark(self):
        print("Dog is barking.")

dog = Dog()

dog.eat()
dog.bark()
````

# 📝 Explanation

Relationship

```text
Animal
   │
   ▼
Dog
```

The Dog class inherits the `eat()` method from Animal.

It also defines its own method:

```python
bark()
```

Output

```text
Animals can eat.
Dog is barking.
```

This is called **Single Inheritance** because one child class inherits from one parent class.


# 🌿 2. Multiple Inheritance

In Multiple Inheritance,

one Child Class inherits from **more than one Parent Class**.

### Structure

```text
Parent 1      Parent 2

     \        /

      \      /

       ▼    ▼

      Child Class
```

The Child Class can access features from both Parent Classes.



````python
# Example: Multiple Inheritance

class Father:

    def skills(self):
        print("Programming Skills")

class Mother:

    def hobby(self):
        print("Painting")

class Child(Father, Mother):
    pass

child = Child()

child.skills()
child.hobby()
````

# 📝 Explanation

Relationship

```text
Father      Mother

    \        /

     \      /

      Child
```

The Child class inherits from both Parent Classes.

Output

```text
Programming Skills
Painting
```

This allows one class to reuse functionality from multiple classes.


# 🌳 3. Multilevel Inheritance

In Multilevel Inheritance,

one Child Class becomes the Parent Class for another class.

### Structure

```text
Grandparent

      │

      ▼

 Parent

      │

      ▼

 Child
```

Each level inherits the features of the previous level.


````python
# Example: Multilevel Inheritance

class Animal:

    def eat(self):
        print("Animals can eat.")

class Dog(Animal):

    def bark(self):
        print("Dog barks.")

class Puppy(Dog):

    def weep(self):
        print("Puppy is weeping.")

puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.weep()
````

# 📝 Explanation

Relationship

```text
Animal

   │

   ▼

Dog

   │

   ▼

Puppy
```

The Puppy class inherits:

From Animal

```python
eat()
```

From Dog

```python
bark()
```

Its own method

```python
weep()
```

Output

```text
Animals can eat.
Dog barks.
Puppy is weeping.
```

This is called **Multilevel Inheritance**.


# 🌲 4. Hierarchical Inheritance

In Hierarchical Inheritance,

multiple Child Classes inherit from the same Parent Class.

### Structure

```text
        Parent

      /    |    \

     ▼     ▼     ▼

 Child1 Child2 Child3
```

All child classes share the common features of the Parent Class.


````python
# Example: Hierarchical Inheritance

class Person:

    def introduce(self):
        print("I am a Person.")

class Student(Person):
    pass

class Teacher(Person):
    pass

student = Student()
teacher = Teacher()

student.introduce()
teacher.introduce()
````

# 📝 Explanation

Relationship

```text
         Person

       /        \

      ▼          ▼

 Student      Teacher
```

Both Student and Teacher inherit the same method.

```python
introduce()
```

Output

```text
I am a Person.
I am a Person.
```

This avoids writing the same code in multiple classes.


# 💎 5. Hybrid Inheritance

Hybrid Inheritance is a **combination of two or more inheritance types**.

For example,

it may combine:

- Single Inheritance
- Multiple Inheritance
- Multilevel Inheritance

### Structure

```text
        A

      /   \

     B     C

      \   /

        D
```

Hybrid Inheritance is possible in Python because Python supports Multiple Inheritance.

However, it can make programs more complex if not designed carefully.

For beginners, it is enough to understand the concept.

As you gain more experience, you'll encounter Hybrid Inheritance in larger applications and frameworks.


````python
# Example: Hybrid Inheritance

class A:

    def display_a(self):
        print("This is Class A")


class B(A):

    def display_b(self):
        print("This is Class B")


class C(A):

    def display_c(self):
        print("This is Class C")


class D(B, C):

    def display_d(self):
        print("This is Class D")


obj = D()

obj.display_a()
obj.display_b()
obj.display_c()
obj.display_d()
````

# 📝 Explanation

Let's understand the class hierarchy.

```text
          A
         / \
        /   \
       B     C
        \   /
         \ /
          D
```

### Step 1

`A` is the **Parent Class**.

It contains:

```python
display_a()
```



### Step 2

`B` inherits from `A`.

So,

`B` gets:

- `display_a()`
- `display_b()`



### Step 3

`C` also inherits from `A`.

So,

`C` gets:

- `display_a()`
- `display_c()`



### Step 4

Finally,

`D` inherits from both:

```python
B
```

and

```python
C
```

Therefore,

`D` can access all four methods.

```
display_a()

display_b()

display_c()

display_d()
```

Output

```text
This is Class A
This is Class B
This is Class C
This is Class D
```

### 📌 Why is this called Hybrid Inheritance?

Because this example combines **two different inheritance types**.

- **Hierarchical Inheritance**

```text
      A
     / \
    B   C
```

Here, one Parent Class (`A`) has multiple Child Classes (`B` and `C`).



- **Multiple Inheritance**

```text
   B     C
    \   /
     \ /
      D
```

Here, one Child Class (`D`) inherits from multiple Parent Classes (`B` and `C`).

Since this example combines **Hierarchical Inheritance** and **Multiple Inheritance**, it is called **Hybrid Inheritance**.


# 📊 Comparison of Inheritance Types

| Type | Description | Structure |
|------|-------------|-----------|
| Single | One Parent → One Child | `A → B` |
| Multiple | Multiple Parents → One Child | `A, B → C` |
| Multilevel | Parent → Child → Grandchild | `A → B → C` |
| Hierarchical | One Parent → Multiple Children | `A → B, C, D` |
| Hybrid | Combination of Multiple Types | Mixed Structure |



## 📝 Remember

✔ **Single** → One Parent, One Child.

✔ **Multiple** → Many Parents, One Child.

✔ **Multilevel** → Parent → Child → Grandchild.

✔ **Hierarchical** → One Parent, Many Children.

✔ **Hybrid** → Combination of different inheritance types.


# ⚠️ Common Mistakes

While learning Inheritance, beginners often make the following mistakes.



## ❌ Mistake 1: Forgetting to Inherit from the Parent Class

Incorrect

```python
class Animal:
    pass

class Dog:
    pass
```

Here, `Dog` is an independent class.

Correct

```python
class Animal:
    pass

class Dog(Animal):
    pass
```

Always mention the Parent Class inside parentheses.



## ❌ Mistake 2: Rewriting Existing Code

Incorrect

```python
class Animal:

    def eat(self):
        print("Eating")


class Dog:

    def eat(self):
        print("Eating")
```

Correct

```python
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):
    pass
```

Reuse existing methods instead of writing duplicate code.



## ❌ Mistake 3: Forgetting to Use `super()`

Incorrect

```python
class Student(Person):

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

Correct

```python
class Student(Person):

    def __init__(self, name, age):
        super().__init__(name, age)
```

Use `super()` whenever you need to reuse the Parent Class constructor or methods.



## ❌ Mistake 4: Overriding Without Understanding the Parent Method

If you completely replace the Parent Class method,

you may accidentally remove useful functionality.

Sometimes,

calling

```python
super().method_name()
```

before adding new functionality is the better approach.



## ❌ Mistake 5: Choosing the Wrong Inheritance Type

Not every problem requires Multiple or Hybrid Inheritance.

Whenever possible,

prefer **Single Inheritance** because it is simpler and easier to maintain.


# 💡 Best Practices

Follow these best practices while using Inheritance.



## ✅ Keep Parent Classes Generic

The Parent Class should contain only common features.

Example

```text
Person

Vehicle

Animal

Employee
```



## ✅ Keep Child Classes Specialized

Child Classes should add only features specific to themselves.

Example

```text
Student

Teacher

Car

Dog
```



## ✅ Prefer Code Reusability

Write common code once inside the Parent Class.

Avoid unnecessary duplication.



## ✅ Use `super()` Whenever Appropriate

Reuse Parent Class constructors and methods instead of copying code.



## ✅ Prefer Composition When Needed

If two classes do not have an **"is-a"** relationship,

Inheritance may not be the right choice.

Example

```
Car → Engine
```

A Car **has an** Engine.

It is **not an** Engine.

In such cases, Composition is a better design choice.



## ✅ Keep the Class Hierarchy Simple

Avoid creating very deep inheritance chains.

A simpler hierarchy is easier to understand, debug, and maintain.
