# 🧠 What is a Constructor?

A **Constructor** is a special method that is automatically executed whenever an object is created.

Its main purpose is to initialize the object's attributes with initial values.

In Python, the constructor is written using the special method:

```python
__init__()
```

Unlike normal methods, we do **not** call the constructor manually.

Python automatically calls it when we create an object.

### 📌 Definition

> **A Constructor is a special method that automatically initializes an object when it is created.**



# ✨ Features of Constructors

Some important characteristics of constructors are:

- ✅ Automatically called when an object is created.
- ✅ Used to initialize object attributes.
- ✅ Every object gets its own data.
- ✅ Written using the special `__init__()` method.
- ✅ Makes objects ready for use immediately after creation.
- ✅ Improves code readability and maintainability.

Constructors are one of the most frequently used features in Object-Oriented Programming.



# 📝 Syntax of a Constructor

In Python, constructors are created using the special method **`__init__()`**.

### Basic Syntax

```python
class ClassName:

    def __init__(self):
        # Initialization Code
```

### Syntax Breakdown

- **class** → Used to create a class.
- **def** → Used to define a method.
- **__init__** → Special constructor method.
- **self** → Refers to the current object.
- Constructor body → Executes automatically whenever an object is created.

Unlike normal methods, we do not call the constructor manually.

Python automatically executes it when an object is created.



````python
# Example 1: Simple Constructor

class Student:

    def __init__(self):
        print("Constructor Called Successfully!")

student1 = Student()
````

# 📝 Explanation

Let's understand the above code step by step.

```python
class Student:
```

Creates a class named **Student**.



```python
def __init__(self):
```

Defines the constructor.

The name `__init__` is fixed and cannot be changed.


```python
print("Constructor Called Successfully!")
```

This statement is executed automatically whenever an object is created.



```python
student1 = Student()
```

As soon as Python creates the object, it automatically calls the constructor.

Therefore the output becomes:

```text
Constructor Called Successfully!
```

Notice that we never called `__init__()` manually.

Python did it automatically.



````python
# Example 2: Constructor in another class

class Car:

    def __init__(self):
        print("A new Car object has been created.")

car1 = Car()
````

# 📝 Explanation

Python performs two tasks automatically.

Output:

# 👤 Understanding the `self` Keyword

The `self` keyword is one of the most important concepts in Python's Object-Oriented Programming.

Many beginners think that `self` is a keyword like `if`, `for`, or `while`.

❌ It is **NOT** a Python keyword.

Instead, `self` is simply a **reference to the current object**.

Whenever an object calls a method, Python automatically passes that object as the first argument to the method.

By convention, we write this first parameter as **`self`**.



## 📌 Definition

> **`self` refers to the current object (instance) of the class.**

It allows the object to access its own data and methods.



# 🤔 Why Do We Need `self`?

Imagine there are three students.

👤 Rahul

👤 Priya

👤 Aman

All of them belong to the same **Student** class.

But each student has different information.

```
Student Class

        │
        ▼

 ┌─────────────┐
 │ Rahul       │
 └─────────────┘

 ┌─────────────┐
 │ Priya       │
 └─────────────┘

 ┌─────────────┐
 │ Aman        │
 └─────────────┘
```

When Rahul accesses his data,

Python must know:

👉 "Which student's data should I use?"

That's exactly what **`self`** tells Python.

`self` always points to the object that is currently using the method.



````python
# Example 1: Using self in a constructor

class Student:

    def __init__(self):
        print("Constructor Executed")

student1 = Student()
````

# 📝 Explanation

Let's understand what happens behind the scenes.

When Python executes:

```python
student1 = Student()
```

Internally, Python does something similar to this:

```python
Student.__init__(student1)
```

Notice that Python automatically sends the object (`student1`) as the first argument.

That first argument is received by:

```python
self
```

So,

```
self
      ↓
student1
```

This is why we never pass `self` manually while creating an object.

Python handles it automatically.



````python
# Example 2: Creating multiple objects

class Student:

    def __init__(self):
        print(self)

student1 = Student()
student2 = Student()
student3 = Student()
````

# 📝 Explanation

This example prints the value stored inside `self`.

Output may look like this:

```text
<__main__.Student object at 0x000001...>
<__main__.Student object at 0x000001...>
<__main__.Student object at 0x000001...>
```

Notice that every object has a different memory address.

```
student1  → self

student2  → self

student3  → self
```

Although the variable name is always `self`, it refers to the object that is currently calling the constructor.



````python
# Example 3: Comparing self and object

class Student:

    def __init__(self):
        print("Self :", self)

student = Student()

print("Object:", student)
````

# 📝 Explanation

You will notice that both values are exactly the same.

Example Output

```text
Self : <__main__.Student object at 0x...>

Object: <__main__.Student object at 0x...>
```

This proves an important point.

```
student
      │
      ▼
same object
      ▲
      │
self
```

`self` and the object variable point to the same object in memory.

The only difference is:

- `student` is used outside the class.
- `self` is used inside the class.


````python
# Example 4: self can have any name (Not Recommended)

class Student:

    def __init__(my_object):
        print(my_object)

student = Student()
````

# 📝 Explanation

This code works perfectly.

Why?

Because Python only cares about the **first parameter**.

It does not require the name to be `self`.

```python

```

```python

```

```python

```

```python

```

However...

### ⚠️ Best Practice

Always use:

```python

```

Why?

Because every Python developer follows this convention.

Using `self` makes your code easier to read and understand.



# 💡 Key Takeaways

✔ `self` is **not** a Python keyword.

✔ `self` refers to the current object (instance).

# 🏷️ Initializing Object Attributes

So far, our constructors were only printing a message.

Example:

```python
class Student:

    def __init__(self):
        print("Constructor Called")
```

Although the constructor runs successfully, the object still does not store any useful information.

For example, a student should have:

- 👤 Name
- 🎂 Age
- 🆔 Roll Number
- 📚 Course

Instead of just printing a message, we can use the constructor to initialize these values.

The data stored inside an object is called an **Attribute** (or **Instance Variable**).



## 📌 Definition

> **An Attribute (Instance Variable) is a variable that belongs to an object and stores information about that object.**

These attributes are usually created inside the constructor using the `self` keyword.



# 🤔 Why Do We Need Attributes?

Imagine a classroom with three students.

| Student | Age | Course |
|---------|-----|---------|
| Rahul | 20 | Python |
| Priya | 21 | Data Science |
| Aman | 19 | AI & ML |

Although all of them belong to the same **Student** class, their information is different.

So where should Python store this information?

The answer is:

Inside each object using **attributes**.

Each object stores its own independent data.

This is why two objects created from the same class can have completely different values.



````python
# Example 1: Initializing attributes

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Rahul", 20)

print(student1.name)
print(student1.age)
````

# 📝 Explanation

Let's understand every line carefully.

```python
def __init__(self, name, age):
```

The constructor now accepts two additional values:

- name
- age

When we create an object,

```python
student1 = Student("Rahul", 20)
```

Python internally does something similar to:

```python
Student.__init__(student1, "Rahul", 20)
```

So,

```
self  → student1

name  → Rahul

age   → 20
```

Now look at these lines.

```python
self.name = name
```

This stores the value of `name` inside the object.

Similarly,

```python
self.age = age
```

stores the age inside the object.

Now the object contains:

```
student1

│

├── name = Rahul

└── age = 20
```

Whenever we write:

```python
print(student1.name)
```

Python retrieves the value stored inside the object.

Output

```text
Rahul
20
```



````python
# Example 2: Multiple objects with different data

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Rahul", 20)
student2 = Student("Priya", 21)
student3 = Student("Aman", 19)

print(student1.name, student1.age)
print(student2.name, student2.age)
print(student3.name, student3.age)
````

# 📝 Explanation

Notice that all three objects belong to the same class.

```
Student Class

        │
        ▼

Rahul   (20)

Priya   (21)

Aman    (19)
```

Each object stores its own data independently.

Output

```text
Rahul 20
Priya 21
Aman 19
```

Even though all objects are created from the same blueprint (class), their attribute values are different.

This is one of the biggest advantages of Object-Oriented Programming.



````python
# Example 3: Car class

class Car:

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

car1 = Car("Toyota", "Fortuner")
car2 = Car("Tesla", "Model Y")

print(car1.brand, car1.model)
print(car2.brand, car2.model)
````

# 📝 Explanation

This example shows that constructors are useful for every type of object.

Each car stores its own information.

```
car1

├── brand = Toyota

└── model = Fortuner
```

```
car2

├── brand = Tesla

└── model = Model Y
```

Output

```text
Toyota Fortuner
Tesla Model Y
```

The constructor initializes the object's data as soon as the object is created.

This makes every object unique.



# 💡 Important Observation

Notice the following line carefully:

```python
self.name = name
```

Many beginners get confused here because both sides use the word **name**.

They are actually different.

| Left Side | Right Side |
|-----------|------------|
| `self.name` | `name` |
| Object Attribute | Constructor Parameter |
| Stored inside the object | Temporary value received by the constructor |

Think of it like this:

```
Temporary Value
      │
      ▼

name = "Rahul"

      │

      ▼

Stored Permanently

self.name = "Rahul"
```

The value comes through the constructor parameter and is then saved inside the object using `self`.

This is why we write:

```python
self.name = name
```

instead of just:

```python
name = name
```



# ⚙️ What are Methods?

A **Method** is a function that is defined inside a class.

Methods describe the **behavior** or **actions** that an object can perform.

For example,

Consider a **Car** object.

The car has some information:

- 🚗 Brand
- 🎨 Color
- ⛽ Fuel Type

These are **attributes**.

Now think about what a car can do.

- ▶️ Start
- ⏹️ Stop
- 🚗 Drive
- 🔊 Honk

These are **methods** because they represent actions.



## 📌 Definition

> **A Method is a function defined inside a class that performs actions using the object's data.**

Methods usually use the `self` keyword to access the object's attributes.



# 🤔 Why Do We Need Methods?

Imagine we create a Student object.

```text
Student

Name : Rahul

Age : 20

Course : Python
```

The object stores data successfully.

But...

Can it display its own details?

❌ No.

Can it update its own data?

❌ No.

Can it perform any task?

❌ No.

That's why we create **methods**.

Methods allow objects to perform useful operations using their own data.



````python
# Example 1: First Method

class Student:

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student Name:", self.name)

student1 = Student("Rahul")

student1.display()
````

# 📝 Explanation

Let's understand this code step by step.

### Step 1

```python
def __init__(self, name):
```

The constructor receives the student's name.



### Step 2

```python
self.name = name
```

The name is stored inside the object.



### Step 3

```python
def display(self):
```

This creates a method named `display()`.

Unlike the constructor, this method is **not called automatically**.



### Step 4

```python
student1.display()
```

Here we are manually calling the method.

Inside the method,

```python
self.name
```

refers to the name stored inside the current object.

Output

```text
Student Name: Rahul
```



````python
# Example 2: Multiple objects using the same method

class Student:

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Student:", self.name)

student1 = Student("Rahul")
student2 = Student("Priya")
student3 = Student("Aman")

student1.display()
student2.display()
student3.display()
````

# 📝 Explanation

All three objects use the same method.

```
Student Class

        │

        ▼

Rahul ---- display()

Priya ---- display()

Aman  ---- display()
```

Although the method is the same,

the value of `self` changes depending on which object calls it.

When

```python
student1.display()
```

`self` refers to **student1**.

When

```python
student2.display()
```

`self` refers to **student2**.

When

```python
student3.display()
```

`self` refers to **student3**.

Output

```text
Student: Rahul
Student: Priya
Student: Aman
```

This is why a single method can work for many different objects.



````python
# Example 3: Car Example

class Car:

    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def car_info(self):
        print(f"Brand : {self.brand}")
        print(f"Model : {self.model}")

car1 = Car("Toyota", "Fortuner")
car2 = Car("Tesla", "Model Y")

car1.car_info()

print("-" * 25)

car2.car_info()
````

# 📝 Explanation


The method

Notice that we didn't create two different methods.

The **same method** works for every object because `self` automatically refers to the object that called it.

This avoids code duplication and makes programs more organized.



````python
# Example 4: Bank Account Example

class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def account_details(self):
        print(f"Account Holder : {self.name}")
        print(f"Balance        : ₹{self.balance}")

account1 = BankAccount("Rahul", 25000)
account2 = BankAccount("Priya", 48000)

account1.account_details()

print("-" * 30)

account2.account_details()
````

`self` refers to **account1**.

When

```python
account2.account_details()
```

Even though the method is written only once, it works correctly for every account object.

# 🔄 Constructor vs Method

Beginners often get confused between **Constructors** and **Methods** because both are functions inside a class.

However, they serve different purposes.

A **Constructor** is used to initialize an object.

A **Method** is used to perform actions using the object's data.

Let's understand the difference with an example.



````python
# Constructor vs Method Example

class Student:

    def __init__(self, name):
        self.name = name
        print("Constructor Executed")

    def display(self):
        print("Student Name:", self.name)

student1 = Student("Rahul")

student1.display()
````

# 📝 Explanation

Let's see what happens step by step.

### Step 1

```python
student1 = Student("Rahul")
```

Python automatically calls the constructor.

Output

```text
Constructor Executed
```



### Step 2

The name is stored inside the object.

```
student1

│

└── name = Rahul
```



### Step 3

Now we manually call

```python
student1.display()
```

Output

```text
Student Name: Rahul
```

Notice the difference.

The constructor ran automatically.

The method ran only because we called it.



# 📊 Constructor vs Method

| Constructor | Method |
|-------------|--------|
| Special method | Normal function inside a class |
| Name is `__init__()` | Can have any valid name |
| Called automatically | Called manually |
| Initializes object data | Performs actions |
| Runs once when an object is created | Can run multiple times |
| Usually creates object attributes | Usually uses object attributes |

### 📌 Remember

Think of it like this:

🏗️ Constructor → Builds and prepares the object.

⚙️ Method → Makes the object perform work.



# 🧩 Multiple Methods in One Class

A class is not limited to only one method.

We can create multiple methods inside the same class.

Each method can perform a different task.

For example,

A Bank Account object can:

- 💰 Show Balance
- ➕ Deposit Money
- ➖ Withdraw Money

Each action can be written as a separate method.

This makes programs clean and organized.



````python
# Example: Multiple Methods

class Student:

    def __init__(self, name, course):
        self.name = name
        self.course = course

    def display_name(self):
        print("Name:", self.name)

    def display_course(self):
        print("Course:", self.course)

student1 = Student("Rahul", "Python")

student1.display_name()
student1.display_course()
````

# 📝 Explanation

This class contains two different methods.

```
display_name()

display_course()
```

Both methods use the same object.

```
student1

├── name = Rahul

└── course = Python
```

Output

```text
Name: Rahul
Course: Python
```

Instead of writing one large method, we divided the work into smaller methods.

This improves readability and makes the code easier to maintain.



# 📥 Methods with Parameters

Just like normal Python functions, methods can also accept parameters.

Parameters allow us to pass additional information to a method when it is called.

For example,

A Bank Account object may need to receive an amount for deposit.

Instead of writing a fixed value inside the method,

we can pass it as a parameter.



````python
# Example: Method with Parameters

class Calculator:

    def add(self, num1, num2):
        result = num1 + num2
        print("Addition:", result)

calc = Calculator()

calc.add(10, 20)
calc.add(50, 30)
calc.add(100, 200)
````

# 📝 Explanation

The method

```python
add(self, num1, num2)
```

accepts two parameters.

Whenever we call

```python
calc.add(10, 20)
```

Python internally does something similar to:

```python
Calculator.add(calc, 10, 20)
```

Here,

- `self` → `calc`
- `num1` → `10`
- `num2` → `20`

Output

```text
Addition: 30
Addition: 80
Addition: 300
```

This example shows that methods can receive different values every time they are called.



# 🌍 Real-World Example

````python
class BankAccount:

    def __init__(self, holder, balance):
        self.holder = holder
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"₹{amount} deposited successfully.")

    def show_balance(self):
        print(f"Current Balance: ₹{self.balance}")

account = BankAccount("Rahul", 10000)

account.show_balance()

account.deposit(5000)

account.show_balance()
````

# 📝 Explanation

This example combines everything we have learned so far.

### Constructor

Stores:

- Account Holder
- Initial Balance



### Method 1

```python
deposit()
```

Receives the deposit amount and updates the balance.



### Method 2

```python
show_balance()
```

Displays the latest account balance.

Output

```text
Current Balance: ₹10000

₹5000 deposited successfully.

Current Balance: ₹15000
```

This is a simple example of how constructors and methods work together to build real-world applications.

# ⚠️ Common Mistakes

While learning Constructors and Methods, beginners often make the following mistakes.


## ❌ Mistake 1: Forgetting the `self` Parameter

Incorrect

```python
class Student:

    def __init__():
        print("Hello")
```

Correct

```python
class Student:

    def __init__(self):
        print("Hello")
```

Every instance method, including the constructor, must have `self` as its first parameter.



## ❌ Mistake 2: Forgetting to Store Values Using `self`

Incorrect

```python
class Student:

    def __init__(self, name):
        name = name
```

Correct

```python
class Student:

    def __init__(self, name):
        self.name = name
```

Without `self`, the value is not stored inside the object.



## ❌ Mistake 3: Calling the Constructor Manually

Incorrect

```python
student = Student("Rahul")

student.__init__("Rahul")
```

Correct

```python
student = Student("Rahul")
```

Python automatically calls the constructor when an object is created.



## ❌ Mistake 4: Forgetting Parentheses While Calling Methods

Incorrect

```python
student.display
```

Correct

```python
student.display()
```

Methods execute only when parentheses are used.


## ❌ Mistake 5: Confusing Constructor and Method

Remember:

- Constructor initializes the object.
- Method performs actions using object data.



# 💡 Best Practices

Follow these best practices while working with Constructors and Methods.

## ✅ Use Meaningful Class Names

Good Examples

- Student
- Employee
- BankAccount
- Car



## ✅ Always Use `self`

Although another name can be used, always follow Python's standard convention.

```python
def __init__(self):
```



## ✅ Keep Constructors Simple

The constructor should mainly initialize object data.

Avoid writing unnecessary complex logic inside `__init__()`.



## ✅ Use Separate Methods for Separate Tasks

Instead of one large method,

create smaller methods.

Example

```text
display_details()

deposit()

withdraw()

calculate_total()
```

This improves readability.


## ✅ Follow Python Naming Conventions

Classes

```python
Student
BankAccount
```

Methods

```python
display_details()

show_balance()

deposit()
```

Use PascalCase for classes and snake_case for methods.
