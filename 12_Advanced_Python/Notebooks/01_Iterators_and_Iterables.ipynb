# 🔄 Iterators and Iterables in Python

# 📖 Introduction

Python provides powerful tools for working with collections of data efficiently.

Whenever you use a `for` loop to iterate over a list, tuple, string, dictionary, or any other collection, Python follows a built-in iteration mechanism behind the scenes.

This mechanism is based on two important concepts:

- **Iterables**
- **Iterators**

Although these terms sound similar, they have different purposes.

An **iterable** is an object that can be traversed, while an **iterator** is an object that keeps track of the current position during iteration and returns one element at a time.

Understanding these concepts helps you write more efficient Python programs and forms the foundation for advanced topics such as **Generators**, **Comprehensions**, and many built-in Python functions.

By the end of this notebook, you will understand how Python performs iteration internally and how to work with iterables and iterators effectively.

# 🤔 Why Do We Need Iterators?

Imagine you have a list containing one million numbers.

If Python tried to process all elements at once, it would require unnecessary memory and reduce efficiency.

Instead, Python processes one element at a time during iteration.

This is possible because of **Iterators**.

An iterator remembers its current position and returns the next element only when requested.

This approach provides several advantages:

- 💾 Better memory efficiency.
- ⚡ Faster processing of large datasets.
- 🔄 Sequential access to data.
- 📦 Foundation for lazy evaluation.

Without iterators, many of Python's built-in features such as `for` loops and generators would not work efficiently.

# 🌍 Real-World Analogy

Imagine a library with thousands of books.

There are two ways to read them.

### 📚 Method 1

Someone places **all the books** on your desk at once.

This requires a huge amount of space and becomes difficult to manage.



### 📖 Method 2

A librarian brings **one book at a time**.

After you finish reading one book, the librarian brings the next.

This approach uses less space and is much more organized.

Python Iterators work in the same way.

Instead of loading everything at once, they provide one item at a time whenever it is requested.

This makes iteration efficient, especially for large collections of data.

# 📌 What is an Iterable?

An **Iterable** is any Python object that can be traversed or iterated over one element at a time.

In simple words,

an iterable is an object whose elements can be accessed sequentially using a loop such as `for`.

Many built-in Python data types are iterable.

Examples include:

- 📋 List
- 📦 Tuple
- 🔤 String
- 📖 Dictionary
- 🎯 Set
- 📏 Range

These objects can be passed to the `iter()` function to create an iterator.



## 📌 Definition

> **An Iterable is an object that can return an iterator using the built-in `iter()` function.**

````python
# List

numbers = [10, 20, 30, 40]

for num in numbers:
    print(num)
````

````python
# String

text = "Python"

for char in text:
    print(char)
````

# 📝 Explanation

Both the list and the string are **iterable objects**.

Python automatically retrieves one element at a time during the `for` loop.

Behind the scenes, Python creates an iterator and repeatedly requests the next element until all elements have been processed.

# 📌 What is an Iterator?

An **Iterator** is an object that allows you to access the elements of an iterable **one at a time**.

Unlike an iterable, an iterator remembers its current position.

Every time you request the next value, the iterator moves forward.

Once all elements have been returned, the iterator raises the `StopIteration` exception.



## 📌 Definition

> **An Iterator is an object that produces one element at a time and keeps track of its current position during iteration.**

````python
numbers = [10, 20, 30, 40]

iterator = iter(numbers)

print(iterator)
````

# 📝 Explanation

The function

```python
iter(numbers)
```

creates an iterator from the list.

The variable

```python
iterator
```

does not store the list itself.

Instead, it stores an object that knows:

- Which iterable it belongs to.
- Which element should be returned next.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 📝 Explanation

The built-in `next()` function requests the next element from the iterator.

Each call to `next()` returns one value and automatically advances the iterator to the next position.

The iterator remembers where it stopped, so it does not start from the beginning each time.

# 🔄 How Python Uses Iterators Internally

When you write:

```python
numbers = [10, 20, 30]

for num in numbers:
    print(num)
```

Python internally performs operations similar to the following:

```python
iterator = iter(numbers)

while True:
    try:
        value = next(iterator)
        print(value)
    except StopIteration:
        break
```

The `for` loop automatically creates an iterator, repeatedly calls `next()`, and stops when the iterator raises the `StopIteration` exception.

# 💡 Did You Know?

Every `for` loop in Python works using the **Iterator Protocol**.

Even though you don't explicitly call `iter()` and `next()`, Python does it internally.

This is why `for` loops can iterate over many different types of objects, such as lists, tuples, strings, dictionaries, sets, files, generators, and many custom objects.

# 🔄 Iterable vs Iterator

Although the terms **Iterable** and **Iterator** sound similar, they represent different concepts in Python.

An **Iterable** is a collection of elements that can produce an iterator.

An **Iterator** is an object that retrieves those elements one by one while keeping track of its current position.

In simple words:

- 📦 Iterable → Collection of data.
- 🔄 Iterator → Object used to traverse the collection.

# 📊 Iterable vs Iterator

| Iterable | Iterator |
|-----------|----------|
| Stores a collection of data. | Traverses the collection one element at a time. |
| Can create an iterator using `iter()`. | Created from an iterable using `iter()`. |
| Does not remember the current position. | Remembers the current position. |
| Cannot be used directly with `next()`. | Can be used with `next()`. |
| Example: `list`, `tuple`, `set`, `dict`, `string`, `range` | Example: `list_iterator`, `tuple_iterator`, `str_iterator` |


````python
numbers = [10, 20, 30]

# Iterable
print(type(numbers))

# Iterator
iterator = iter(numbers)

print(type(iterator))
````

# 📝 Explanation

In this example,

```python
numbers
```

is an **Iterable** because it stores a collection of values.

The statement

```python
iterator = iter(numbers)
```

creates an **Iterator**.

The iterator is responsible for returning one element at a time during iteration.

# 📌 The `iter()` Function

The built-in `iter()` function converts an iterable into an iterator.

### Syntax

```python
iter(iterable)
```

The function accepts any iterable object and returns an iterator.

````python
fruits = ["Apple", "Banana", "Mango"]

fruit_iterator = iter(fruits)

print(fruit_iterator)
````

# 📝 Explanation

The `iter()` function does not return the original list.

Instead, it creates a new iterator object that can access the list one element at a time.

# 📌 The `next()` Function

The built-in `next()` function retrieves the next element from an iterator.

### Syntax

```python
next(iterator)
```

Each call to `next()` moves the iterator forward by one position.

````python
numbers = [100, 200, 300]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 📝 Explanation

The iterator remembers its current position.

```python
next(iterator)
```

returns:

- First call → `100`
- Second call → `200`
- Third call → `300`

The iterator never starts from the beginning unless a **new iterator** is created.

# ⚠️ StopIteration Exception

What happens if we request another element after all elements have been consumed?

Python raises a special exception called **`StopIteration`**.

````python
numbers = [10, 20]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

# 📝 Explanation

The iterator successfully returns:

- `10`
- `20`

After the last element,

there are no more values available.

Therefore,

```python
next(iterator)
```

raises the `StopIteration` exception.

This exception tells Python that the iteration has finished.

The `for` loop automatically catches this exception internally, which is why you normally don't see it while using a `for` loop.

````python
numbers = [10, 20, 30]

iterator = iter(numbers)

while True:
    try:
        print(next(iterator))
    except StopIteration:
        print("Iteration Completed")
        break
````

# 💡 Did You Know?

Every iterator is also an **iterable**.

This means you can pass an iterator to the `iter()` function again.

```python
iterator = iter([1, 2, 3])

print(iter(iterator) is iterator)
```

Output

```text
True
```

This happens because an iterator returns itself when passed to `iter()`.

# 🛠️ Creating a Custom Iterator

Python not only provides built-in iterators for objects like lists, tuples, and strings, but also allows developers to create their own custom iterators.

A custom iterator is useful when you want complete control over how data is generated or traversed.

To create a custom iterator, a class must follow the **Iterator Protocol**.

The Iterator Protocol requires two special methods:

- `__iter__()`
- `__next__()`

Together, these methods allow Python to iterate over custom objects using a `for` loop or the `next()` function.

# 📌 The Iterator Protocol

An object is considered an iterator if it implements the following methods.

## 1️⃣ `__iter__()`

Returns the iterator object itself.



## 2️⃣ `__next__()`

Returns the next value in the sequence.

When there are no more values available,

it must raise the

```python
StopIteration
```

exception.

These two methods together form the **Iterator Protocol** in Python.

# 📊 Iterator Protocol Flow

```text
Create Object
      │
      ▼
Call iter(object)
      │
      ▼
__iter__()
      │
      ▼
Iterator Object Returned
      │
      ▼
Call next(iterator)
      │
      ▼
__next__()
      │
      ▼
Return Next Value
      │
      ▼
More Values?
      │
 ┌────┴────┐
 │         │
Yes       No
 │         │
 ▼         ▼
Return   Raise
Value   StopIteration
```

````python
class CountUp:

    def __init__(self, limit):
        self.limit = limit
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):

        if self.current <= self.limit:
            number = self.current
            self.current += 1
            return number

        raise StopIteration
````

# 📝 Explanation

Let's understand each method.

### `__init__()`

Initializes the iterator.

```python
self.limit
```

stores the maximum value.

```python
self.current
```

stores the current position.

---

### `__iter__()`

```python
return self
```

returns the iterator object itself.

Whenever Python executes

```python
iter(object)
```

this method is called.



### `__next__()`

Each call to

```python
next(iterator)
```

executes this method.

If values are available,

the next value is returned.

Otherwise,

```python
StopIteration
```

is raised to indicate that the iteration has finished.

````python
counter = CountUp(5)

for number in counter:
    print(number)
````

# 📝 Explanation

When the `for` loop starts,

Python internally performs the following steps.

```python
iterator = iter(counter)
```

Python calls

```python
__iter__()
```

to obtain the iterator.

Next,

Python repeatedly executes

```python
next(iterator)
```

which calls

```python
__next__()
```

The iterator returns:

```text
1
2
3
4
5
```

Finally,

`StopIteration` is raised,

and the `for` loop terminates automatically.

````python
counter = CountUp(3)

print(next(counter))
print(next(counter))
print(next(counter))
````

# 📝 Explanation

The `next()` function can also be used directly with a custom iterator.

Each call returns one value.

After all values have been generated,

calling

```python
next(counter)
```

again would raise

```python
StopIteration
```

because no more values are available.

# ⚠️ Common Mistakes

While learning Iterators and Iterables, beginners often make the following mistakes.

## ❌ Mistake 1: Assuming Every Iterable is an Iterator

Incorrect

```python
numbers = [10, 20, 30]

next(numbers)
```

Output

```text
TypeError:
'list' object is not an iterator
```

**Correct**

```python
numbers = [10, 20, 30]

iterator = iter(numbers)

print(next(iterator))
```

Always convert an iterable into an iterator before using `next()`.

---

## ❌ Mistake 2: Forgetting `StopIteration`

```python
numbers = [1, 2]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

The third call raises:

```text
StopIteration
```

Always remember that iterators have a finite number of elements unless designed otherwise.

---

## ❌ Mistake 3: Forgetting `return self` in `__iter__()`

Incorrect

```python
def __iter__(self):
    pass
```

Correct

```python
def __iter__(self):
    return self
```

Without returning the iterator object, Python cannot iterate over your custom iterator.

---

## ❌ Mistake 4: Forgetting to Raise `StopIteration`

Every custom iterator must eventually stop.

Incorrect

```python
def __next__(self):
    return self.current
```

Correct

```python
def __next__(self):

    if condition:
        return value

    raise StopIteration
```

Without raising `StopIteration`, the iterator may never terminate correctly.

---

## ❌ Mistake 5: Confusing `iter()` and `next()`

Remember:

- `iter()` → Creates an iterator.
- `next()` → Retrieves the next element from an iterator.
