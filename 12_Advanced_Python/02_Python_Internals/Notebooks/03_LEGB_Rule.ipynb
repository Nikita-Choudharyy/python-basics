# 🔍 LEGB Rule in Python

# 📖 Introduction

The **LEGB Rule** describes the order Python follows when searching for a name.

LEGB stands for:

```text
L → Local
E → Enclosing
G → Global
B → Built-in
```

When Python encounters a name such as:

```python
print(name)
```

it needs to determine which `name` the code refers to.

Python searches through different scopes in a specific order.

The general lookup order is:

```text
Local
   ↓
Enclosing
   ↓
Global
   ↓
Built-in
```

Python stops searching when it finds the name.

This process is called **name resolution**.

# 🧠 What is Name Resolution?

**Name Resolution** is the process Python uses to determine which object a name refers to.

For example:

```python
name = "Nikita"

print(name)
```

Python needs to resolve:

```text
name
```

to the object:

```text
"Nikita"
```

When multiple scopes contain the same name, Python uses the **LEGB Rule** to determine which binding should be selected.

# 📊 LEGB Overview

The LEGB Rule follows this order:

```text
┌─────────────────┐
│      Local      │  ← 1st
└────────┬────────┘
         ↓
┌─────────────────┐
│    Enclosing    │  ← 2nd
└────────┬────────┘
         ↓
┌─────────────────┐
│     Global      │  ← 3rd
└────────┬────────┘
         ↓
┌─────────────────┐
│     Built-in    │  ← 4th
└─────────────────┘
```

Python searches from **inside to outside**.

As soon as Python finds the requested name, the search stops.

# 🟢 L — Local Scope

The first place Python looks is the **Local Scope**.

Consider:

```python
name = "Global"


def greet():

    name = "Local"

    print(name)
```

When:

```python
greet()
```

runs, Python first checks the Local Scope of `greet()`.

It finds:

```text
name → "Local"
```

Therefore, `"Local"` is printed.

````python
name = "Global"


def greet():

    name = "Local"

    print(name)


greet()
````

# 🔍 Step-by-Step Lookup

For:

```python
print(name)
```

inside `greet()`:

Python checks:

```text
1. Local      → name found ✅
```

Because the name was found locally, Python does **not** need to continue searching:

```text
Enclosing
Global
Built-in
```

So the result is:

```text
Local
```

# 🏠 E — Enclosing Scope

If Python cannot find a name in the Local Scope, it checks the **Enclosing Scope**.

This occurs with nested functions.

Example:

```python
def outer():

    name = "Outer"

    def inner():

        print(name)
```

The `inner()` function does not define `name` locally.

Python therefore checks the enclosing function:

```text
inner()
   ↓
outer()
```

It finds:

```text
name → "Outer"
```

````python
def outer():

    name = "Outer"

    def inner():

        print(name)

    inner()


outer()
````

# 🔍 Enclosing Lookup

For:

```python
print(name)
```

inside `inner()`:

Python checks:

```text
1. Local      → not found ❌
2. Enclosing  → found ✅
```

Therefore:

```text
Outer
```

is printed.

# 🌍 G — Global Scope

If Python cannot find a name in the Local or Enclosing Scope, it checks the **Global Scope**.

Example:

```python
name = "Global"


def greet():

    print(name)
```

There is no local `name`.

There is no enclosing function.

Python therefore searches the Global Scope and finds:

```text
name → "Global"
```

````python
name = "Global"


def greet():

    print(name)


greet()
````

# 🔍 Global Lookup

For:

```python
print(name)
```

Python checks:

```text
1. Local      → not found ❌
2. Enclosing  → not found ❌
3. Global     → found ✅
```

Therefore:

```text
Global
```

is printed.

# 🐍 B — Built-in Scope

If Python cannot find a name in Local, Enclosing, or Global Scope, it finally checks the **Built-in Scope**.

Python provides many built-in names, such as:

```python

min

range
```

For example:

```python

```

Python resolves:

```text

```

through the Built-in Scope.

````python
numbers = [10, 20, 30]

print(len(numbers))
````

# 🔍 Built-in Lookup

For:


Python searches:

```text
1. Local      → not found ❌
2. Enclosing  → not found ❌
3. Global     → not found ❌
4. Built-in   → found ✅
```

Therefore Python uses the built-in `len()` function.

# 📌 First Important Rule

The most important thing to remember is:

> **Python searches LEGB from the nearest scope outward.**

The order is:

```text

Enclosing
   ↓
Global
   ↓
Built-in
```

Python does not randomly choose between variables.

It follows a defined name-resolution process.

# 🎭 Name Shadowing

**Name Shadowing** occurs when a closer scope defines a name that also exists in an outer scope.

Example:

```python
name = "Global"

def show():

    name = "Local"

    print(name)
```

The Local `name` shadows the Global `name` inside `show()`.

The Global variable still exists.

It is simply not selected by the LEGB lookup while the Local binding is available.

````python
name = "Global"


def show():

    name = "Local"

    print("Inside :", name)


show()

print("Outside:", name)
````

# 🧩 Shadowing Across Three Levels

The same name can be defined at three different levels:

```python
value = "Global"

def outer():

    value = "Enclosing"

    def inner():

        value = "Local"

        print(value)
```

LEGB selects:

```text
Local → "Local"
```

because Local is the closest scope.

````python
value = "Global"


def outer():

    value = "Enclosing"

    def inner():

        value = "Local"

        print("Selected:", value)

    inner()


outer()
````

# 🔧 Recovering a Shadowed Built-in

If a built-in name has been shadowed globally, the built-in can still be accessed through the `builtins` module.


````python
import builtins

len = 100

print("Shadowed len:", len)

print(
    "Built-in len:",
    builtins.len([10, 20, 30])
)
````

# 🧠 Why Does `UnboundLocalError` Happen?

Consider:

```python
x = 10

def test():

    print(x)

    x = 20
```

It may look like Python should:

```text
Local → not found
Enclosing → not found
Global → x = 10
```

But the assignment:

```python
x = 20
```

causes Python to treat `x` as a **local binding** in that function.

Therefore the earlier:

```python
print(x)
```

tries to read the local `x` before it has been assigned.

Result:

```text
UnboundLocalError
```

The key idea is:

> Assignment inside a function can affect how Python determines the binding for that name throughout the function.

# 🌍 `global` and LEGB

The `global` keyword changes which binding an assignment refers to.

Example:

```python
x = 10

def change():

    global x

    x = 20
```

Without `global`, the assignment:

```python
x = 20
```

would create a Local binding.

With:

```python
global x
```

Python is told that `x` refers to the Global binding.

So:

```text
Local assignment
        ↓
Local x

global x
        ↓
Global x
```

````python
x = 10


def change():

    global x

    x = 20


print("Before:", x)

change()

print("After :", x)
````

# 🏠 `nonlocal` and LEGB

The `nonlocal` keyword is used when a nested function needs to modify a binding from an enclosing function scope.

Example:

```python
def outer():

    x = 10

    def inner():

        nonlocal x

        x = 20
```

Here:

```text
inner Local Scope
        ↓
nonlocal x
        ↓
Enclosing Scope
        ↓
x = 20
```

`nonlocal` does not refer directly to the Global Scope.

````python
def outer():

    x = 10

    def inner():

        nonlocal x

        x = 20

    inner()

    print("Outer x:", x)


outer()
````

# 🔍 LEGB with `global` and `nonlocal`

Remember:

```text
Normal name lookup:

Local
  ↓
Enclosing
  ↓
Global
  ↓
Built-in
```

But `global` and `nonlocal` affect **where assignments are bound**.

### `global`

```python
global x
```

means:

> The assignment to `x` refers to the module-level Global binding.

### `nonlocal`

```python
nonlocal x
```

means:

> The assignment to `x` refers to an existing binding in an enclosing function scope.

# ⚠️ `global` Must Be Declared Before Assignment

If a function needs to assign to a global name using `global`, the declaration must appear before the relevant use or assignment in that scope.

Correct:

```python
x = 10

def update():

    global x

    x = 20
```

The `global` declaration tells Python how `x` should be treated within that function.

````python
x = 10


def update():

    global x

    x = 20


update()

print(x)
````

# ⚠️ `nonlocal` Requires an Enclosing Binding

Consider:

```python
def outer():

    def inner():

        nonlocal x
```

This is invalid because there is no enclosing function binding for `x`.

`nonlocal` requires an existing binding in an enclosing function scope.

Example of a valid use:

```python
def outer():

    x = 10

    def inner():

        nonlocal x
```

````python
def outer():

    x = 10

    def inner():

        nonlocal x

        x += 1

    inner()

    print(x)


outer()
````

# 🗑️ `del` and Name Resolution

The `del` statement removes a name binding from a namespace.

Example:

```python

del x
```

After deleting the Global binding, attempting to access `x` can produce `NameError` if no other applicable binding exists.

`del` removes the binding; it does not simply assign a special "empty" value.

````python
temporary_value = 100

print("Before deletion:", temporary_value)

del temporary_value

# Uncomment to observe NameError:
# print(temporary_value)
````

# 🧠 `del` in Nested Scopes

In nested functions, `del` can remove a local binding.

However, if a name belongs to an enclosing scope, directly deleting that enclosing binding from the inner function is not the same as ordinary local deletion.

For learning purposes, remember:

```text
del name
    ↓
removes the binding of name
from the applicable namespace
```

Name resolution still follows LEGB when Python needs to find a name.

Now the Global name `greet` refers to the string instead of the function.

Calling:

```python
greet()
```

will therefore fail.

````python
def greet():

    print("Hello Python")


print("Before shadowing:")

greet()


greet = "Python"

print("After shadowing:")
print(greet)

# Uncomment to observe the error:
# greet()
````

# 🧠 LEGB and Lambda Functions

Lambda functions follow the same name-resolution rules.

Example:

```python
x = 10

multiply = lambda number: number * x
```

The lambda does not define `x` locally.

When it uses `x`, Python searches outward and can find the Global `x`.

````python
x = 10

multiply = lambda number: number * x

print(multiply(5))
````

# 🔍 Lambda with Local Shadowing

Consider:

```python
x = 10

def calculate():

    x = 20

    multiply = lambda number: number * x

    return multiply
```

The lambda does not have its own local `x`.

It therefore finds the `x` from the enclosing function scope.

This demonstrates that lambdas also participate in Python's name-resolution mechanism.

````python
x = 10


def calculate():

    x = 20

    multiply = lambda number: number * x

    return multiply


result = calculate()

print(result(5))
````

# 🔗 Closures and LEGB

Closures provide a practical example of Enclosing Scope.

Example:

```python
def create_multiplier(factor):

    def multiply(number):

        return number * factor

    return multiply
```

When `multiply()` searches for:

```python
factor
```

its lookup is:

```text
Local
   ↓
Enclosing → factor found
   ↓
Global
   ↓
Built-in
```

The search stops at the Enclosing Scope.

````python
def create_multiplier(factor):

    def multiply(number):

        return number * factor

    return multiply


double = create_multiplier(2)

print(double(10))
````

# 🧠 Closure State and Name Lookup

Consider:

```python
def create_counter():

    count = 0

    def increment():

        nonlocal count

        count += 1

        return count

    return increment
```

The name:

```python
count
```

belongs to the Enclosing Scope.

The `nonlocal` declaration tells Python that assignments to `count` should update that enclosing binding.

````python
def create_counter():

    count = 0

    def increment():

        nonlocal count

        count += 1

        return count

    return increment


counter = create_counter()

print(counter())
print(counter())
print(counter())
````

# 🔬 Inspecting Closure Variables

A Closure can be inspected using:

```python
function.__code__.co_freevars
```

This tells us which variables from an enclosing scope are referenced by the function.

We can also inspect:

```python
function.__closure__
```

which contains the closure cells holding the captured values.

````python
def create_multiplier(factor):

    def multiply(number):

        return number * factor

    return multiply


double = create_multiplier(2)

print(
    "Free Variables:",
    double.__code__.co_freevars
)

print(
    "Closure:",
    double.__closure__
)

if double.__closure__:

    for cell in double.__closure__:

        print(
            "Captured Value:",
            cell.cell_contents
        )
````

# 🎯 LEGB with a Closure

Consider:

```python
x = "Global"

def outer():

    x = "Enclosing"

    def inner():

        return x

    return inner
```

When `inner()` executes:

```python
return x
```

Python searches:

```text
Local
   ↓
Enclosing → FOUND
   ↓
Global
   ↓
Built-in
```

Therefore:

```text
"Enclosing"
```

is returned.

````python
x = "Global"


def outer():

    x = "Enclosing"

    def inner():

        return x

    return inner


function = outer()

print(function())
````

# 🔥 Tricky Example — Closure Value

Consider:

```python
def outer():

    x = 10

    def inner():

        return x

    x = 20

    return inner
```

What will:

```python
outer()()
```

return?

The inner function refers to the enclosing binding of `x`.

The binding contains:

```text
x → 20
```

when the returned function is later called.

Therefore the result is:

```text
20
```

````python
def outer():

    x = 10

    def inner():

        return x

    x = 20

    return inner


function = outer()

print(function())
````

# 🧠 Important Closure Insight

A Closure does not simply store an independent copy of every variable.

It retains access to the relevant enclosing binding through closure cells.

This is why changes to an enclosing variable using `nonlocal` can be observed by the inner function.

````python
def outer():

    value = 10

    def inner():

        return value

    def update():

        nonlocal value

        value = 50

    return inner, update


show_value, update_value = outer()

print("Before:", show_value())

update_value()

print("After :", show_value())
````

# ⚠️ Common Mistakes

### ❌ Mistake 1 — Memorizing LEGB without understanding lookup

Don't just remember:

```text
Local → Enclosing → Global → Built-in
```

Understand that Python searches from the nearest applicable scope outward.

---

### ❌ Mistake 2 — Thinking Global is always selected

A Global variable can be shadowed by a Local or Enclosing variable.

---

### ❌ Mistake 3 — Forgetting Built-in Scope

Names such as:

```python
len
sum
print
type
```

can come from Built-in Scope.

---

### ❌ Mistake 4 — Shadowing built-ins

Avoid:

```python
list = []
sum = 100
len = 10
```

---

### ❌ Mistake 5 — Confusing NameError with UnboundLocalError

`NameError` means Python cannot resolve the requested name.

`UnboundLocalError` commonly occurs when a name is treated as Local but is accessed before its Local binding has been assigned.

---

### ❌ Mistake 6 — Assuming `nonlocal` means Global

`nonlocal` refers to an existing binding in an enclosing function scope.

It does not directly refer to the Global Scope.

# ✅ Best Practices

Follow these practices when working with LEGB:

- ✔ Understand name lookup instead of only memorizing the acronym.
- ✔ Prefer clear and meaningful variable names.
- ✔ Avoid unnecessary name shadowing.
- ✔ Never intentionally shadow built-in functions unless there is a very specific reason.
- ✔ Keep Global variables to a minimum.
- ✔ Use `global` only when modification of module-level state is genuinely required.
- ✔ Use `nonlocal` when a Closure needs to modify an enclosing binding.
- ✔ Keep nested functions focused.
- ✔ Understand the difference between reading and assigning a name.
- ✔ Use debugging and tracing to understand unexpected name resolution.

- ✔ Global Scope is checked after Enclosing Scope.
- ✔ Built-in Scope is checked last.
