# 🧠 Namespaces in Python

# 📖 Introduction

In Python, a **Namespace** is a system that stores names and maps them to the objects they refer to.

Whenever we create a variable, function, class, or import a module, Python needs a way to keep track of these names.

A Namespace provides this organization.

For example:

```python
name = "Nikita"
age = 20
```

Here:

- `name` is a name.
- `age` is a name.
- `"Nikita"` is an object.
- `20` is an object.

Python maintains a mapping between these names and their corresponding objects.

Conceptually:

```text
name ───────► "Nikita"
age  ───────► 20
```

Namespaces help Python determine **which object a particular name refers to**.

They are an important part of understanding:

- Variable management
- Scope
- LEGB Rule
- Functions
- Classes
- Modules
- Object references

# 🤔 Why Do We Need Namespaces?

Imagine a large Python program containing hundreds of variables and functions.

Without an organized system, Python would have difficulty determining which name refers to which object.

Namespaces solve this problem by organizing names into separate environments.

For example, two different functions can have variables with the same name:

```python
def first():
    name = "Alice"


def second():
    name = "Bob"
```

Both functions contain a variable called:

```python
name
```

However, these names exist in different namespaces.

Therefore, they do not conflict with each other.

Namespaces provide **organization and separation of names**.

A student named `"Rahul"` can exist in both the Computer Science and Mathematics departments without causing confusion because they belong to different contexts.

Different namespaces can contain the same name while referring to different objects.

The **context** determines which name Python is referring to.

# 📚 Types of Namespaces

Python mainly works with different namespaces during program execution.

Important namespaces include:

1. 🌍 Global Namespace
2. 🔵 Local Namespace
3. 🏠 Enclosing Namespace
4. 🐍 Built-in Namespace
5. 📦 Module Namespace

These namespaces help Python organize names and determine which object a particular name refers to.

# 🌍 Global Namespace

The **Global Namespace** contains names that are defined at the top level of a Python module or script.

For example:

```python
name = "Nikita"
age = 20

def greet():
    print("Hello")
```

Here, the names:

```text
name
age
greet
```

belong to the Global Namespace of that module.

The Global Namespace normally exists for the lifetime of the module.

````python
name = "Nikita"
age = 20


def greet():
    print("Hello Python")


print(globals()["name"])
print(globals()["age"])
print(globals()["greet"])
````

# 🔵 Local Namespace

A **Local Namespace** is created when a function starts executing.

It contains names created inside that function.

Example:

```python
def greet():

    message = "Hello"

    print(message)
```

Here:

```text
message
```

belongs to the Local Namespace of `greet()`.

The Local Namespace exists while the function is executing.

````python
def greet():

    message = "Hello Python"

    print(locals())


greet()
````

# 📝 Understanding `locals()`

Inside a function,

```python
locals()
```

returns information about the current Local Namespace.

For example:

```python
def calculate():

    x = 10
    y = 20

    print(locals())
```

The Local Namespace contains names such as:

```text
x
y
```

along with their associated objects.

Conceptually:

```text
x → 10
y → 20
```

````python
def calculate():

    x = 10
    y = 20

    namespace = locals()

    print("x =", namespace["x"])
    print("y =", namespace["y"])


calculate()
````

# 🏠 Enclosing Namespace

The **Enclosing Namespace** appears when we have nested functions.

For example:

```python
def outer():

    message = "Hello"

    def inner():

        print(message)
```

Here:

- `inner()` has its own Local Namespace.
- `message` belongs to the enclosing `outer()` function.

The `inner()` function can access `message` from its enclosing namespace.

This concept becomes especially important when learning:

- Closures
- Nested Functions
- LEGB Rule

````python
def outer():

    message = "Hello from Outer"

    def inner():

        print(message)

    inner()


outer()
````

# 🐍 Built-in Namespace

Python provides a **Built-in Namespace** containing names that are available throughout a Python program.

Examples include:

```python

max()
min()
type()
range()
```

These names are provided by Python itself.

For example:

```python

```

Python can find `print` because it exists in the Built-in Namespace.

````python
print(len("Python"))

print(sum([10, 20, 30]))

print(type(100))
````

# 📦 Module Namespace

Every Python module has its own namespace.

Consider a file:

```text
calculator.py
```

containing:

```python
def add(a, b):
    return a + b

PI = 3.14159
```

The module has its own namespace containing:

```text
add
PI
```

When the module is imported:

```python
import calculator
```

we can access these names through the module:

```python
calculator.add()
calculator.PI
```

This separation helps prevent naming conflicts between different modules.

# 📊 Namespace Overview

| Namespace | Created When | Example |
|---|---|---|
| Global | Module starts | `name`, `age` |
| Local | Function executes | `x`, `y` inside function |
| Enclosing | Nested function exists | Outer function variables |
| Built-in | Python starts | `print`, `len`, `sum` |
| Module | Module is loaded | Functions/classes/variables in module |

These namespaces work together to organize names in Python.

Namespaces are not the same thing as Scope.

A **Namespace** is the collection/mapping of names to objects.

A **Scope** determines where a particular name can be accessed in the program.

For example:

```text

    ↓
Stores names and objects

Scope
    ↓
Determines where those names are accessible
```

# 🔍 Inspecting the Global Namespace

Python provides the built-in function:

```python
globals()
```

It returns a dictionary representing the current Global Namespace.

For example:

```python
name = "Nikita"
age = 20

print(globals())
```

The returned dictionary contains names defined in the current global environment.

Conceptually:

```text
{
    "name": "Nikita",
    "age": 20,
    ...
}
```

In a Jupyter Notebook, the dictionary may contain many additional names created by the notebook environment.

````python
name = "Nikita"
age = 20

global_namespace = globals()

print("name:", global_namespace["name"])
print("age:", global_namespace["age"])
````

# 🔎 Checking Whether a Name Exists

Because `globals()` returns a dictionary, we can use normal dictionary operations to check whether a name exists.

For example:

```python
"name" in globals()
```

returns `True` if the name exists in the Global Namespace.

````python
name = "Nikita"

print("name" in globals())
print("city" in globals())
````

# ➕ Adding a Name to the Global Namespace

Since `globals()` returns a dictionary representing the Global Namespace, we can technically add a new name through it.

Example:

```python
globals()["language"] = "Python"
```

This creates a global name called `language`.

This technique is useful for understanding how namespaces work internally, but directly modifying `globals()` is generally **not recommended in normal application code**.

Prefer normal assignment:

```python
language = "Python"
```

````python
globals()["language"] = "Python"

print(language)
print(globals()["language"])
````

# ➖ Removing a Name from the Global Namespace

Because the Global Namespace behaves like a dictionary, a name can also technically be removed using:

```python
del globals()["language"]
```

Again, this is mainly useful for understanding Namespace behavior.

In normal Python programs, explicit variable management is usually clearer than dynamically modifying `globals()`.

````python
globals()["temporary_value"] = 100

print(temporary_value)

del globals()["temporary_value"]

print("temporary_value" in globals())
````

# 🧠 Namespace and Name Shadowing

Two different namespaces can contain the same name.


The function uses its local `name`.

This is an important concept that will become clearer when studying the **LEGB Rule**.

````python
name = "Global"


def show():

    name = "Local"

    print("Inside function :", name)


show()

print("Outside function:", name)
````

# 🏠 Enclosing Namespace in Nested Functions

When one function is defined inside another function, the inner function has access to the namespace of its enclosing function.

Example:

```python
def outer():

    message = "Hello"

    def inner():

        print(message)
```

Here:

- `message` is local to `outer()`.
- For `inner()`, `message` belongs to the **Enclosing Namespace**.
- `inner()` can access `message` even though it is not defined inside `inner()`.

This relationship is important for understanding **Closures**.

````python
def outer():

    message = "Hello from Outer"

    def inner():

        print("Inner:", message)
        print("Inner Local Namespace:", locals())

    inner()


outer()
````

# 🔗 Namespace and Closures

A Closure allows an inner function to retain access to variables from its enclosing function.

Example:

```python
def outer():

    value = 100

    def inner():

        return value

    return inner
```

After:

```python
closure = outer()
```

the `outer()` function has finished executing.

However, `closure` can still access:

```python
value
```

because the Closure preserves access to the required enclosing variable.

This demonstrates the connection between:

```text
Namespace
     ↓
Enclosing Variable
     ↓
Closure
     ↓
Inner Function
```

````python
def outer():

    value = 100

    def inner():

        return value

    return inner


closure = outer()

print("Result:", closure())

print("Free Variables:", closure.__code__.co_freevars)
````

# 🔬 Inspecting Closure Cells

A Closure stores references to its captured variables in **closure cells**.

We can inspect them using:

```python
closure.__closure__
```

Each cell contains a value captured from the enclosing scope.

For example:

```python
closure.__closure__[0].cell_contents
```

can be used to inspect the value stored in the first closure cell.

This is an advanced introspection technique and is useful for understanding how Python represents captured variables.

````python
def outer():

    message = "Hello Python"

    def inner():

        return message

    return inner


closure = outer()

print("Closure:", closure.__closure__)

if closure.__closure__:

    for cell in closure.__closure__:

        print("Stored Value:", cell.cell_contents)
````

# 🌍 Function Namespace and `__globals__`

Functions also maintain a reference to the Global Namespace in which they were defined.

This can be inspected using:

```python
function.__globals__
```

For example:

```python
def greet():
    print(message)
```

The function's `__globals__` attribute refers to the global namespace associated with the function.

This is useful when studying Python's internal name-resolution behavior.

````python
message = "Hello from Global"


def greet():

    print(message)


print(greet.__globals__["message"])

greet()
````

# 🧠 `__globals__` vs `globals()`

Both are related to the Global Namespace, but they are used in slightly different contexts.

### `globals()`

Returns the Global Namespace of the **current execution context**.

```python
globals()
```

### `function.__globals__`

Provides the Global Namespace associated with a **specific function**.

```python
greet.__globals__
```

Conceptually:

```text
globals()
     ↓
Current Global Namespace


greet.__globals__
     ↓
Global Namespace associated with greet
```

````python
course = "Advanced Python"


def show_course():

    print(course)


print(show_course.__globals__["course"])

show_course()
````

# 🧩 Same Name in Multiple Namespaces

The same name can exist simultaneously in different namespaces.

Example:

```python
name = "Global"


def outer():

    name = "Outer"

    def inner():

        name = "Inner"

        print(name)

    inner()


outer()
```

Here, three different `name` bindings exist:

```text
Global Namespace
    name → "Global"

Enclosing Namespace
    name → "Outer"

Local Namespace
    name → "Inner"
```

When `inner()` executes:

```python
print(name)
```

the local `name` is selected.

The exact lookup order is defined by the **LEGB Rule**, which we will study separately.

````python
name = "Global"


def outer():

    name = "Outer"

    def inner():

        name = "Inner"

        print("Selected:", name)

    inner()


outer()

print("Global:", name)
````

# ⚠️ Common Namespace Confusion

A common beginner mistake is assuming that a variable name represents a single value throughout the entire program.

For example:

```python
value = 10


def test():

    value = 20

    print(value)
```

There are two different bindings:

```text
Global Namespace
value → 10

Local Namespace
value → 20
```

The local `value` does not overwrite the global `value`.

It **shadows** it inside the function.

After the function finishes:

```python
value
```

in the global scope is still:

```text
10
```

````python
value = 10


def test():

    value = 20

    print("Inside:", value)


test()

print("Outside:", value)
````
