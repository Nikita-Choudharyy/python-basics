# 🎯 Scope in Python

# 📖 Introduction

In Python, **Scope** refers to the region of a program where a particular name can be accessed directly.

While a Namespace is responsible for storing names and their associated objects, Scope determines **where those names are accessible**.

For example:

```python
def greet():

    message = "Hello Python"

    print(message)
```

The variable:

```python
message
```

is accessible inside `greet()` because it belongs to the function's local scope.

However, trying to access it outside the function will result in an error.

Scope helps Python control the visibility and accessibility of names in a program.

# 🤔 Why Do We Need Scope?

Consider a large Python application containing hundreds of variables.

If every variable were accessible from every part of the program, it would become difficult to:

- Track where a variable came from.
- Avoid accidental modifications.
- Prevent naming conflicts.
- Understand function behavior.
- Maintain large programs.

Scope provides a controlled environment for names.

It determines where a name can be accessed and helps keep different parts of a program organized.

# 🟢 Local Scope

A **Local Scope** is created when a function executes.

Variables defined inside a function normally belong to that function's Local Scope.

Example:

```python
def greet():

    message = "Hello Python"

    print(message)
```

Here:

```python
message
```

is a local variable.

It can be accessed inside `greet()`.

````python
def greet():

    message = "Hello Python"

    print(message)


greet()
````

# ⚠️ Accessing a Local Variable Outside Its Scope

A local variable is not directly accessible outside the function where it was created.

For example:

```python
def greet():

    message = "Hello Python"


greet()

print(message)
```

This will raise:

```text
NameError
```

because `message` belongs to the Local Scope of `greet()`.

````python
def greet():

    message = "Hello Python"

    print("Inside:", message)


greet()

# Uncomment to see the error:
# print(message)
````

# 🧠 Important Concept

A variable's accessibility depends on its Scope.

```text
def greet():

    message = "Hello"

    └── message is accessible here
```

Outside the function:

```text
message
   ↓
Not directly accessible
```

This separation helps functions keep their internal variables independent from the rest of the program.

# 🌍 Global Scope

A variable defined at the top level of a Python module belongs to the **Global Scope** of that module.

Example:

```python
course = "Python"


def show_course():

    print(course)
```

The function can access the global variable because the name is available in an outer scope.

Global names can generally be accessed from functions, provided they are not shadowed by a local name.

````python
course = "Python"


def show_course():

    print("Course:", course)


show_course()
````

# 📝 Local vs Global Scope

Consider:

```python
course = "Python"


def show():

    topic = "Scope"

    print(course)
    print(topic)
```

Here:

```text
course → Global Scope

topic → Local Scope
```

The function can access the global `course` and its own local `topic`.

However, code outside the function cannot directly access the local `topic`.

````python
course = "Python"


def show():

    topic = "Scope"

    print("Global variable:", course)
    print("Local variable :", topic)


show()
````

# 🏠 Enclosing Scope

An **Enclosing Scope** occurs when functions are nested.

Consider:

```python
def outer():

    message = "Outer"

    def inner():

        print(message)
```

Here:

- `inner()` has its own Local Scope.
- `message` belongs to the Scope of `outer()`.
- For `inner()`, the scope of `outer()` is the **Enclosing Scope**.

The inner function can access names from its enclosing function.

````python
def outer():

    message = "Hello from Outer"

    def inner():

        print(message)

    inner()


outer()
````

Examples include:


These names are available without defining them ourselves.

Example:

```python
numbers = [10, 20, 30]
```

Python can find `len` through the Built-in Scope.

````python
numbers = [10, 20, 30]

print("Length:", len(numbers))
print("Sum   :", sum(numbers))
print("Max   :", max(numbers))
print("Type  :", type(numbers))
````

# 🧠 Scope vs Namespace

It is important not to confuse **Scope** with **Namespace**.

### Namespace

A Namespace stores mappings between names and objects.

```text
Name → Object
```

### Scope

Scope determines where a name can be accessed.

```text
Scope → Accessibility
```

For example:

```python
def greet():

    message = "Hello"
```

The Local Namespace stores the name:

```text
message → "Hello"
```

The Local Scope determines where `message` can be accessed.

So:

> **Namespace is about where names are stored, while Scope is about where names are accessible.**

# 🔄 Scope and Variable Assignment

Variable assignment behaves differently depending on the Scope in which it occurs.

Consider:

```python
x = 10

def change():
    x = 20
```

The assignment:

```python
x = 20
```

creates a **local variable** inside `change()`.

It does not automatically modify the global `x`.

This distinction is very important when working with functions.

````python
x = 10


def change():

    x = 20

    print("Inside Function :", x)


change()

print("Outside Function:", x)
````

# 🧠 Why Didn't the Global Variable Change?

In the previous example, Python sees:

```python
x = 20
```

inside the function.

Because the assignment happens inside the function, Python treats `x` as a **local name**.

Therefore:

```text
Global Scope
x → 10

Local Scope
x → 20
```

The two names have the same spelling but belong to different scopes.

````python
number = 100


def update():

    number = 500

    print("Local number :", number)


update()

print("Global number:", number)
````

# 🌍 Modifying a Global Variable

If we actually want to modify a global variable from inside a function, Python provides the:

```python
global
```

keyword.

Syntax:

```python
global variable_name
```

This tells Python that the name refers to the existing Global Scope rather than creating a new local name.

````python
count = 0


def increment():

    global count

    count += 1

    print("Inside Function:", count)


increment()
increment()
increment()

print("Final Count:", count)
````

# ⚠️ Common Error with Global Variables

Consider:

```python

    count += 1
```

This raises:

```text
UnboundLocalError
```

Why?

Because Python treats `count` as local due to the assignment:

```python
count += 1
```

but the local variable has not been initialized before it is read.

The solution, when modification of the global binding is genuinely intended, is:

```python
global count
```

````python
count = 0


def increment():

    global count

    count += 1


increment()

print(count)
````

# 🏠 Enclosing Scope and `nonlocal`

The `nonlocal` keyword is used when we want to modify a variable belonging to an **enclosing function scope**.

Example:

```python
def outer():

    count = 0

    def inner():

        nonlocal count

        count += 1
```

Here:

- `count` belongs to `outer()`.
- `inner()` is nested inside `outer()`.
- `nonlocal count` tells Python to use the enclosing `count`.

````python
def outer():

    count = 0

    def inner():

        nonlocal count

        count += 1

        print("Count:", count)

    inner()
    inner()
    inner()


outer()
````

# 🔗 `global` vs `nonlocal`

Both keywords modify names outside the current Local Scope, but they refer to different scopes.

### `global`

Used for a name in the **Global Scope**.

```python
global count
```

### `nonlocal`

Used for a name in an **Enclosing Function Scope**.

```python
nonlocal count
```

Comparison:

```text
global
   ↓
Global Scope


nonlocal
   ↓
Enclosing Function Scope
```

````python
value = 10


def global_example():

    global value

    value = 20


global_example()

print("Global value:", value)
````

````python
def enclosing_example():

    value = 10

    def change():

        nonlocal value

        value = 20

    change()

    print("Enclosing value:", value)


enclosing_example()
````

# 🪆 Nested Function Scope

Nested functions create multiple levels of scope.

Example:

```python
def outer():

    outer_value = "Outer"

    def middle():

        middle_value = "Middle"

        def inner():

            inner_value = "Inner"
```

Here:

```text
outer()
│
├── outer_value
│
└── middle()
    │
    ├── middle_value
    │
    └── inner()
        │
        └── inner_value
```

Each function has its own Local Scope.

For `inner()`, the scopes surrounding it form its Enclosing Scopes.

````python
def outer():

    outer_value = "Outer"

    def middle():

        middle_value = "Middle"

        def inner():

            inner_value = "Inner"

            print(inner_value)
            print(middle_value)
            print(outer_value)

        inner()

    middle()


outer()
````

# 🔎 Scope and Shadowing

A local variable can have the same name as a variable in an outer scope.

Example:

```python
message = "Global"

def outer():

    message = "Outer"

    def inner():

        message = "Inner"

        print(message)
```

Here:

```text
Global Scope
message → "Global"

Enclosing Scope
message → "Outer"

Local Scope
message → "Inner"
```

The `message` inside `inner()` refers to the local binding.

This is called **name shadowing**.

````python
message = "Global"


def outer():

    message = "Outer"

    def inner():

        message = "Inner"

        print("Inside inner:", message)

    inner()

    print("Inside outer:", message)


outer()

print("Global:", message)
````

# 🧪 Practical Example — Counter

Closures provide a useful example of Enclosing Scope.

```python
def counter():

    count = 0

    def increment():

        nonlocal count

        count += 1

        return count

    return increment
```

The variable `count` belongs to the enclosing function.

The inner function modifies it using:

```python
nonlocal count
```

This allows the returned function to preserve state between calls.

````python
def counter():

    count = 0

    def increment():

        nonlocal count

        count += 1

        return count

    return increment


counter_function = counter()

print(counter_function())
print(counter_function())
print(counter_function())
````

Notice the difference:

```python
global
```

works with the Global Scope, while:

```python
nonlocal
```

works with an Enclosing Function Scope.

Neither keyword is needed merely to **read** an outer variable.

They are mainly needed when an assignment should modify an existing binding outside the current Local Scope.

# 📦 Mutable Objects and Scope

Scope determines where a **name** is accessible.

It does not make the object itself local or global.

Consider:

```python
numbers = [1, 2, 3]

def add_number():

    numbers.append(4)
```

The function can modify the list because it is accessing the same mutable object referenced by the global name.

Notice that we are not rebinding the name `numbers`.

We are modifying the object that `numbers` already refers to.

````python
numbers = [1, 2, 3]


def add_number():

    numbers.append(4)


add_number()

print(numbers)
````

# 🧠 Rebinding vs Mutating

This distinction is extremely important.

### Mutating an Object

```python
numbers.append(4)
```

The existing list object is modified.



### Rebinding a Name

```python
numbers = [1, 2, 3, 4]
```

A new binding is created in the current scope.

This difference explains many confusing `global` and scope-related behaviors.

````python
numbers = [1, 2, 3]


def mutate():

    numbers.append(4)


mutate()

print("After mutation:", numbers)
````

````python
numbers = [1, 2, 3]


def rebind():

    numbers = [10, 20, 30]

    print("Inside Function:", numbers)


rebind()

print("Outside Function:", numbers)
````

# 🪆 Nested Scope in Depth

# 🔗 Closures and Scope

Closures are closely connected to Enclosing Scope.

A Closure occurs when an inner function remembers and retains access to variables from its enclosing function even after the enclosing function has finished executing.

Example:

```python
def create_multiplier(factor):

    def multiply(number):

        return number * factor

    return multiply
```

Here:

```python
factor
```

belongs to the enclosing scope of `multiply()`.

The returned function retains access to it.

````python
def create_multiplier(factor):

    def multiply(number):

        return number * factor

    return multiply


double = create_multiplier(2)
triple = create_multiplier(3)

print(double(10))
print(triple(10))
````

# 🧠 Why Does the Closure Remember `factor`?

After:

```python
double = create_multiplier(2)
```

the `create_multiplier()` function has finished executing.

However, `double` still remembers:

```text
factor → 2
```

Similarly:

```text
triple
   ↓
factor → 3
```

Each Closure maintains access to its own enclosing value.

This is an important practical example of Enclosing Scope.

````python
print("Double closure:", double.__closure__)
print("Triple closure:", triple.__closure__)

print(
    "Double free variables:",
    double.__code__.co_freevars
)

print(
    "Triple free variables:",
    triple.__code__.co_freevars
)
````

# 🧩 Scope and Function Factories

Function factories are a practical application of Enclosing Scope.

A function factory creates and returns customized functions.

Example:

```python
def power_factory(exponent):

    def power(number):

        return number ** exponent

    return power
```

Here:

```python
exponent
```

belongs to the Enclosing Scope of `power()`.

The returned function remembers it.

````python
def power_factory(exponent):

    def power(number):

        return number ** exponent

    return power


square = power_factory(2)
cube = power_factory(3)

print("Square:", square(5))
print("Cube  :", cube(5))
````

## 🔢 Counter System

Build a reusable counter system using a Closure.

### Requirements

Create:

```python
create_counter(start)
```

The function should return another function that increments the counter.

Example:

```python
counter = create_counter(10)

print(counter())
print(counter())
print(counter())
```

Expected output:

```text
11
12
13
```

### Requirements

- Use a Nested Function.
- Use Enclosing Scope.
- Use `nonlocal`.
- Preserve the counter state between calls.

### Bonus

Inspect the returned function using:

```python
__closure__
```

and:

```python
__code__.co_freevars
```

````python
def create_counter(start):

    count = start

    def increment():

        nonlocal count

        count += 1

        return count

    return increment


counter = create_counter(10)

print(counter())
print(counter())
print(counter())
````
