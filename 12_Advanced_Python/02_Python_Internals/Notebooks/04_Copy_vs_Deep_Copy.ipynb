# 🔗 Assignment Does Not Automatically Create a Copy

Consider:

```python
numbers = [10, 20, 30]

other = numbers
```

It is tempting to think:

```text
numbers → [10, 20, 30]

other   → [10, 20, 30]
```

as two independent lists.

But that is not what happens.

Instead:

```text
numbers ─────┐
             ↓
         [10, 20, 30]
             ↑
             └───── other
```

Both names refer to the **same list object**.

````python
numbers = [10, 20, 30]

other = numbers

print("numbers:", numbers)
print("other  :", other)
````

# 🔍 Checking Object Identity with `id()`

Python's built-in `id()` function returns an identity value for an object during its lifetime.

We can use it to check whether two names refer to the same object.

Example:

```python
numbers = [10, 20, 30]

other = numbers
```

Then:

```python
id(numbers)
id(other)
```

will refer to the same object identity.

This helps us understand the difference between:

- Same object
- Different objects with equal values

````python
numbers = [10, 20, 30]

other = numbers

print("ID of numbers:", id(numbers))
print("ID of other  :", id(other))

print("Same object:", numbers is other)
````

# 🆚 `==` vs `is`

These operators answer different questions.

## `==`

Checks whether two objects have equal values.

```python
a == b
```

means approximately:

> Do these objects compare equal?

---

## `is`

Checks object identity.

```python
a is b
```

means:

> Are these two names referring to the same object?

Example:

```python
a = [1, 2, 3]
b = [1, 2, 3]
```

The values may be equal:

```python
a == b
```

but they are separate list objects, so:

```python
a is b
```

is normally `False`.

````python
a = [1, 2, 3]
b = [1, 2, 3]

print("a == b:", a == b)
print("a is b:", a is b)

print("ID of a:", id(a))
print("ID of b:", id(b))
````

# 🧰 Creating a Shallow Copy

For a list, one common method is:

```python
shallow = original.copy()
```

Other common techniques include:

```python
shallow = original[:]
```

and:

```python
shallow = list(original)
```

The `copy` module also provides:

```python
copy.copy(original)
```

These approaches can create a shallow copy for appropriate objects.

````python
original = [1, 2, 3]

shallow = original.copy()

print("Original:", original)
print("Shallow :", shallow)

print("Same object:", original is shallow)
print("Equal values:", original == shallow)
````

# 🔍 Shallow Copy with Nested Lists

Consider:

```python
original = [
    [1, 2],
    [3, 4]
]

shallow = original.copy()
```

The outer lists are different:

```python
original is shallow
```

is `False`.

But the first nested list is shared:

```python
original[0] is shallow[0]
```

is `True`.

This is the key behavior of a shallow copy.

````python
original = [
    [1, 2],
    [3, 4]
]

shallow = original.copy()

print("Outer objects same:",
      original is shallow)

print("Nested objects same:",
      original[0] is shallow[0])

print("Nested objects same:",
      original[1] is shallow[1])
````

# ⚠️ Shallow Copy and Nested Mutation

Because nested mutable objects can be shared, changing a nested list can affect both structures.

Example:

```python
original = [
    [1, 2],
    [3, 4]
]

shallow = original.copy()

shallow[0].append(99)
```

The nested list is shared.

Therefore the modification can be visible through both `original` and `shallow`.

````python
original = [
    [1, 2],
    [3, 4]
]

shallow = original.copy()

shallow[0].append(99)

print("Original:", original)
print("Shallow :", shallow)
````

# 🔴 Deep Copy

A **Deep Copy** creates a new outer object and recursively creates copies of the nested objects as well.

Consider:

```python
original = [
    [1, 2],
    [3, 4]
]
```

A Deep Copy creates:

```text
original ─────→ Outer List A
                    │
                    ├──→ [1, 2]
                    └──→ [3, 4]


deep ─────────→ Outer List B
                    │
                    ├──→ [1, 2]
                    └──→ [3, 4]
```

The important difference is:

```text
Outer objects → Different
Nested objects → Different
```

Therefore, modifying a nested mutable object in the Deep Copy does not modify the corresponding nested object in the original structure.

# 🧰 `copy` Module

Python provides the `copy` module for copying objects.

Import it using:

```python
import copy
```

The module provides two important functions:

```python
copy.copy()
```

and:

```python
copy.deepcopy()
```

### `copy.copy()`

Creates a **Shallow Copy**.

### `copy.deepcopy()`

Creates a **Deep Copy**.

````python
import copy

original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)

deep = copy.deepcopy(original)

print("Original:", original)
print("Shallow :", shallow)
print("Deep    :", deep)
````

# 🔍 Comparing Object Identity

Let's compare the objects created using Shallow Copy and Deep Copy.

For the outer list:

```python
original is shallow
```

and:

```python
original is deep
```

should both be:

```text
False
```

because both copying operations create a new outer object.

The important difference appears when we inspect nested objects.

````python
import copy

original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

print("Original is Shallow:", original is shallow)
print("Original is Deep   :", original is deep)
````

# 🧩 Comparing Nested Object Identity

Now check the first nested list:

```python
original[0] is shallow[0]
```

For a Shallow Copy, this is:

```text
True
```

because the nested list is shared.

But:

```python
original[0] is deep[0]
```

is:

```text
False
```

because Deep Copy creates a separate nested object.

````python
import copy

original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

print(
    "Original[0] is Shallow[0]:",
    original[0] is shallow[0]
)

print(
    "Original[0] is Deep[0]:",
    original[0] is deep[0]
)
````

# 🆚 Shallow Copy vs Deep Copy

The main difference:

| Feature | Shallow Copy | Deep Copy |
|---|---|---|
| New outer object | ✅ | ✅ |
| Nested mutable objects copied | ❌ | ✅ |
| Nested objects shared | Often yes | No, for copied nested objects |
| Function | `copy.copy()` | `copy.deepcopy()` |
| Suitable for nested mutable structures | Depends | Often safer |

The key idea:

```text
Shallow Copy
    ↓
Copy outer object
    ↓
Reuse references to nested objects


Deep Copy
    ↓
Copy outer object
    ↓
Recursively copy nested objects
```

# 🧪 Practical Comparison

Let's modify a nested list in both copies.

For the Shallow Copy:

```python
shallow[0].append(100)
```

The nested list is shared with the original.

Therefore the original can also reflect the change.

For the Deep Copy:

```python
deep[0].append(200)
```

the nested list is independent.

Therefore the original is not modified by this nested mutation.

````python
import copy

original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

shallow[0].append(100)

deep[1].append(200)

print("Original:", original)
print("Shallow :", shallow)
print("Deep    :", deep)
````

# 🔬 Inspecting the IDs

We can use `id()` to understand what is happening internally.

Check:

```python
id(original)
id(shallow)
id(deep)
```

The outer IDs should be different.

Now check:

```python
id(original[0])
id(shallow[0])
id(deep[0])
```

For the Shallow Copy:

```text
original[0] and shallow[0]
```

refer to the same nested object.

For the Deep Copy:

```text
original[0] and deep[0]
```

refer to different nested objects.

````python
import copy

original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

print("Outer IDs")
print("Original:", id(original))
print("Shallow :", id(shallow))
print("Deep    :", id(deep))

print("\nNested IDs")
print("Original[0]:", id(original[0]))
print("Shallow[0] :", id(shallow[0]))
print("Deep[0]    :", id(deep[0]))
````

# 🏗️ Nested Dictionaries

The same concept applies to dictionaries containing nested mutable objects.

Example:

```python
original = {
    "student": {
        "name": "Nikita",
        "skills": ["Python", "SQL"]
    }
}
```

A Shallow Copy creates a new outer dictionary.

However, the nested dictionary and its nested list can still be shared.

A Deep Copy recursively copies the nested structure.

````python
import copy

original = {
    "student": {
        "name": "Nikita",
        "skills": ["Python", "SQL"]
    }
}

shallow = copy.copy(original)
deep = copy.deepcopy(original)

print(
    "Outer dictionary same:",
    original is shallow
)

print(
    "Nested dictionary same:",
    original["student"] is shallow["student"]
)

print(
    "Nested list same:",
    original["student"]["skills"]
    is shallow["student"]["skills"]
)

print(
    "Deep nested dictionary same:",
    original["student"] is deep["student"]
)

print(
    "Deep nested list same:",
    original["student"]["skills"]
    is deep["student"]["skills"]
)
````

# ⚠️ Nested Mutation in Dictionaries

Consider:

```python
shallow["student"]["skills"].append("Machine Learning")
```

Because the nested list is shared by the Shallow Copy and the original, the change can be visible in both.

With a Deep Copy:

```python
deep["student"]["skills"].append("Machine Learning")
```

the copied nested list is independent from the original nested list.

````python
import copy

original = {
    "student": {
        "name": "Nikita",
        "skills": ["Python", "SQL"]
    }
}

shallow = copy.copy(original)
deep = copy.deepcopy(original)

shallow["student"]["skills"].append("Machine Learning")

deep["student"]["skills"].append("Statistics")

print("Original:", original)
print("Shallow :", shallow)
print("Deep    :", deep)
````

```python

```

and:

```python
original = [[1, 2], [3, 4]]
```

# 🔄 Mutation vs Reassignment

Copying problems become easier to understand when we distinguish between:

### Mutation

Changing an existing object.

Example:

```python
items.append(100)
```

The existing list is modified.

---

### Reassignment

Making a name refer to another object.

Example:

```python
items = [100]
```

The name now refers to a different list.

These two operations can behave differently when working with copies.

````python
import copy


original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)


# Mutation
shallow[0].append(100)

print("After mutation:")
print("Original:", original)
print("Shallow :", shallow)
````

````python
import copy


original = [
    [1, 2],
    [3, 4]
]

shallow = copy.copy(original)


# Reassignment
shallow[0] = [100, 200]

print("After reassignment:")
print("Original:", original)
print("Shallow :", shallow)
````

# 🧠 Why Did Reassignment Behave Differently?

Initially:

```text
original
   │
   ├──→ [1, 2]
   └──→ [3, 4]

shallow
   │
   ├──→ [1, 2]
   └──→ [3, 4]
```

The nested `[1, 2]` is shared.

But when we do:

```python
shallow[0] = [100, 200]
```

we replace the reference stored at index `0` in the **shallow outer list**.

The original outer list still points to the original nested list.

Therefore:

```text
Mutation
→ changes shared object

Reassignment
→ changes reference in one container
```

Now consider:

```python

    [1, 2],
```

This is not a shallow copy.

It is simply **assignment**.

Both names refer to the same outer list and therefore the same nested objects.

````python
original = [
    [1, 2],
    [3, 4]
]

other = original

other[0].append(99)

print("Original:", original)
print("Other   :", other)

print(
    "Same outer object:",
    original is other
)
````

# 📊 Three-Way Comparison

There are three different situations:

## 1. Assignment

```python
b = a
```

```text
a ─────┐
       ↓
     Object
       ↑
       └───── b
```

Same object.

---

## 2. Shallow Copy

```python
b = copy.copy(a)
```

```text
a → Outer A
      │
      └──→ Nested Object

b → Outer B
      │
      └──→ Same Nested Object
```

New outer object, shared nested references.

---

## 3. Deep Copy

```python
b = copy.deepcopy(a)
```

```text
a → Outer A
      │
      └──→ Nested A

b → Outer B
      │
      └──→ Nested B
```

Nested objects are recursively copied.

# 🧠 Choosing the Right Approach

Use simple assignment when:

```python
b = a
```

is intentionally meant to create another reference to the same object.

Use Shallow Copy when:

- You need a new outer container.
- Sharing nested objects is acceptable.
- The nested structure does not need independent mutation.

Use Deep Copy when:

- You need an independent nested structure.
- Nested mutable objects must not be shared.
- You need to modify the copy without affecting the original nested data.

However, Deep Copy should not automatically be the default.

Always consider whether independent recursive copying is actually necessary.

# ⚠️ Important Performance Consideration

Deep Copy can be more expensive than Shallow Copy because it may need to recursively copy many nested objects.

For a large and deeply nested structure:

```python
copy.deepcopy(data)
```

can require significantly more work than:

```python
copy.copy(data)
```

Therefore:

> Use the simplest copying strategy that correctly satisfies your requirement.

Do not use Deep Copy just because it sounds safer.

````python
import copy


data = {
    "student": {
        "name": "Nikita",
        "skills": ["Python", "SQL"],
        "scores": {
            "Python": 90,
            "SQL": 85
        }
    }
}


shallow = copy.copy(data)
deep = copy.deepcopy(data)


print("Outer dictionary:")
print("Original is Shallow:", data is shallow)
print("Original is Deep   :", data is deep)


print("\nStudent dictionary:")
print(
    "Original student is Shallow student:",
    data["student"] is shallow["student"]
)

print(
    "Original student is Deep student:",
    data["student"] is deep["student"]
)
````

# 🔗 Shared References Inside One Structure

A data structure can contain the **same object more than once**.

Example:

```python
shared = [1, 2, 3]

data = [shared, shared]
```

Here:

```python
data[0] is data[1]
```

is `True`.

Both positions refer to the same list object.

This is called a **shared reference**.

````python
shared = [1, 2, 3]

data = [shared, shared]

print("data:", data)

print(
    "Same nested object:",
    data[0] is data[1]
)
````

# 🧠 Deep Copy and Shared References

Deep Copy does more than simply create independent objects.

It also preserves the **internal sharing relationships** of the copied structure.

Consider:

```python
shared = [1, 2, 3]

data = [shared, shared]
```

After:

```python
deep = copy.deepcopy(data)
```

we expect:

```python
deep[0] is deep[1]
```

to remain `True`.

But:

```python
data[0] is deep[0]
```

should be `False`.

So Deep Copy creates a new corresponding shared object rather than unnecessarily creating two unrelated copies.

````python
import copy


shared = [1, 2, 3]

data = [shared, shared]

deep = copy.deepcopy(data)


print(
    "Original shared:",
    data[0] is data[1]
)

print(
    "Deep shared:",
    deep[0] is deep[1]
)

print(
    "Original vs Deep:",
    data[0] is deep[0]
)
````

# 🧩 Why Preserving Shared References Matters

Suppose two parts of a data structure intentionally point to the same object.

If copying created unrelated objects instead, the relationship between those parts would be lost.

Deep Copy therefore tracks objects that it has already copied during the copying process.

This helps preserve the structure of references inside the copied object graph.

# 📚 Copying Dictionaries

Dictionaries can also be copied using:

```python
dictionary.copy()
```

This creates a shallow copy.

Example:

```python

    "name": "Python",
    "topics": ["OOP", "Functions"]
}
```

The outer dictionary is copied.

But the nested `topics` list can remain shared.

````python
original = {
    "name": "Python",
    "topics": ["OOP", "Functions"]
}

copied = original.copy()

print("Same dictionary:", original is copied)

print(
    "Same topics list:",
    original["topics"] is copied["topics"]
)
````

# 🧮 Copying Sets

Sets are mutable objects.

A set can be shallow-copied using:

```python
set.copy()
```

Example:

```python
original = {1, 2, 3}

copied = original.copy()
```

The two sets are separate objects.

Since a set cannot directly contain mutable elements such as lists or dictionaries, the simple set case does not have the same nested-mutable issue as a list containing lists.

````python
original = {1, 2, 3}

copied = original.copy()

print("Original:", original)
print("Copied  :", copied)

print("Same object:", original is copied)
print("Equal values:", original == copied)
````

# 📦 Copying Tuples

A tuple is immutable, but it may contain mutable objects.

Example:

```python
original = (
    [1, 2],
    [3, 4]
)
```

A shallow copy can reuse the same tuple object because the tuple itself is immutable.

Deep Copy may create a new tuple when its contents require copying.

The important point is:

> Copy behavior depends on the entire object graph, not only the outer container type.

````python
import copy


original = (
    [1, 2],
    [3, 4]
)


shallow = copy.copy(original)
deep = copy.deepcopy(original)


print("Original is Shallow:", original is shallow)
print("Original is Deep   :", original is deep)

print(
    "Nested list shared with Shallow:",
    original[0] is shallow[0]
)

print(
    "Nested list shared with Deep:",
    original[0] is deep[0]
)
````

# ⚠️ Deep Copy Is Not Always Necessary

It may be tempting to use:

```python
copy.deepcopy()
```

for every copying operation.

That is not always a good idea.

Deep Copy can:

- consume more memory
- take more time
- copy objects that do not need independent copies
- make the intention of the code less clear

A better approach is:

```text
First understand the data structure.
        ↓
Determine whether nested objects are shared.
        ↓
Choose the simplest correct copying strategy.
```

# 🧠 Shallow Copy Can Be the Right Choice

Suppose we have:

```python
config = {
    "name": "ML Project",
    "settings": {
        "learning_rate": 0.01
    }

```

If we only need an independent outer dictionary and will not modify the nested settings, a Shallow Copy may be sufficient.

Use Deep Copy when the nested mutable objects also need to become independent.

# 🔥 Advanced Example — Selective Reassignment

Consider:

```python
original = {
    "skills": ["Python", "SQL"],
    "scores": [90, 80]
}

shallow = original.copy()
```

Now:

```python
shallow["skills"] = ["Python", "SQL", "ML"]
```

This is **reassignment** of the `skills` reference in the shallow dictionary.

It does not mutate the previously shared list.

Therefore, the original dictionary can remain unchanged.

This is different from:

```python
shallow["skills"].append("ML")
```

which mutates the shared nested list.

````python
original = {
    "skills": ["Python", "SQL"],
    "scores": [90, 80]
}

shallow = original.copy()


shallow["skills"] = [
    "Python",
    "SQL",
    "Machine Learning"
]


print("Original:", original)
print("Shallow :", shallow)
````

# 🆚 Mutation vs Reassignment — Final Comparison

### Mutation

```python
shallow["skills"].append("ML")
```

The shared nested list is modified.

Result:

```text
Original → Changed
Shallow  → Changed
```

---

### Reassignment

```python
shallow["skills"] = ["Python", "SQL", "ML"]
```

The shallow dictionary now points to a new list.

Result:

```text
Original → Original List
Shallow  → New List
```

This distinction is one of the most important concepts in understanding shallow copies.

````python
original = {
    "skills": ["Python", "SQL"]
}

shallow = original.copy()


# Mutation
shallow["skills"].append("ML")

print("After mutation:")
print("Original:", original)
print("Shallow :", shallow)


# Reset
original = {
    "skills": ["Python", "SQL"]
}

shallow = original.copy()


# Reassignment
shallow["skills"] = [
    "Python",
    "SQL",
    "ML"
]

print("\nAfter reassignment:")
print("Original:", original)
print("Shallow :", shallow)
````
