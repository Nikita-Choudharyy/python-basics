# 🎨 Decorators in Python

# 📖 Introduction

As software applications become larger and more complex, developers often need to add extra functionality to existing functions without modifying their original code.

For example, we may want to:

- Measure the execution time of a function.
- Log function calls.

- Validate input data.
- Handle exceptions.
- Cache expensive computations.

One of the biggest advantages of Decorators is that they **extend the behavior of existing functions without changing their original implementation**.

Decorators are built using **Functions**, **Closures**, and **Higher-Order Functions**, making them one of the most important concepts in Functional Programming.

They are widely used in Python frameworks such as:

- 🌐 Flask

# 🤔 Why Do We Need Decorators?

Suppose you have ten different functions.

Now imagine you want to:

- Log every function call.
- Measure execution time.
- Check whether the user is authenticated.
- Validate function inputs.

One approach is to add the same code inside every function.

This creates:

- ❌ Code duplication.
- ❌ Poor maintainability.
- ❌ Difficult debugging.
- ❌ Reduced code readability.

Decorators solve this problem.

Instead of modifying each function individually,

a Decorator wraps the original function and automatically adds the required functionality.

This follows an important software engineering principle:

> **Extend behavior without modifying existing code.**

This makes programs cleaner, reusable, and easier to maintain.

# 🌍 Real-World Analogy

Imagine buying a mobile phone.

The phone works perfectly on its own.

Now you decide to add:

- 📱 A protective case.
- 🛡️ A screen protector.
- 🔋 A power bank.

These accessories add new functionality and protection,

but they **do not change the original phone itself**.

Decorators work in exactly the same way.

The original function remains unchanged.

The Decorator simply adds extra functionality around it.

Instead of modifying the function,

it **wraps** the function with additional behavior.

This makes Decorators powerful and reusable.

# 📌 What is a Decorator?

A **Decorator** is a function that takes another function as its input, adds additional functionality to it, and returns a new enhanced function.

The original function is **not modified**.

Instead, the Decorator wraps the original function with extra behavior.

This allows developers to extend the functionality of existing functions without changing their source code.



## 📌 Definition

> **A Decorator is a function that accepts another function, adds extra functionality, and returns a new wrapped function.**

# 📌 Higher-Order Functions

Before learning Decorators, it is important to understand **Higher-Order Functions**.

A Higher-Order Function is a function that:

- Accepts another function as an argument, or
- Returns another function as its result.

Since Python treats functions as **first-class objects**, functions can be passed, stored, and returned just like any other object.

Decorators are built using Higher-Order Functions.

````python
def greet():

    print("Hello, Python!")


def display(function):

    print("Before Function Call")

    function()

    print("After Function Call")


display(greet)
````

# 📝 Explanation

In this example,

```python
greet
```

is passed as an argument to

```python
display()
```

The `display()` function accepts another function and executes it.

Since it accepts another function as an argument,

it is called a **Higher-Order Function**.

This is the basic idea behind Decorators.

# 📦 Function Wrapping

A Decorator wraps an existing function by executing additional code:

- Before the original function.
- After the original function.

The original function remains unchanged.

This process is called **Function Wrapping**.

````python
def decorator(function):

    def wrapper():

        print("Before Function Execution")

        function()

        print("After Function Execution")

    return wrapper


def message():

    print("Welcome to Python")


decorated_function = decorator(message)

decorated_function()
````

# 📝 Explanation

Let's understand the execution flow.

### Step 1

The original function:

```python
message()
```

is passed to:

```python
decorator(message)
```

---

### Step 2

The Decorator creates a new function:

```python
wrapper()
```

The wrapper adds extra functionality before and after calling the original function.

---

### Step 3

The Decorator returns:

```python
wrapper
```

instead of the original function.

---

### Step 4

When we execute:

```python
decorated_function()
```

Python actually executes:

```python
wrapper()
```

The wrapper then calls the original function internally.

This is how Decorators extend functionality without modifying the original function.

# 🔄 Execution Flow of a Decorator

```text
Original Function
        │
        ▼
Pass Function to Decorator
        │
        ▼
Decorator Creates Wrapper
        │
        ▼
Wrapper Adds Extra Functionality
        │
        ▼
Decorator Returns Wrapper
        │
        ▼
Call Wrapped Function
        │
        ▼
Before Code
        │
        ▼
Original Function
        │
        ▼
After Code
```

# 💡 Did You Know?

A Decorator does **not** modify the original function.

Instead,

it creates and returns a **new function** that wraps the original function.

This is why Decorators are considered a clean and reusable way to extend functionality in Python.

Decorators are built using **Closures**, because the `wrapper()` function remembers the original function passed to the Decorator.

# 🎨 The `@` Decorator Syntax

In the previous examples, we applied a Decorator manually.

Example:

```python
decorated_function = decorator(message)

decorated_function()
```

Although this works, Python provides a cleaner and more readable syntax using the **`@` symbol**.

The `@` syntax automatically applies the Decorator to the function.

It is simply a shortcut for manually wrapping a function with a Decorator.



## 📌 Equivalent Forms

These two pieces of code are exactly the same.

### Method 1 (Manual)

```python
message = decorator(message)

message()
```



### Method 2 (`@` Syntax)

```python
@decorator
def message():
    print("Hello")
```

Python automatically converts the second form into the first one.

````python
def decorator(function):

    def wrapper():

        print("Before Function Execution")

        function()

        print("After Function Execution")

    return wrapper


@decorator
def message():

    print("Welcome to Python")


message()
````

# 📝 Explanation

When Python sees:

```python
@decorator
```

it automatically performs:

```python
message = decorator(message)
```

This means the original function is replaced with the wrapped version returned by the Decorator.

From this point onward,

every call to:

```python
message()
```

actually executes:

```python
wrapper()
```

instead of the original function directly.

# 📥 Decorators with Function Arguments

Until now,

our Decorators worked only with functions that accepted no arguments.

However,

real-world functions usually take one or more parameters.

To support any number of positional and keyword arguments,

Decorators use:

- `*args`
- `**kwargs`

This makes the Decorator flexible and reusable.

````python
def decorator(function):

    def wrapper(*args, **kwargs):

        print("Before Function Execution")

        function(*args, **kwargs)

        print("After Function Execution")

    return wrapper


@decorator
def greet(name):

    print(f"Hello, {name}!")


greet("Nikita")
````

# 📝 Explanation

The wrapper accepts:

```python
*args
```

for positional arguments and

```python
**kwargs
```

for keyword arguments.

These arguments are then forwarded to the original function.

This allows the same Decorator to work with many different functions regardless of their parameters.

# 📤 Returning Values from Decorated Functions

Sometimes,

the original function returns a value.

The Decorator should preserve that return value.

Otherwise,

the caller will receive `None`.

````python
def decorator(function):

    def wrapper(*args, **kwargs):

        print("Calculating...")

        result = function(*args, **kwargs)

        print("Calculation Completed")

        return result

    return wrapper


@decorator
def add(a, b):

    return a + b


answer = add(10, 20)

print("Result :", answer)
````

# 📝 Explanation

The wrapper stores the original function's return value:

```python
result = function(*args, **kwargs)
```

After performing additional operations,

it returns the same value back to the caller.

This ensures that the Decorator does not change the original behavior of the function.

# 🌍 Real-World Example – Logging Function Calls

One common use of Decorators is logging.

Whenever a function is called,

the Decorator records information before executing the function.

````python
def logger(function):

    def wrapper(*args, **kwargs):

        print(f"Calling Function: {function.__name__}")

        return function(*args, **kwargs)

    return wrapper


@logger
def multiply(a, b):

    return a * b


print(multiply(5, 4))
````

# 📝 Explanation

The Decorator prints the function name before execution.

After logging,

it calls the original function and returns its result.

This technique is commonly used in:

- Application logging
- Debugging
- Monitoring
- Performance tracking

This flexibility makes Decorators one of Python's most powerful features for writing reusable and maintainable code.

# 🎯 Decorators with Arguments

Sometimes, we want a Decorator to behave differently depending on some configuration.

For example:

- Log messages with different levels.
- Restrict access based on user roles.
- Retry a function a specified number of times.
- Set custom execution limits.

In such cases, the Decorator itself must accept arguments.

This requires **three nested functions**:

1. The outer function accepts the Decorator arguments.
2. The Decorator function accepts the original function.
3. The Wrapper function executes the original function.

This structure allows us to customize the behavior of the Decorator.

````python
def repeat(times):

    def decorator(function):

        def wrapper(*args, **kwargs):

            for _ in range(times):

                function(*args, **kwargs)

        return wrapper

    return decorator


@repeat(3)
def greet():

    print("Hello Python")


greet()
````

# 📝 Explanation

Let's understand the execution flow.

Python first executes:

```python
repeat(3)
```

This returns the actual Decorator.

Then,

that Decorator receives:

```python
greet
```

Finally,

the Wrapper executes the function three times.

Execution Flow

```text
repeat(3)
      │
      ▼
Decorator
      │
      ▼
Wrapper
      │
      ▼
greet()
```

# 🧱 Multiple Decorators

Python allows multiple Decorators to be applied to the same function.

Decorators are applied **from bottom to top**, but they execute **from top to bottom**.

Each Decorator wraps the result of the previous one.

````python
def decorator_one(function):

    def wrapper():

        print("Decorator One - Before")

        function()

        print("Decorator One - After")

    return wrapper


def decorator_two(function):

    def wrapper():

        print("Decorator Two - Before")

        function()

        print("Decorator Two - After")

    return wrapper


@decorator_one
@decorator_two
def message():

    print("Welcome!")


message()
````

# 📝 Explanation

Python internally converts:

```python
@decorator_one
@decorator_two
def message():
```

into:

```python
message = decorator_one(
            decorator_two(message)
          )
```

This is why understanding the order of Decorators is important.

# 🛡️ Preserving Function Metadata with `functools.wraps`

When a function is decorated,

Python replaces the original function with the Wrapper function.

As a result,

information such as:

- Function Name
- Documentation String (Docstring)

may be lost.

To preserve the original metadata,

Python provides:

```python
functools.wraps
```

````python
from functools import wraps


def decorator(function):

    @wraps(function)
    def wrapper():

        print("Before Execution")

        function()

    return wrapper


@decorator
def greet():

    """Greeting Function"""

    print("Hello!")


print(greet.__name__)

print(greet.__doc__)
````

# 📝 Explanation

Without `@wraps`,

the decorated function would appear as:

```text
wrapper
```

instead of:

```text
greet
```

Using `@wraps` preserves important metadata, making debugging, documentation, and introspection much easier.

For this reason, professional Python code almost always uses `functools.wraps` when writing custom Decorators.

# ⚠️ Common Mistakes

While learning Decorators, beginners often make the following mistakes.

## ❌ Mistake 1: Forgetting to Return the Wrapper Function

Incorrect

```python
def decorator(function):

    def wrapper():

        function()
```

The Decorator returns `None` because the wrapper function is never returned.

---

Correct

```python
def decorator(function):

    def wrapper():

        function()

    return wrapper
```

Always return the wrapper function.

---

## ❌ Mistake 2: Forgetting to Call the Original Function

Incorrect

```python
def decorator(function):

    def wrapper():

        print("Before Execution")

    return wrapper
```

The original function is never executed.

---

Correct

```python
def decorator(function):

    def wrapper():

        print("Before Execution")

        function()

    return wrapper
```

---

## ❌ Mistake 3: Not Using *args and **kwargs

Incorrect

```python
def decorator(function):

    def wrapper():

        function()
```

This works only for functions without parameters.

---

Correct

```python
def decorator(function):

    def wrapper(*args, **kwargs):

        return function(*args, **kwargs)

    return wrapper
```

This Decorator can now decorate almost any function.

---

## ❌ Mistake 4: Forgetting to Return the Result

Incorrect

```python
def wrapper(*args, **kwargs):

    function(*args, **kwargs)
```

If the original function returns a value,

it will be lost.

---

Correct

```python
def wrapper(*args, **kwargs):

    return function(*args, **kwargs)
```

---

## ❌ Mistake 5: Forgetting functools.wraps

Without

```python
@wraps(function)
```

Python replaces the original metadata.

Example

```python
print(greet.__name__)
```

Output

```text
wrapper
```

Instead of

```text
greet
```

Always use `functools.wraps` in professional Decorators.

# ✅ Best Practices

Follow these best practices while working with Decorators.

### 🎯 Keep Decorators Focused

Each Decorator should perform only one responsibility.

---

### 📦 Always Use `*args` and `**kwargs`

This makes Decorators reusable for different function signatures.

---

### 🛡️ Use `functools.wraps`

Preserve the original function's metadata.

---

### 🔄 Return the Original Result

If the original function returns a value,

the Decorator should return it as well.

---

### 🧩 Create Reusable Decorators

Avoid writing duplicate wrapper logic.

Instead,

create reusable Decorators that can be applied to multiple functions.

---

### 📝 Write Meaningful Names

Use descriptive names for Decorators such as:

- `logger`
- `timer`
- `authenticate`
- `retry`

instead of generic names like `decorator1`.
