# ⚡ Generators in Python

# 📖 Introduction

As Python applications grow larger, they often need to process large amounts of data efficiently.

Suppose you want to generate one million numbers.

One approach is to create a list containing all the numbers.

Although this works, it consumes a significant amount of memory because every value is stored in memory at the same time.

Python provides a more efficient solution called **Generators**.

A Generator produces values **one at a time**, only when they are requested, instead of storing all values in memory.

This technique is known as **Lazy Evaluation**.

Generators are built on the **Iterator Protocol**, which means every generator is automatically an iterator.

Because generators produce values on demand, they are highly memory-efficient and are widely used in Data Science, Machine Learning, Web Development, and Python's standard library.

By the end of this notebook, you will understand how generators work, how to create them using the `yield` keyword, and why they are preferred over lists in many real-world situations.

Instead of creating all values at once,

# 📌 What is a Generator?

A **Generator** is a special type of function that produces values **one at a time** instead of returning all values at once.

Unlike a normal function, a generator does not terminate immediately after producing a value.

Instead, it **pauses its execution**, remembers its current state, and resumes from the same point when the next value is requested.

Generators are created using the **`yield`** keyword.

Since generators produce values only when needed, they are highly memory-efficient and are ideal for processing large datasets.



## 📌 Definition

> **A Generator is a special function that uses the `yield` keyword to produce values one at a time while preserving its execution state.**

# 🔄 Generator vs Normal Function

Although Generators and normal functions look similar, they behave very differently.

| Normal Function | Generator Function |
|-----------------|-------------------|
| Uses `return` | Uses `yield` |
| Returns all results at once | Produces one value at a time |
| Execution ends after `return` | Execution pauses after `yield` |
| Does not remember its previous state | Remembers its previous state |
| Less memory-efficient for large data | Highly memory-efficient |


````python
# Normal Function

def numbers():
    return [1, 2, 3, 4, 5]

result = numbers()

print(result)
````

````python
# Generator Function

def numbers():

    yield 1
    yield 2
    yield 3
    yield 4
    yield 5

result = numbers()

print(result)
````

# 📝 Explanation

The normal function returns the complete list immediately.

However,

the Generator Function does **not** execute immediately.

Instead,

```python
numbers()
```

returns a **Generator Object**.

The values are produced only when requested using:

- `next()`
- `for` loop

This is what makes generators memory-efficient.

# 🏷️ The `yield` Keyword

The **`yield`** keyword is used to create Generator Functions.

Whenever Python encounters a `yield` statement,

it:

- Returns the current value.
- Pauses the function.
- Saves the current execution state.
- Waits until the next value is requested.

When execution resumes,

the function continues **exactly after the previous `yield` statement**.

Unlike `return`, `yield` does not terminate the function permanently.

````python
def greet():

    print("Start")

    yield "Hello"

    print("Middle")

    yield "Python"

    print("End")


generator = greet()

print(next(generator))

print(next(generator))
````

# 📝 Explanation

Let's understand the execution step by step.

### First `next()`

Python starts executing the function.

```text
Start
```

is printed.

Execution reaches:

```python
yield "Hello"
```

The value

```text
Hello
```

is returned,

and the function **pauses**.



### Second `next()`

Execution resumes **after the first `yield`**.

Python prints:

```text
Middle
```

Execution reaches:

```python
yield "Python"
```

The value

```text
Python
```

is returned,

and the function pauses again.

The function has **not restarted** from the beginning—it continues from where it previously paused.

# 💡 Did You Know?

A Generator Function is **not executed when it is called**.

For example,

```python
generator = greet()
```

does **not** execute the function body.

The function starts executing only when the first value is requested using:

```python
next(generator)
```

or by iterating over it with a `for` loop.

This behavior is known as **Lazy Evaluation**, and it is one of the key reasons why generators are memory-efficient.

# ▶️ Using `next()` with Generators

A Generator produces values only when they are requested.

The built-in `next()` function requests the next available value from the Generator.

Each call to `next()`:

- Executes the Generator until the next `yield` statement.
- Returns the yielded value.
- Pauses the Generator.
- Saves its current execution state.

When `next()` is called again, the Generator resumes from where it previously paused.

````python
def numbers():

    yield 10
    yield 20
    yield 30


generator = numbers()

print(next(generator))
print(next(generator))
print(next(generator))
````

# 📝 Explanation

When the Generator is created,

```python
generator = numbers()
```

no values are generated immediately.

The Generator starts executing only when:

```python
next(generator)
```

is called.

Each call returns one value and pauses the function at the corresponding `yield` statement.

This process continues until all values have been produced.

# ⚠️ StopIteration in Generators

Just like Iterators,

Generators also raise the **`StopIteration`** exception when all values have been generated.

````python
def colors():

    yield "Red"
    yield "Green"


generator = colors()

print(next(generator))
print(next(generator))
print(next(generator))
````

# 📝 Explanation

The Generator produces:

- `"Red"`
- `"Green"`

After the second value,

there are no more `yield` statements.

Therefore,

calling

```python
next(generator)
```

again raises:

```python
StopIteration
```

This indicates that the Generator has finished producing values.

# 🔁 Using a `for` Loop with Generators

Instead of manually calling `next()`,

Generators are usually used with a `for` loop.

The `for` loop automatically:

- Creates an iterator.
- Calls `next()`.
- Handles the `StopIteration` exception.

This makes the code cleaner and easier to read.

````python
def fruits():

    yield "Apple"
    yield "Banana"
    yield "Mango"


for fruit in fruits():
    print(fruit)
````

# 📝 Explanation

The `for` loop automatically requests one value at a time from the Generator.

Internally, Python performs operations similar to:

```python
generator = fruits()

while True:
    try:
        value = next(generator)
        print(value)

    except StopIteration:
        break
```

The `for` loop hides these implementation details, making Generator usage much simpler.

# 🧠 How Generators Work Internally

The execution flow of a Generator can be visualized as follows.

```text
Call Generator Function
          │
          ▼
Generator Object Created
          │
          ▼
next() Called
          │
          ▼
Execute Until yield
          │
          ▼
Return Value
          │
          ▼
Pause Execution
          │
          ▼
next() Called Again
          │
          ▼
Resume From Previous yield
          │
          ▼
Repeat Until No More yield
          │
          ▼
Raise StopIteration
```

````python
def countdown():

    print("3")
    yield

    print("2")
    yield

    print("1")
    yield

    print("Done")


generator = countdown()

next(generator)
next(generator)
next(generator)

try:
    next(generator)
except StopIteration:
    pass
````

# 📝 Explanation

Notice that the Generator remembers its execution state.

Each call to `next()` continues exactly from the previous `yield` statement.

This ability to pause and resume execution is what makes Generators different from normal functions.

# 💡 Did You Know?

A Generator can be **iterated only once**.

After it has been completely consumed,

it cannot be restarted.

To generate the values again,

you must create a **new Generator object** by calling the Generator Function again.

```python
generator = numbers()
```

This creates a fresh Generator that starts from the beginning.

# ⚡ Generator Expressions

Just as Python provides **List Comprehensions** for creating lists, it also provides **Generator Expressions** for creating generators.

A Generator Expression creates values **one at a time** instead of storing all values in memory.

The syntax is very similar to a List Comprehension.

The only difference is:

- List Comprehension uses **square brackets `[]`**.
- Generator Expression uses **parentheses `()`**.

Generator Expressions are useful when working with large datasets because they consume much less memory.

````python
# List Comprehension

numbers = [x for x in range(5)]

print(numbers)
````

````python
# Generator Expression

numbers = (x for x in range(5))

print(numbers)
````

# 📝 Explanation

The List Comprehension immediately creates and stores all values in memory.

The Generator Expression does not generate the values immediately.

Instead,

it creates a **Generator Object**.

The values are generated only when requested using:

- `next()`
- `for` loop

This makes Generator Expressions much more memory-efficient.

````python
numbers = (x for x in range(5))

for number in numbers:
    print(number)
````

# 📊 List vs Generator

| List | Generator |
|------|-----------|
| Uses `[]` | Uses `()` |
| Stores all values immediately | Generates values on demand |
| Higher memory usage | Lower memory usage |
| Faster for small datasets | Better for large datasets |
| Can be reused multiple times | Can be consumed only once |

# 💾 Memory Efficiency

Suppose you need one million numbers.

### Using a List

```python
numbers = [x for x in range(1_000_000)]
```

Python creates and stores all one million numbers in memory before the program can use them.



### Using a Generator

```python
numbers = (x for x in range(1_000_000))
```

Python generates each number only when it is requested.

As a result,

memory usage remains much lower.

````python
numbers = (x for x in range(5))

print(next(numbers))
print(next(numbers))
print(next(numbers))
````

# 📝 Explanation

The Generator does not create all values at once.

Each call to

```python
next()
```

produces exactly one value.

This is the main reason why Generators are preferred when processing large datasets.

# ♾️ Infinite Generators

One of the biggest advantages of Generators is that they can produce an unlimited sequence of values.

Since values are generated one at a time,

memory usage remains constant.

Infinite Generators are useful for:

- Random number generation
- Sensor data streams
- Live data feeds
- Continuous event processing

Example : 

```python
def infinite_numbers():

    number = 1

    while True:
        yield number
        number += 1

generator = infinite_numbers()

print(next(generator))
print(next(generator))
print(next(generator))
print(next(generator))
print(next(generator))
```

# ⚠️ Important Note

The following code should **never** be executed.

```python
for number in infinite_numbers():
    print(number)
```

The Generator never stops producing values.

Therefore,

the loop will continue indefinitely until it is manually interrupted.

# ⚠️ Common Mistakes

While learning Generators, beginners often make the following mistakes.

## ❌ Mistake 1: Using `return` Instead of `yield`

Incorrect

```python
def numbers():
    return 1
    return 2
```

**Correct**

```python
def numbers():
    yield 1
    yield 2
```

A Generator Function must use the `yield` keyword.

---

## ❌ Mistake 2: Expecting a Generator to Return a List

```python
def numbers():
    yield 1
    yield 2

result = numbers()

print(result)
```

Output

```text
<generator object numbers at 0x...>
```

A Generator returns a **Generator Object**, not a list.

To retrieve the values, use:

- `next()`
- `for` loop
- `list()`

---

## ❌ Mistake 3: Reusing an Exhausted Generator

```python
generator = (x for x in range(3))

for value in generator:
    print(value)

for value in generator:
    print(value)
```

The second loop produces no output because the Generator has already been consumed.

Create a new Generator if you want to iterate again.

---

## ❌ Mistake 4: Forgetting that `yield` Pauses Execution

Some beginners think that every call starts the function from the beginning.

In reality,

the Generator resumes execution from the previous `yield` statement.

---

## ❌ Mistake 5: Using a Generator When Multiple Iterations Are Required

Generators can be consumed only once.

If the data needs to be traversed multiple times,

consider using a list instead.
