# 🔒 Closures in Python

# 📖 Introduction

Functions are one of the most powerful features of Python.

In addition to defining and calling functions, Python also treats functions as **first-class objects**, which means they can be passed as arguments, returned from other functions, and stored in variables.

This capability makes advanced concepts such as **Closures** possible.

A **Closure** is created when an inner function remembers and accesses variables from its enclosing (outer) function, even after the outer function has finished executing.

Unlike ordinary nested functions, a closure preserves the state of the outer function, allowing the inner function to continue using those variables whenever it is called.

Closures are widely used in Functional Programming and serve as the foundation for many advanced Python features, including **Decorators**, callbacks, function factories, and stateful functions.

By the end of this notebook, you will understand how closures work, how they remember data, and why they are an essential concept in professional Python programming.

# 🤔 Why Do We Need Closures?

Sometimes we need a function to **remember certain information** even after another function has finished executing.

For example,

suppose we want to create multiple greeting functions.

Instead of writing separate functions like:

- `greet_english()`
- `greet_hindi()`
- `greet_japanese()`

we can write one function that creates customized greeting functions.

Closures make this possible.

They allow a function to retain access to variables from its enclosing scope, making the code more reusable, modular, and flexible.

Closures are especially useful for:

- 🔄 Creating customized functions.
- 🧠 Preserving state.
- ⚡ Reducing global variables.
- 🏭 Building function factories.
- 🎯 Preparing for Decorators.

# 🌍 Real-World Analogy

Imagine a teacher writes a student's marks inside a sealed envelope.

The teacher then gives the envelope to the student.

Even after the teacher leaves the classroom,

the student can still open the envelope later and read the marks.

The information remains safely stored inside the envelope.

Closures work in a similar way.

The **outer function** stores some data,

and the **inner function** carries that data with it,

even after the outer function has finished executing.

This ability to "remember" data is what makes Closures powerful.

# 📌 What is a Nested Function?

A **Nested Function** is a function that is defined inside another function.

The function defined inside is called the **Inner Function**, while the function containing it is called the **Outer Function**.

Nested functions help organize code, improve readability, and hide implementation details from the outside world.

By default, the inner function is accessible only within the outer function.



## 📌 Definition

> **A Nested Function is a function defined inside another function.**

````python
def outer():

    print("Outer Function")

    def inner():

        print("Inner Function")

    inner()


outer()
````

# 📝 Explanation

In this example,

- `outer()` is the **Outer Function**.
- `inner()` is the **Nested (Inner) Function**.

The inner function exists only inside the outer function.

It cannot be accessed directly from outside the outer function.

````python
def outer():

    def inner():

        print("Hello")

    inner()


outer()

# inner()   ❌ This will raise an error
````

# 📝 Explanation

The `inner()` function has **local scope** within the `outer()` function.

Trying to call:

```python
inner()
```

outside the outer function results in a `NameError` because the inner function is not visible outside its enclosing scope.

# 📌 What is a Closure?

A **Closure** is a special type of Nested Function.

A Nested Function becomes a **Closure** when it remembers and uses variables from its enclosing (outer) function **even after the outer function has finished executing**.

In other words,

a Closure "closes over" the variables from its outer scope and keeps them alive for future use.



## 📌 Definition

> **A Closure is a Nested Function that remembers the variables of its enclosing function, even after the enclosing function has finished execution.**

````python
def outer():

    message = "Hello Python"

    def inner():

        print(message)

    return inner


closure = outer()

closure()
````

# 📝 Explanation

Let's understand what happens step by step.

1. `outer()` is called.
2. The variable `message` is created.
3. The `inner()` function is defined.
4. Instead of calling `inner()`, the outer function **returns** it.
5. Normally, the local variable `message` should disappear after `outer()` finishes.
6. However, the returned `inner()` function still remembers `message`.

This is exactly what a **Closure** does.

It preserves the required variables from the enclosing scope even after that scope no longer exists.

# 🆓 Free Variables

A **Free Variable** is a variable that is **used inside the inner function but defined in the outer function**.

The inner function does not create this variable.

Instead, it accesses the variable from its enclosing scope.

In the previous example,

```python
message
```

is a **Free Variable** because:

- It is defined in `outer()`.
- It is used inside `inner()`.

````python
def outer():

    language = "Python"

    def inner():

        print(language)

    return inner


closure = outer()

closure()
````

# 📝 Explanation

In this example,

```python
language
```

is a **Free Variable**.

Even though `outer()` has finished executing,

the Closure still remembers its value and prints:

```text
Python
```

# 🌐 Lexical Scoping

Python follows a concept called **Lexical Scoping** (also known as **Static Scoping**).

This means that Python determines where a variable belongs based on **where the function is defined**, not where it is called.

When an inner function accesses a variable, Python searches in the following order:

1. Local Scope
2. Enclosing Scope
3. Global Scope
4. Built-in Scope

This search order is known as the **LEGB Rule**, which we'll study in detail in a later notebook.

Closures work because Python allows the inner function to access variables from the **Enclosing Scope**.

# 💡 Did You Know?

You can inspect the variables remembered by a Closure using the `__closure__` attribute.

Example:

```python
def outer():

    value = 100

    def inner():
        print(value)

    return inner


closure = outer()

print(closure.__closure__)
```

The `__closure__` attribute stores information about the free variables captured by the Closure.

This is an advanced feature that helps demonstrate how Closures preserve state internally.

# 🏭 Function Factory Using Closures

One of the most common uses of Closures is creating **Function Factories**.

A Function Factory is a function that creates and returns another function.

Each returned function remembers its own data from the enclosing function.

This allows us to create customized functions without duplicating code.

````python
def multiplier(factor):

    def multiply(number):

        return number * factor

    return multiply


double = multiplier(2)

triple = multiplier(3)

print(double(10))

print(triple(10))
````

# 📝 Explanation

Let's understand what happens.

When we execute:

```python
double = multiplier(2)
```

Python creates a Closure that remembers:

```python
factor = 2
```

Similarly,

```python
triple = multiplier(3)
```

creates another Closure that remembers:

```python
factor = 3
```

Although both functions are created from the same outer function,

each Closure preserves its own independent state.

# 🧠 How Closures Preserve State

One of the biggest advantages of Closures is that they remember data even after the outer function has finished executing.

Each Closure keeps its own copy of the captured variables.

This preserved information is called the **Closure State**.

````python
def counter():

    count = 0

    def increment():

        nonlocal count

        count += 1

        return count

    return increment


counter1 = counter()

print(counter1())

print(counter1())

print(counter1())
````

# 📝 Explanation

In this example,

the variable:

```python
count
```

belongs to the outer function.

The inner function modifies it using:

```python
nonlocal count
```

Even after the outer function has finished,

the Closure remembers the updated value of `count`.

This allows the counter to continue increasing every time the inner function is called.

````python
counter2 = counter()

print(counter2())

print(counter2())
````

# 📝 Explanation

Notice that:

```python
counter1
```

and

```python
counter2
```

maintain **independent states**.

Each Closure remembers its own value of `count`.

This demonstrates that every Closure has its own enclosed environment.

# 🔄 Nested Function vs Closure

Not every Nested Function is a Closure.

A Nested Function becomes a Closure only if it:

- Is returned from the outer function.
- Uses one or more variables from the enclosing scope.


# 📊 Nested Function vs Closure

| Nested Function | Closure |
|-----------------|----------|
| Defined inside another function. | A Nested Function that remembers outer variables. |
| May not use outer variables. | Uses free variables from the enclosing scope. |
| Usually executes inside the outer function. | Can execute even after the outer function finishes. |
| Does not necessarily preserve state. | Preserves state. |

````python
# Nested Function (Not a Closure)

def outer():

    def inner():

        print("Hello")

    inner()


outer()
````

````python
# Closure

def outer():

    message = "Hello Python"

    def inner():

        print(message)

    return inner


closure = outer()

closure()
````

# ⚠️ Common Mistakes

While learning Closures, beginners often make the following mistakes.

## ❌ Mistake 1: Confusing Nested Functions with Closures

Every Closure is a Nested Function.

However,

**not every Nested Function is a Closure.**

A Nested Function becomes a Closure only if it remembers variables from its enclosing scope.

---

## ❌ Mistake 2: Forgetting to Return the Inner Function

Incorrect

```python
def outer():

    message = "Hello"

    def inner():

        print(message)

    inner()
```

The function is executed immediately.

No Closure is created.

---

Correct

```python
def outer():

    message = "Hello"

    def inner():

        print(message)

    return inner
```

Returning the inner function creates a Closure.

---

## ❌ Mistake 3: Forgetting the `nonlocal` Keyword

Incorrect

```python
def counter():

    count = 0

    def increment():

        count += 1
```

This raises an error because Python treats `count` as a new local variable.

---

Correct

```python
def counter():

    count = 0

    def increment():

        nonlocal count

        count += 1
```

Use `nonlocal` when modifying variables from the enclosing scope.

---

## ❌ Mistake 4: Expecting the Outer Function to Stay Active

The outer function finishes execution.

Only the required variables are preserved by the Closure.

The entire outer function is **not** kept running.

---

## ❌ Mistake 5: Using Global Variables Instead of Closures

Many beginners solve state-related problems using global variables.

Closures provide a cleaner, safer, and more modular solution.

# ✅ Best Practices

Follow these best practices while working with Closures.

### 🔒 Use Closures to Preserve State

Closures are ideal when a function needs to remember information between function calls.

---

### 🏭 Use Closures for Function Factories

Closures make it easy to create customized functions without duplicating code.

---

### 🚫 Avoid Unnecessary Global Variables

Prefer Closures over global variables whenever possible.

This improves modularity and reduces side effects.

---

### 📝 Use `nonlocal` Carefully

Use the `nonlocal` keyword only when you need to modify variables from the enclosing scope.

---

### 🎯 Keep Closures Focused

A Closure should have a single, well-defined responsibility.

Avoid placing unrelated logic inside the same Closure.
