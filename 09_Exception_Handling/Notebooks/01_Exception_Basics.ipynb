# ❓ What is an Error?

An **error** is a problem in a program that prevents it from working correctly.

Errors may occur:

- While writing the program.
- While running the program.
- Because of invalid user input.
- Because of missing files or resources.

Whenever Python encounters an error, it generates an error message to help identify the problem.

## 💻 Example 1

````python
print("Hello World")
````

This program executes successfully because it contains no errors.

## 💻 Example 2

````python
print("Hello World"
````

Python cannot execute this program because the closing parenthesis is missing.

This type of error occurs before the program starts running.

# 🤔 Why Do Errors Occur?

Errors can occur for many reasons, such as:

- Typing mistakes
- Invalid syntax
- Incorrect calculations
- Missing files
- Invalid user input
- Dividing by zero
- Using undefined variables
- Accessing invalid indexes

# 📂 Types of Errors

Python errors are generally divided into two major categories:

1. Syntax Errors
2. Runtime Errors (Exceptions)

## 1️⃣ Syntax Errors

A **Syntax Error** occurs when the rules (syntax) of Python are violated.

The program cannot start until the error is fixed.

````python
if True
    print("Hello")
````

### Explanation

The colon (`:`) is missing after the `if` statement.

Python checks syntax before executing the program, so execution stops immediately.

## 2️⃣ Runtime Errors

A **Runtime Error** occurs while the program is executing.

The syntax is correct, but something unexpected happens during execution.

````python
number = 10

print(number / 0)
````

### Explanation

The program starts successfully because the syntax is correct.

However, it crashes when it attempts to divide a number by zero.

# ⚠️ What is an Exception?

An **exception** is a type of runtime error that occurs while a program is executing.

When an exception occurs, Python stops the normal flow of the program and displays an error message unless the exception is handled properly.

## 💡 Examples of Exceptions

Some common exceptions are:

- ZeroDivisionError
- ValueError
- TypeError
- NameError
- IndexError
- KeyError
- FileNotFoundError

## 💻 Example 3

````python
numbers = [10, 20, 30]

print(numbers[5])
````

### Explanation

The list contains only three elements.

Trying to access an element that doesn't exist raises an `IndexError`.

# 🔄 Exception Flow

When an exception occurs, Python follows this sequence:

1. Execute the program.
2. Encounter an exception.
3. Stop normal execution.
4. Display an error message.
5. Terminate the program (if the exception is not handled).

### Flow Diagram

```text
Program Starts
      │
      ▼
Execute Statements
      │
      ▼
Exception Occurs?
      │
 ┌────┴────┐
 │         │
No        Yes
 │         │
 ▼         ▼
Continue   Stop Execution
            │
            ▼
     Display Error
            │
            ▼
     Program Ends
```

# 🎯 Why Exception Handling?

Imagine you're using an application to transfer money, book a ticket, or upload a file.

If something goes wrong and the application suddenly crashes, it creates a poor user experience.

Exception handling allows programmers to detect and manage unexpected errors, making programs more reliable and user-friendly.

## 🤔 Why is Exception Handling Important?

Without exception handling:

- Programs may stop unexpectedly.
- Users may lose their work.
- Important tasks may remain incomplete.
- Applications become unreliable.

Exception handling helps developers write programs that can respond gracefully when unexpected situations occur.

# ✅ Advantages of Exception Handling

Exception handling provides several benefits:

- Prevents unexpected program termination.
- Improves program reliability.
- Makes debugging easier.
- Helps identify the cause of errors.
- Improves the user experience.
- Makes applications more robust and maintainable.

# 🌍 Real-World Examples

## 📌 Example 1: ATM Machine

Imagine an ATM machine.

If a customer enters an incorrect PIN, the ATM should display an error message instead of shutting down.

The machine continues to serve other customers safely.

## 📌 Example 2: Online Login System

When a user enters an incorrect password, the website displays a message such as:

> **"Invalid username or password."**

The application continues running instead of crashing.

## 📌 Example 3: File Reader

A program attempts to open a file.

If the file is missing, the application should inform the user instead of terminating unexpectedly.

This makes the software more reliable and easier to use.

# 💡 Key Points

- Errors are common in programming.
- Syntax errors occur before program execution.
- Runtime errors occur while the program is running.
- Exceptions are runtime errors.
- Unhandled exceptions stop program execution.
- Exception handling helps programs deal with unexpected situations gracefully.

# ❌ Common Mistakes

Beginners often make these mistakes:

- Confusing syntax errors with exceptions.
- Assuming every error is an exception.
- Ignoring error messages.
- Trying to fix errors without understanding their cause.
- Not reading the complete traceback provided by Python.

# ✅ Best Practices

Follow these best practices while learning exception handling:

- Read error messages carefully.
- Understand the root cause before fixing an error.
- Test programs with different inputs.
- Learn common built-in exceptions.
- Write clear and readable code.
