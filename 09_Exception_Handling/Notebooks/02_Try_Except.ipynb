# 🛡️ Try and Except in Python

## 📖 Introduction

In the previous notebook, you learned that an exception is a runtime error that can interrupt the normal execution of a program.

In this notebook, you'll learn how to use `try` and `except` to write more reliable Python programs.

# 🤔 Why Do We Need `try` and `except`?

Without exception handling, a single runtime error can stop the execution of an entire program.

Using `try` and `except`, we can detect exceptions and decide how the program should respond instead of allowing it to crash.

## 🌍 Real-World Example

Imagine a banking application.

A customer enters an amount to withdraw.

If the customer enters invalid data, the application should display an appropriate message instead of closing unexpectedly.

This is one of the most common uses of exception handling.

# 📦 What is `try`?

The `try` block contains the code that might generate an exception.

Python executes the statements inside the `try` block normally.

If no exception occurs, the program continues executing normally.

If an exception occurs, Python immediately stops executing the remaining statements inside the `try` block and looks for a matching `except` block.

# 📦 What is `except`?

The `except` block specifies what should happen if an exception occurs inside the corresponding `try` block.

If an exception is raised, Python skips the remaining statements in the `try` block and executes the matching `except` block.

# 💻 Example : Basic `try-except`

```python
try:
    number = 10 / 0
    print(number)

except:
    print("An error occurred.")
```

### Explanation

The statement `10 / 0` raises a `ZeroDivisionError`.

Instead of stopping the program and displaying the default traceback, Python executes the `except` block and prints a custom message.

# 💻 Example : Program Continues After Handling the Exception

```python
print("Program Started")

try:
    number = 10 / 0
    print(number)

except:
    print("Cannot divide by zero.")

print("Program Finished")
```

### Explanation

Although an exception occurs, the program does not terminate unexpectedly.

After executing the `except` block, Python continues executing the remaining statements in the program.

# 💡 Key Observation

When an exception occurs:

1. Python immediately stops executing the remaining statements inside the `try` block.
2. Python searches for a matching `except` block.
3. The `except` block is executed.
4. The program continues with the statements after the `try-except` block.

# 🎯 Catching Specific Exceptions

Instead of catching every exception using a general `except` block, it is a good practice to catch only the exceptions that you expect.

This makes your program more readable and easier to debug.

## 💻 Example : Catching a Specific Exception

```python
try:
    result = 10 / 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### Explanation

The statement raises a `ZeroDivisionError`.

Since the `except` block is specifically looking for a `ZeroDivisionError`, it handles the exception and displays the custom message.

# 💻 Example : Handling an IndexError

```python
numbers = [10, 20, 30]

try:
    print(numbers[5])

except IndexError:
    print("Index is out of range.")
```

### Explanation

The list contains only three elements.

Trying to access an element that does not exist raises an `IndexError`, which is handled by the matching `except` block.

# 📚 Multiple `except` Blocks

A single `try` block may raise different types of exceptions.

Python allows us to define multiple `except` blocks so that each exception can be handled differently.

## 💻 Example 

```python
numbers = [10, 20, 30]

try:
    print(numbers[10])

except ZeroDivisionError:
    print("Division by zero is not allowed.")

except IndexError:
    print("Invalid index.")

except ValueError:
    print("Invalid value.")
```

### Explanation

The program raises an `IndexError`.

Python checks the `except` blocks one by one and executes the matching `IndexError` block.

# 📌 Using `Exception as e`

Sometimes we want to know the exact error message generated by Python.

We can store the exception object in a variable using the syntax `Exception as e`.

The variable `e` contains details about the exception.

## 💻 Example 

```python
try:
    number = 10 / 0

except Exception as e:
    print("Error:", e)
```

### Explanation

Instead of displaying a custom message, the program prints the actual error message stored in the exception object.

## 💻 Example 

```python
try:
    numbers = [1, 2, 3]
    print(numbers[10])

except Exception as e:
    print("Error:", e)
```

### Explanation

The variable `e` contains the original error message generated by Python, making debugging much easier.

# 💡 Why Use `Exception as e`?

Using `Exception as e` provides several advantages:

- Displays the original error message.
- Makes debugging easier.
- Helps identify the exact cause of an error.
- Useful for logging errors in real-world applications.

# 🏗️ Nested `try-except`

A **nested `try-except`** means placing one `try-except` block inside another.

This is useful when different parts of a program can raise different exceptions, and you want to handle them separately.

## 💻 Example 

```python
try:
    print("Outer Try")

    try:
        result = 10 / 0
        print(result)

    except ZeroDivisionError:
        print("Inner: Cannot divide by zero.")

except Exception:
    print("Outer Exception")
```

### Explanation

The `ZeroDivisionError` occurs inside the inner `try` block.

Since the inner `except` block handles it successfully, the outer `except` block is never executed.

## 💻 Example 

```python
try:
    print("Outer Try")

    try:
        numbers = [10, 20]
        print(numbers[5])

    except ZeroDivisionError:
        print("Division Error")

except Exception as e:
    print("Outer:", e)
```

### Explanation

The inner `except` block only handles `ZeroDivisionError`.

Since an `IndexError` occurs, it is not handled inside the inner block.

Python passes the exception to the outer `except` block, where it is handled.

# ❌ Common Mistakes

Beginners often make these mistakes:

- Writing code that cannot raise an exception inside `try`.
- Catching every exception with a general `except` unnecessarily.
- Ignoring the original error message.
- Using too many nested `try-except` blocks, making code difficult to read.
- Assuming that every exception will be handled automatically.

# ✅ Best Practices

- Catch specific exceptions whenever possible.
- Keep the `try` block as small as possible.
- Read and understand error messages.
- Use meaningful error messages for users.
- Avoid unnecessary nested `try-except` blocks.
- Test your programs with different types of invalid inputs.
