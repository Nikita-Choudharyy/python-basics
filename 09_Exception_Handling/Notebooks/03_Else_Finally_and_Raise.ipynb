# ⚙️ else, finally, and raise in Python

## 📖 Introduction

In the previous notebook, you learned how to handle exceptions using `try` and `except`.

Python provides three additional keywords—`else`, `finally`, and `raise`—that make exception handling more flexible and powerful.

These keywords help you execute code only when no exception occurs, perform cleanup tasks, or intentionally raise exceptions when needed.

## 🎯 Learning Objectives

After completing this notebook, you will be able to:

- Understand the purpose of `else`.
- Understand the purpose of `finally`.

- Understand re-raising exceptions.
- Write cleaner and more reliable exception handling code.

# 📌 What is `else`?

The `else` block is used with a `try-except` statement.

It is executed **only if no exception occurs** inside the `try` block.

If an exception is raised, the `else` block is skipped.

# 💻 Example 

```python
try:
    number = 20 / 4
    print("Result:", number)

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Division completed successfully.")
```

### Explanation

No exception occurs while dividing `20` by `4`.

Therefore, Python skips the `except` block and executes the `else` block.

# 💻 Example 

```python
try:
    result = 20 / 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Division completed successfully.")
```

### Explanation

Since a `ZeroDivisionError` occurs, Python executes the `except` block.

The `else` block is skipped because the `try` block did not complete successfully.

# 💡 When Should You Use `else`?

Use the `else` block when you want to execute code **only if the `try` block succeeds**.

Common use cases include:

- Displaying success messages.
- Processing data after successful validation.
- Saving results after an operation completes without errors.

# 📌 What is `finally`?

The `finally` block is used to execute code that **must run regardless of whether an exception occurs or not**.

Whether the `try` block succeeds, an exception is handled, or an exception is not handled, the `finally` block is always executed.

# 💻 Example 

```python
try:
    number = 20 / 4
    print("Result:", number)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Execution completed.")
```

### Explanation

Since no exception occurs, the `try` block executes successfully.

After completing the `try` block, Python executes the `finally` block.

# 💻 Example 

```python
try:
    result = 20 / 0
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Execution completed.")
```

### Explanation

Even though a `ZeroDivisionError` occurs, the `finally` block still executes.

This makes `finally` ideal for cleanup operations that must always be performed.

# 🌍 Real-World Example

Suppose a program opens a file to read data.

Whether the file is read successfully or an error occurs while processing it, the file should always be closed.

The `finally` block ensures that cleanup operations are performed reliably.

# 🔍 Difference Between `else` and `finally`

| `else` | `finally` |
|---------|-----------|
| Executes only when no exception occurs | Executes whether an exception occurs or not |
| Used for success-related code | Used for cleanup-related code |
| Optional | Optional |
| Skipped if an exception occurs | Always executes |

# 💻 Example : Using `else` and `finally` Together

```python
try:
    result = 50 / 10
    print("Result:", result)

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Division was successful.")

finally:
    print("Program execution completed.")
```

### Explanation

Since no exception occurs:

- The `try` block executes successfully.
- The `except` block is skipped.
- The `else` block executes because the operation was successful.
- The `finally` block executes at the end, regardless of the outcome.

# 💡 Key Observation

Execution order when no exception occurs:

1. `try`
2. `else`
3. `finally`

Execution order when an exception occurs:

1. `try`
2. `except`
3. `finally`

# 📌 What is `raise`?

The `raise` keyword is used to **manually generate an exception**.

Normally, Python raises exceptions automatically when an error occurs.

However, there are situations where you may want to stop the program or indicate that invalid data has been provided. In such cases, you can use the `raise` statement.

## 🤔 Why Do We Use `raise`?

The `raise` statement is useful when:

- User input is invalid.
- Business rules are violated.
- Required conditions are not met.
- You want to prevent incorrect data from being processed.

The `raise` keyword creates an exception manually.

You can choose the exception type and provide a meaningful error message.

# 💻 Example 

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
```

### Explanation

The age is invalid because it is less than zero.

Instead of allowing the program to continue with incorrect data, we manually raise a `ValueError`.

# 💻 Example 

```python
marks = 120

if marks > 100:
    raise ValueError("Marks cannot be greater than 100.")
```

### Explanation

The maximum valid marks are 100.

Since the value is outside the valid range, the program raises a `ValueError`.

# 📌 Using `raise` Inside `try`

```python
try:
    age = -10

    if age < 0:
        raise ValueError("Age cannot be negative.")

except ValueError as e:
    print(e)
```

### Explanation

The exception is raised manually using `raise`.

Since it occurs inside the `try` block, it is immediately caught by the matching `except` block.

# 🔄 Re-raising Exceptions

Sometimes, you may want to perform some action when an exception occurs and then allow the exception to continue.

This is called **re-raising an exception**.

## 💻 Example 

```python
try:
    result = 10 / 0

except ZeroDivisionError:
    print("Division by zero occurred.")
    raise
```

### Explanation

The exception is first caught by the `except` block.

After printing a message, `raise` is used without specifying an exception, so Python raises the same exception again.

# 🌍 Real-World Examples

### 📌 Example 1: Age Validation

A registration form should reject negative ages by raising a `ValueError`.

### 📌 Example 2: Exam Marks

A school management system should reject marks less than 0 or greater than 100.

### 📌 Example 3: Product Price

An online shopping application should not allow products with negative prices.

### 📌 Example 4: Bank Withdrawal

A banking application can raise an exception if the withdrawal amount is invalid or exceeds the available balance.

# 📌 Quick Reference Table

| Concept | Description |
|---------|-------------|
| `else` | Executes only if the `try` block completes without an exception |
| `finally` | Always executes, whether an exception occurs or not |
| `raise` | Manually raises an exception |
| `raise ExceptionType("message")` | Raises a specific exception with a custom message |
| `raise` | Re-raises the current exception inside an `except` block |

# 💡 Key Points

- `else` executes only when no exception occurs.
- `finally` always executes.
- `raise` creates an exception manually.
- `raise` can be used inside a `try` block.
- `raise` without an exception re-raises the current exception.

# ❌ Common Mistakes

Beginners often make these mistakes:

- Using `raise` without a valid reason.
- Raising the wrong exception type.
- Forgetting to provide a meaningful error message.
- Assuming `finally` executes only when an exception occurs.
- Confusing `else` with `finally`.

# ✅ Best Practices

- Raise exceptions only for truly invalid situations.
- Choose the most appropriate exception type.
- Write clear and descriptive error messages.
- Use `finally` for cleanup tasks.
- Keep `try` blocks focused on code that may raise exceptions.
