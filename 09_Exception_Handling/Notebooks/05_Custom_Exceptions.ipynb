# 📚 Custom Exceptions in Python

## 📖 Introduction

Python provides many built-in exceptions such as `ValueError`, `TypeError`, and `IndexError`.

However, in real-world applications, these built-in exceptions are sometimes not enough to describe business-specific problems.

In such situations, Python allows us to create our own exceptions, known as **Custom Exceptions**.

Custom exceptions make programs more meaningful, easier to debug, and simpler to maintain.

# 🤔 What are Custom Exceptions?

A **Custom Exception** is a user-defined exception created by the programmer.

Instead of relying only on Python's built-in exceptions, you can define exceptions that describe errors specific to your application's requirements.

Custom exceptions improve code readability and make error messages more meaningful.

## 📌 Simple Example

Imagine you are creating a banking application.

If a user tries to withdraw more money than their account balance, a built-in exception like `ValueError` does not clearly describe the problem.

A custom exception named `InsufficientBalanceError` provides a much more meaningful description of the issue.

# 💡 Why Do We Need Custom Exceptions?

Custom exceptions help us:

- Represent application-specific errors.
- Improve code readability.
- Make debugging easier.
- Provide meaningful error messages.
- Build professional and maintainable applications.

# ⚖️ Built-in Exception vs Custom Exception

| Built-in Exception | Custom Exception |
|--------------------|------------------|
| Provided by Python | Created by the programmer |
| General-purpose | Application-specific |
| Examples: `ValueError`, `TypeError` | Examples: `InvalidAgeError`, `InsufficientBalanceError` |
| Suitable for common runtime errors | Suitable for business logic and project-specific rules |

# 🧬 Inheriting from the Exception Class

Every custom exception should inherit from Python's built-in `Exception` class.

By inheriting from `Exception`, your custom class behaves like any other Python exception and can be raised or handled using `try` and `except`.

## 📌 Basic Syntax

```python
class CustomError(Exception):
    pass
```

### Explanation

In this example:

- `CustomError` is a user-defined exception.
- It inherits from the `Exception` class.
- The `pass` statement indicates that the class has no additional functionality yet.

Even without additional code, it behaves like a standard Python exception.

# 🚀 Creating Your First Custom Exception

Now let's create and use our first custom exception.

```python
class InvalidAgeError(Exception):
    pass
```

### Explanation

Here, we created a new exception class named `InvalidAgeError`.

This exception can now be raised whenever an invalid age is encountered in our program.

# 📌 Why Inherit from Exception?

Python recognizes classes that inherit from `Exception` as valid exceptions.

If a class does not inherit from `Exception` (or one of its subclasses), Python will not allow it to be raised as an exception.

# 🚀 Raising a Custom Exception

## 📖 What is `raise`?

The `raise` keyword is used to manually generate an exception.

When Python executes a `raise` statement, it immediately stops the normal flow of the program and raises the specified exception.

## 📌 Syntax

 - raise ExceptionName

### Explanation

Replace `ExceptionName` with the exception you want to raise.

It can be either:

- A built-in exception
- A custom exception

## 💻 Example : Raising a Custom Exception

```python
class InvalidAgeError(Exception):
    pass

raise InvalidAgeError
```

### Explanation

Here, Python creates an object of the `InvalidAgeError` class and raises it.

Since the exception is not handled, the program stops immediately.

# 🛡️ Handling a Custom Exception

Custom exceptions can be handled using the same `try` and `except` blocks that are used for built-in exceptions.

```python
class InvalidAgeError(Exception):
    pass

try:
    raise InvalidAgeError

except InvalidAgeError:
    print("Invalid age entered.")
```

### Explanation

The exception is raised inside the `try` block.

Instead of crashing the program, the `except` block catches the custom exception and prints an appropriate message.

# 💬 Custom Exception with an Error Message

Sometimes we want our exception to display a meaningful message.

This can be done by passing a message while raising the exception.

```python
class InvalidAgeError(Exception):
    pass

raise InvalidAgeError("Age must be 18 or above.")
```

### Explanation

The message `"Age must be 18 or above."` becomes part of the exception.

It helps users and developers understand the exact reason for the error.

# 🎯 Handling a Custom Exception with a Message

```python
class InvalidAgeError(Exception):
    pass

try:
    raise InvalidAgeError("Age must be 18 or above.")

except InvalidAgeError as error:
    print(error)
```

### Explanation

The exception object is stored in the variable `error`.

Printing this object displays the custom error message passed while raising the exception.

# 🌍 Real-World Example

Imagine an online voting system.

Only users who are 18 years or older are allowed to vote.

If someone enters an age below 18, raising a custom exception such as `InvalidAgeError` makes the program's intent much clearer than using a generic exception.

```python
class InvalidAgeError(Exception):
    pass

age = 16

if age < 18:
    raise InvalidAgeError("You must be at least 18 years old to vote.")
```

# 🛠️ Custom Exception with `__init__()`

## 📖 Why Use `__init__()`?

Sometimes we want our custom exception to store additional information, such as an error message, error code, or any other relevant data.

For this purpose, we can define an `__init__()` method inside our custom exception class.

## 📌 Syntax

```python
class CustomError(Exception):

    def __init__(self, message):
        super().__init__(message)
```

### Explanation

In this example:

- `__init__()` initializes the exception object.
- `message` stores the custom error message.
- `super().__init__(message)` passes the message to the parent `Exception` class.

## 💻 Example

```python
class InvalidAgeError(Exception):

    def __init__(self, message):
        super().__init__(message)

raise InvalidAgeError("Age must be 18 or above.")
```

### Explanation

The message is passed to the parent `Exception` class using `super()`.

When the exception is raised, Python automatically displays the custom message.

# 🚀 Creating Multiple Custom Exceptions

Large applications often require multiple custom exceptions for different situations.

Each exception represents a different type of business rule or application error.

```python
class InvalidAgeError(Exception):
    pass


class InsufficientBalanceError(Exception):
    pass


class InvalidPasswordError(Exception):
    pass
```

### Explanation

Here we created three different custom exceptions.

Each exception represents a completely different error condition.

# 🌍 Real-World Example 1: Banking System

```python
class InsufficientBalanceError(Exception):
    pass


balance = 500

withdraw_amount = 800

if withdraw_amount > balance:
    raise InsufficientBalanceError("Insufficient account balance.")
```

### Explanation

Instead of raising a generic exception, we use a meaningful exception named `InsufficientBalanceError`.

This clearly explains the actual business problem.

# 🌍 Real-World Example 2: Login System

```python
class InvalidPasswordError(Exception):
    pass


password = "12345"

if len(password) < 8:
    raise InvalidPasswordError("Password must contain at least 8 characters.")
```

### Explanation

The program checks whether the password satisfies the minimum length requirement.

If it does not, an `InvalidPasswordError` is raised with a meaningful message.

# 🌍 Real-World Example 3: Student Admission System

```python
class InvalidMarksError(Exception):
    pass


marks = 105

if marks < 0 or marks > 100:
    raise InvalidMarksError("Marks must be between 0 and 100.")
```

### Explanation

The application validates the student's marks.

If the marks are outside the valid range, the custom exception is raised.

# 🌳 Custom Exception Hierarchy

Just like Python's built-in exceptions, custom exceptions can also be organized into a hierarchy.

Creating a base custom exception helps group related exceptions together and makes error handling more structured.

```python
class BankingError(Exception):
    pass


class InsufficientBalanceError(BankingError):
    pass


class InvalidAccountError(BankingError):
    pass
```

### Explanation

In this example:

- `BankingError` is the base custom exception.
- `InsufficientBalanceError` and `InvalidAccountError` inherit from `BankingError`.
- This hierarchy allows related exceptions to be grouped together, making exception handling cleaner in larger applications.

# 💡 When Should You Create a Custom Exception?

Create a custom exception when:

- A built-in exception does not clearly describe the problem.
- Your application has specific business rules.
- You want meaningful error messages.
- You want to organize related application errors.
- You are building medium or large-scale applications.

- Custom exceptions are user-defined exceptions created for application-specific errors.
- Every custom exception should inherit from Python's `Exception` class.
- The `raise` keyword is used to manually generate custom exceptions.
- Custom exceptions can include meaningful error messages.

# ❌ Common Mistakes

Beginners often make these mistakes:

- Forgetting to inherit from the `Exception` class.
- Using a built-in exception when a custom exception would better describe the problem.
- Writing vague or unclear exception messages.
- Creating too many unnecessary custom exceptions.
- Raising a custom exception without handling it when appropriate.
- Forgetting to call `super().__init__(message)` when overriding the `__init__()` method.

# ✅ Best Practices

Follow these best practices when creating custom exceptions:

- Inherit from the `Exception` class.
- Choose meaningful and descriptive exception names.
- Write clear and informative error messages.
- Create custom exceptions only when built-in exceptions are not sufficient.
- Group related exceptions using a base custom exception class.
- Keep exception classes simple and focused on representing specific errors.
